```python
# NLP + K-MEANS CLUSTERING

import pandas as pd
import re
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD

# 1. Load dataset
df = pd.read_csv("dataset_praktikum2.csv")
print(df.head())
print(df.columns)

# 2. Select text column
text_column = "text"   # change if needed
df[text_column] = df[text_column].fillna("").astype(str)

# 3. NLP preprocessing
def clean_text(x):
    x = x.lower()
    x = re.sub(r"http\S+|www\S+", "", x)
    x = re.sub(r"[^a-zA-Z\s]", "", x)
    return x

df["clean_text"] = df[text_column].apply(clean_text)

# 4. TF-IDF
vectorizer = TfidfVectorizer(stop_words="english", max_features=5000)
X = vectorizer.fit_transform(df["clean_text"])

# 5. K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["cluster"] = model.fit_predict(X)

# 6. Show clusters
for i in range(k):
    print(f"\nCLUSTER {i}")
    print(df[df["cluster"] == i][text_column].head(5).to_string(index=False))

# 7. Top words in each cluster
words = vectorizer.get_feature_names_out()
for i, center in enumerate(model.cluster_centers_):
    top = center.argsort()[-10:][::-1]
    print(f"\nCluster {i} words:", ", ".join(words[top]))

# 8. Visualization
X2 = TruncatedSVD(n_components=2, random_state=42).fit_transform(X)

plt.figure(figsize=(8,5))
plt.scatter(X2[:,0], X2[:,1], c=df["cluster"], cmap="viridis")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.title("K-Means NLP Clustering")
plt.show()

# 9. Save result
df.to_csv("dataset_praktikum2_kmeans_result.csv", index=False)
print("\nDone! Result saved.")
```
