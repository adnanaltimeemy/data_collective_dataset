```python
# 1. Import libraries
import pandas as pd
import re
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 2. Load dataset
df = pd.read_csv("aldi_finds_latest.csv")
print(df.head())
print(df.columns)

# 3. Select text column
# Change 'product_name' to your actual text column
text_col = "product_name"

# 4. Clean text
def clean_text(x):
    x = str(x).lower()
    x = re.sub(r'[^a-z\s]', '', x)
    return x

df["clean_text"] = df[text_col].fillna("").apply(clean_text)

# 5. NLP: TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(df["clean_text"])

# 6. K-Means clustering
k = 5
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X)

# 7. Show clusters
print("\nCluster results:")
print(df[[text_col, "cluster"]].head(20))

# 8. PCA visualisation
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X.toarray())

plt.figure(figsize=(8,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=df["cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Aldi Finds - K-Means Clusters")
plt.show()

# 9. Save results
df.to_csv("aldi_finds_kmeans_results.csv", index=False)
print("Done!")
```
