```python
# 1. Import libraries
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# 2. Load dataset
df = pd.read_csv("date-started-by-county.csv")

# 3. Explore data
print(df.head())
print(df.info())

# 4. Clean text/date columns
df = df.dropna()

# Example NLP processing
text_data = df["text_column"].astype(str)

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

X_text = tfidf.fit_transform(text_data)

# 5. K-means clustering
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

df["cluster"] = kmeans.fit_predict(X_text)

# 6. Reduce dimensions for visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_text.toarray())

plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    c=df["cluster"]
)

plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("K-means NLP Clusters")
plt.show()

# 7. View clustered counties
df[["county", "cluster"]].head()
```
