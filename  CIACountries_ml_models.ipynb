```python
df = df.convert_dtypes()

cat_cols = df.select_dtypes(include=["string"]).columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
label_encoders = {}

for col in cat_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le
```
