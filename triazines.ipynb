```python
# NLP + K-Means on triazines.csv

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# Load dataset
df = pd.read_csv("triazines.csv")
print(df.head())
print(df.columns)

# Select text column (change "text" if needed)
text_col = "text"

# TF-IDF
tfidf = TfidfVectorizer(stop_words="english", max_features=2000)
X = tfidf.fit_transform(df[text_col].fillna(""))

# K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# Show results
print(df[[text_col, "Cluster"]].head(10))

# PCA visualization
X2 = PCA(n_components=2).fit_transform(X.toarray())

plt.scatter(X2[:, 0], X2[:, 1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Triazines - K-Means Clustering")
plt.show()

# Save results
df.to_csv("triazines_clustered.csv", index=False)
```
