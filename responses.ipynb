```python
# Install if needed:
# !pip install pandas scikit-learn nltk matplotlib

import pandas as pd
import re
import nltk
import matplotlib.pyplot as plt

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

nltk.download("stopwords")

# 1. Load dataset
df = pd.read_csv("responses.csv")

# Change "response" to your text column name
text_col = "response"

# 2. NLP cleaning
stop = set(stopwords.words("english"))

def clean(text):
    text = re.sub(r"[^a-zA-Z\s]", "", str(text).lower())
    return " ".join(w for w in text.split() if w not in stop)

df["clean"] = df[text_col].apply(clean)

# 3. TF-IDF
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(df["clean"])

# 4. K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["cluster"] = model.fit_predict(X)

# 5. Show results
print(df[[text_col, "cluster"]].head(10))

# 6. Top words in each cluster
words = tfidf.get_feature_names_out()

for i in range(k):
    top = model.cluster_centers_[i].argsort()[-10:][::-1]
    print(f"\nCluster {i}:")
    print(", ".join(words[j] for j in top))

# 7. Visualisation
pca = PCA(n_components=2)
points = pca.fit_transform(X.toarray())

plt.scatter(points[:, 0], points[:, 1], c=df["cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("K-Means NLP Clusters")
plt.show()

# 8. Save results
df.to_csv("responses_clustered.csv", index=False)
```
