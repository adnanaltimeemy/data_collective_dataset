```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD

# 2. Load dataset
df = pd.read_csv("cast_movie_count.csv")

# Check columns
print(df.head())
print(df.columns)

# 3. NLP: convert cast text to TF-IDF
df["cast"] = df["cast"].fillna("").astype(str)

tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(df["cast"])

# 4. K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# 5. Display clusters
print(df[["cast", "Cluster"]].head(20))

# 6. Reduce dimensions for visualization
svd = TruncatedSVD(n_components=2, random_state=42)
X_2D = svd.fit_transform(X)

# 7. Plot clusters
plt.scatter(X_2D[:,0], X_2D[:,1], c=df["Cluster"])
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.title("K-Means Clustering of Cast Data")
plt.show()

# 8. Save result
df.to_csv("cast_movie_count_clustered.csv", index=False)
```
