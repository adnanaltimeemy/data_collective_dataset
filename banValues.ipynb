````python
```python
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Load dataset
df = pd.read_csv("banValues.csv")

# Automatically select text column
text_col = df.select_dtypes(include="object").columns[0]

# Clean text
df["clean"] = df[text_col].fillna("").astype(str).str.lower()
df["clean"] = df["clean"].apply(lambda x: re.sub(r"[^a-zA-Z\s]", "", x))

# NLP - TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(df["clean"])

# K-Means
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X)

# Show results
print(df[[text_col, "cluster"]].head(10))
print("\nCluster counts:")
print(df["cluster"].value_counts())

# Save results
df.to_csv("banValues_clustered.csv", index=False)
print("\nSaved as banValues_clustered.csv")
```
````
