```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("iris-id.csv")

# Show data
print(df.head())

# Select numerical columns and remove ID if present
X = df.select_dtypes(include="number")
X = X.drop(columns=[c for c in X.columns if "id" in c.lower()], errors="ignore")

# Standardise data
X = StandardScaler().fit_transform(X)

# K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# Display clusters
print(df.head())
print("\nCluster counts:")
print(df["Cluster"].value_counts())

# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=df["Cluster"])
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Iris K-Means Clustering")
plt.show()
```
