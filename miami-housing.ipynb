```python
# ============================================================
# Miami Housing Dataset: K-Means + NLP
# Jupyter Notebook
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

import warnings
warnings.filterwarnings("ignore")


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv("miami-housing.csv")

print("Dataset shape:", df.shape)
display(df.head())
print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
display(df.isnull().sum())


# ============================================================
# 3. BASIC DATA EXPLORATION
# ============================================================

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").T)


# ============================================================
# 4. PREPARE NUMERIC FEATURES FOR K-MEANS
# ============================================================

# Select numeric columns
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

print("Numeric columns:")
print(numeric_cols)

# Remove columns that are likely to be IDs
id_cols = [
    col for col in numeric_cols
    if "id" in col.lower()
]

numeric_features = [
    col for col in numeric_cols
    if col not in id_cols
]

print("\nFeatures used for K-Means:")
print(numeric_features)

# Create a copy
X = df[numeric_features].copy()

# Replace infinite values
X = X.replace([np.inf, -np.inf], np.nan)

# Fill missing values with median
X = X.fillna(X.median())

# Standardize data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\nScaled data shape:", X_scaled.shape)


# ============================================================
# 5. FIND OPTIMAL NUMBER OF CLUSTERS - ELBOW METHOD
# ============================================================

inertia = []
K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")
plt.grid(True)
plt.show()


# ============================================================
# 6. SILHOUETTE SCORE
# ============================================================

silhouette_scores = []

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

plt.figure(figsize=(8, 5))
plt.plot(K_range, silhouette_scores, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score")
plt.grid(True)
plt.show()

# Select K with highest silhouette score
best_k = K_range[np.argmax(silhouette_scores)]

print("Recommended number of clusters:", best_k)


# ============================================================
# 7. APPLY K-MEANS
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(X_scaled)

print("\nCluster counts:")
print(df["Cluster"].value_counts().sort_index())


# ============================================================
# 8. VISUALIZE K-MEANS CLUSTERS USING PCA
# ============================================================

pca = PCA(n_components=2, random_state=42)

X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 7))

sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=df["Cluster"],
    palette="viridis",
    s=60
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Miami Housing K-Means Clusters")
plt.legend(title="Cluster")
plt.show()


# ============================================================
# 9. ANALYZE EACH CLUSTER
# ============================================================

cluster_summary = df.groupby("Cluster")[numeric_features].mean()

print("Average values by cluster:")
display(cluster_summary)


# ============================================================
# 10. NLP SECTION
# ============================================================

# Find text columns
text_cols = df.select_dtypes(include=["object"]).columns.tolist()

print("Text columns:")
print(text_cols)


# Combine all text columns into one document per house
if len(text_cols) > 0:

    text_data = df[text_cols].fillna("").astype(str)

    df["combined_text"] = text_data.apply(
        lambda row: " ".join(row),
        axis=1
    )

    print("\nExample text:")
    print(df["combined_text"].head())


    # ========================================================
    # 11. TF-IDF
    # ========================================================

    tfidf = TfidfVectorizer(
        stop_words="english",
        max_features=1000
    )

    X_text = tfidf.fit_transform(df["combined_text"])

    print("\nTF-IDF matrix shape:")
    print(X_text.shape)


    # ========================================================
    # 12. DISPLAY MOST IMPORTANT WORDS
    # ========================================================

    feature_names = tfidf.get_feature_names_out()

    # Average TF-IDF value for every word
    mean_tfidf = np.asarray(
        X_text.mean(axis=0)
    ).flatten()

    word_scores = pd.DataFrame({
        "word": feature_names,
        "score": mean_tfidf
    })

    word_scores = word_scores.sort_values(
        "score",
        ascending=False
    )

    print("\nTop 20 words:")
    display(word_scores.head(20))


    # ========================================================
    # 13. K-MEANS ON TEXT DATA
    # ========================================================

    # Determine a reasonable number of NLP clusters
    n_text_clusters = best_k

    text_kmeans = KMeans(
        n_clusters=n_text_clusters,
        random_state=42,
        n_init=10
    )

    df["Text_Cluster"] = text_kmeans.fit_predict(X_text)

    print("\nNLP cluster counts:")
    print(df["Text_Cluster"].value_counts().sort_index())


    # ========================================================
    # 14. FIND IMPORTANT WORDS FOR EACH TEXT CLUSTER
    # ========================================================

    print("\nImportant words in each NLP cluster:")

    for cluster in range(n_text_clusters):

        cluster_rows = df["Text_Cluster"] == cluster

        cluster_tfidf = X_text[cluster_rows].mean(axis=0)
        cluster_tfidf = np.asarray(cluster_tfidf).flatten()

        top_indices = cluster_tfidf.argsort()[-10:][::-1]

        top_words = feature_names[top_indices]

        print(f"\nCluster {cluster}:")
        print(", ".join(top_words))


    # ========================================================
    # 15. COMBINE NUMERIC + NLP CLUSTERS
    # ========================================================

    print("\nComparison of numerical and text clusters:")

    cluster_comparison = pd.crosstab(
        df["Cluster"],
        df["Text_Cluster"]
    )

    display(cluster_comparison)


# ============================================================
# 16. SAVE RESULTS
# ============================================================

df.to_csv(
    "miami_housing_clustered.csv",
    index=False
)

print("\nSaved file: miami_housing_clustered.csv")


# ============================================================
# 17. FINAL DATASET
# ============================================================

display(df.head())
```
