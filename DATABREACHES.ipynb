```python
# ============================================================
# DATA BREACHES DATASET
# NLP + K-MEANS CLUSTERING
# COMPLETE JUPYTER NOTEBOOK CODE
# ============================================================


# ============================================================
# CELL 1 - INSTALL REQUIRED LIBRARIES
# ============================================================

# Run this cell if the libraries are not already installed.

# !pip install pandas numpy matplotlib seaborn scikit-learn nltk wordcloud


# ============================================================
# CELL 2 - IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import re
import string
import warnings

from collections import Counter

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

warnings.filterwarnings("ignore")


# ============================================================
# CELL 3 - DOWNLOAD NLTK RESOURCES
# ============================================================

nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")


# ============================================================
# CELL 4 - LOAD DATASET
# ============================================================

file_path = "Data Breaches.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print()
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


# ============================================================
# CELL 5 - DISPLAY FIRST RECORDS
# ============================================================

display(df.head())


# ============================================================
# CELL 6 - DISPLAY LAST RECORDS
# ============================================================

display(df.tail())


# ============================================================
# CELL 7 - DISPLAY COLUMN NAMES
# ============================================================

print("Columns in the dataset:")
print()

for i, column in enumerate(df.columns):
    print(i, ":", column)


# ============================================================
# CELL 8 - DATASET INFORMATION
# ============================================================

print("Dataset information:")
print()

df.info()


# ============================================================
# CELL 9 - STATISTICAL SUMMARY
# ============================================================

display(df.describe(include="all").T)


# ============================================================
# CELL 10 - CHECK MISSING VALUES
# ============================================================

missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values,
    "Missing Percentage": (
        missing_values.values / len(df) * 100
    )
})

missing_table = missing_table.sort_values(
    "Missing Values",
    ascending=False
)

display(missing_table)


# ============================================================
# CELL 11 - CHECK DUPLICATE RECORDS
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)


# ============================================================
# CELL 12 - REMOVE DUPLICATES
# ============================================================

df = df.drop_duplicates().reset_index(drop=True)

print("Dataset shape after removing duplicates:")
print(df.shape)


# ============================================================
# CELL 13 - IDENTIFY DATA TYPES
# ============================================================

print("Data types:")
display(
    pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str).values
    })
)


# ============================================================
# CELL 14 - FIND TEXT COLUMNS
# ============================================================

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Text columns:")
print(text_columns)


# ============================================================
# CELL 15 - FIND NUMERICAL COLUMNS
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical columns:")
print(numeric_columns)


# ============================================================
# CELL 16 - PREPARE TEXT DATA
# ============================================================

if len(text_columns) == 0:
    raise ValueError(
        "The dataset does not contain any text columns."
    )

# Replace missing text values
df[text_columns] = df[text_columns].fillna("")


# ============================================================
# CELL 17 - COMBINE TEXT COLUMNS
# ============================================================

df["combined_text"] = (
    df[text_columns]
    .astype(str)
    .agg(" ".join, axis=1)
)

display(
    df[text_columns + ["combined_text"]].head()
)


# ============================================================
# CELL 18 - NLP TEXT CLEANING FUNCTION
# ============================================================

stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()


def clean_text(text):

    # Convert to string
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )

    # Remove email addresses
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # Remove HTML
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )

    # Remove numbers
    text = re.sub(
        r"\d+",
        " ",
        text
    )

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove non-alphabetic characters
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    # Tokenization
    words = text.split()

    # Stopword removal
    words = [
        word for word in words
        if word not in stop_words
    ]

    # Lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]

    return " ".join(words)


# ============================================================
# CELL 19 - APPLY TEXT CLEANING
# ============================================================

df["clean_text"] = df["combined_text"].apply(
    clean_text
)

display(
    df[["combined_text", "clean_text"]].head(10)
)


# ============================================================
# CELL 20 - REMOVE EMPTY TEXT RECORDS
# ============================================================

before = len(df)

df = df[
    df["clean_text"].str.strip() != ""
].copy()

df = df.reset_index(drop=True)

after = len(df)

print("Records before removing empty text:", before)
print("Records after removing empty text:", after)
print("Records removed:", before - after)


# ============================================================
# CELL 21 - TEXT LENGTH ANALYSIS
# ============================================================

df["text_length"] = df["clean_text"].apply(
    lambda x: len(x.split())
)

display(
    df["text_length"].describe()
)


# ============================================================
# CELL 22 - TEXT LENGTH HISTOGRAM
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    df["text_length"],
    bins=30
)

plt.xlabel("Number of Words")
plt.ylabel("Number of Records")
plt.title("Distribution of Text Length")

plt.show()


# ============================================================
# CELL 23 - CREATE TF-IDF FEATURES
# ============================================================

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    min_df=2,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X = vectorizer.fit_transform(
    df["clean_text"]
)

print("TF-IDF matrix:")
print("Rows:", X.shape[0])
print("Features:", X.shape[1])


# ============================================================
# CELL 24 - DISPLAY TF-IDF FEATURES
# ============================================================

feature_names = vectorizer.get_feature_names_out()

print("Number of TF-IDF features:")
print(len(feature_names))

print()
print("First 50 features:")

print(feature_names[:50])


# ============================================================
# CELL 25 - FIND MOST IMPORTANT WORDS
# ============================================================

mean_tfidf = np.asarray(
    X.mean(axis=0)
).flatten()

tfidf_words = pd.DataFrame({
    "Word": feature_names,
    "TF_IDF": mean_tfidf
})

tfidf_words = tfidf_words.sort_values(
    "TF_IDF",
    ascending=False
)

print("Top 30 words according to TF-IDF:")

display(
    tfidf_words.head(30)
)


# ============================================================
# CELL 26 - VISUALIZE TOP WORDS
# ============================================================

top_words = tfidf_words.head(20)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_words,
    x="TF_IDF",
    y="Word"
)

plt.title("Top 20 Words by Average TF-IDF")
plt.xlabel("Average TF-IDF")
plt.ylabel("Word")

plt.tight_layout()

plt.show()


# ============================================================
# CELL 27 - ELBOW METHOD
# ============================================================

inertia = []

k_range = range(2, 11)

for k in k_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)

    inertia.append(
        model.inertia_
    )


# ============================================================
# CELL 28 - PLOT ELBOW CURVE
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_range),
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Selecting K")

plt.xticks(list(k_range))
plt.grid(True)

plt.show()


# ============================================================
# CELL 29 - SILHOUETTE ANALYSIS
# ============================================================

silhouette_scores = []

for k in k_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k} | "
        f"Silhouette Score = {score:.4f}"
    )


# ============================================================
# CELL 30 - PLOT SILHOUETTE SCORES
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_range),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.xticks(list(k_range))
plt.grid(True)

plt.show()


# ============================================================
# CELL 31 - SELECT BEST K
# ============================================================

best_k = list(k_range)[
    np.argmax(silhouette_scores)
]

print(
    "Best K according to silhouette score:",
    best_k
)


# ============================================================
# CELL 32 - CREATE FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X)

df["Cluster"] = cluster_labels

print("K-Means clustering completed.")


# ============================================================
# CELL 33 - CLUSTER COUNTS
# ============================================================

cluster_counts = (
    df["Cluster"]
    .value_counts()
    .sort_index()
)

cluster_table = pd.DataFrame({
    "Cluster": cluster_counts.index,
    "Number of Breaches": cluster_counts.values,
    "Percentage": (
        cluster_counts.values /
        len(df) * 100
    )
})

display(cluster_table)


# ============================================================
# CELL 34 - CLUSTER DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=cluster_table,
    x="Cluster",
    y="Number of Breaches"
)

plt.title("Number of Data Breaches per Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Breaches")

plt.tight_layout()

plt.show()


# ============================================================
# CELL 35 - IMPORTANT WORDS FOR EACH CLUSTER
# ============================================================

terms = vectorizer.get_feature_names_out()

cluster_keywords = {}

for cluster_number in range(best_k):

    cluster_center = (
        kmeans.cluster_centers_[cluster_number]
    )

    top_indices = cluster_center.argsort()[::-1][:20]

    top_terms = [
        terms[index]
        for index in top_indices
    ]

    cluster_keywords[cluster_number] = top_terms

    print()
    print("=" * 70)
    print(f"CLUSTER {cluster_number}")
    print("=" * 70)

    print(
        ", ".join(top_terms)
    )


# ============================================================
# CELL 36 - CREATE CLUSTER KEYWORD TABLE
# ============================================================

keyword_table = []

for cluster_number, words in cluster_keywords.items():

    keyword_table.append({
        "Cluster": cluster_number,
        "Top Keywords": ", ".join(words)
    })

keyword_table = pd.DataFrame(
    keyword_table
)

display(keyword_table)


# ============================================================
# CELL 37 - SHOW SAMPLE BREACHES FROM EACH CLUSTER
# ============================================================

for cluster_number in range(best_k):

    print()
    print("=" * 80)
    print(f"SAMPLE RECORDS - CLUSTER {cluster_number}")
    print("=" * 80)

    cluster_data = df[
        df["Cluster"] == cluster_number
    ]

    display(
        cluster_data[
            text_columns + ["Cluster"]
        ].head(10)
    )


# ============================================================
# CELL 38 - FIND RECORDS CLOSEST TO EACH CLUSTER CENTRE
# ============================================================

# Distance from each record to its assigned cluster centre

distances = kmeans.transform(X)

df["Cluster_Distance"] = [
    distances[i, cluster_labels[i]]
    for i in range(len(df))
]


# ============================================================
# CELL 39 - REPRESENTATIVE RECORDS
# ============================================================

print("Representative records for each cluster:")

for cluster_number in range(best_k):

    cluster_records = df[
        df["Cluster"] == cluster_number
    ].sort_values(
        "Cluster_Distance"
    )

    print()
    print("=" * 70)
    print(
        f"REPRESENTATIVE RECORDS - CLUSTER {cluster_number}"
    )
    print("=" * 70)

    display(
        cluster_records[
            text_columns +
            ["Cluster", "Cluster_Distance"]
        ].head(5)
    )


# ============================================================
# CELL 40 - REDUCE DIMENSIONS USING SVD
# ============================================================

svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_svd = svd.fit_transform(X)

df["SVD_1"] = X_svd[:, 0]
df["SVD_2"] = X_svd[:, 1]

print(
    "Explained variance ratio:",
    svd.explained_variance_ratio_
)

print(
    "Total explained variance:",
    svd.explained_variance_ratio_.sum()
)


# ============================================================
# CELL 41 - VISUALIZE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=df,
    x="SVD_1",
    y="SVD_2",
    hue="Cluster",
    palette="tab10",
    s=70,
    alpha=0.7
)

plt.title(
    "Data Breaches - K-Means Clustering"
)

plt.xlabel("SVD Component 1")
plt.ylabel("SVD Component 2")

plt.legend(
    title="Cluster",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 42 - CALCULATE FINAL SILHOUETTE SCORE
# ============================================================

final_silhouette = silhouette_score(
    X,
    df["Cluster"]
)

print(
    f"Final Silhouette Score: "
    f"{final_silhouette:.4f}"
)


# ============================================================
# CELL 43 - CREATE CLUSTER LABELS
# ============================================================

df["Cluster_Label"] = (
    "Breach Cluster " +
    df["Cluster"].astype(str)
)

display(
    df[
        text_columns +
        ["Cluster", "Cluster_Label"]
    ].head(20)
)


# ============================================================
# CELL 44 - CLUSTER SUMMARY
# ============================================================

cluster_summary = []

for cluster_number in range(best_k):

    cluster_data = df[
        df["Cluster"] == cluster_number
    ]

    cluster_summary.append({
        "Cluster": cluster_number,
        "Number of Breaches": len(cluster_data),
        "Percentage": round(
            len(cluster_data) / len(df) * 100,
            2
        ),
        "Top Keywords": ", ".join(
            cluster_keywords[cluster_number][:10]
        )
    })

cluster_summary = pd.DataFrame(
    cluster_summary
)

display(cluster_summary)


# ============================================================
# CELL 45 - WORD FREQUENCY ANALYSIS
# ============================================================

all_words = []

for text in df["clean_text"]:

    all_words.extend(
        text.split()
    )

word_frequency = Counter(all_words)

frequency_table = pd.DataFrame(
    word_frequency.most_common(30),
    columns=["Word", "Frequency"]
)

display(frequency_table)


# ============================================================
# CELL 46 - WORD FREQUENCY GRAPH
# ============================================================

plt.figure(figsize=(12, 8))

sns.barplot(
    data=frequency_table.head(20),
    x="Frequency",
    y="Word"
)

plt.title(
    "Top 20 Most Frequently Used Words"
)

plt.xlabel("Frequency")
plt.ylabel("Word")

plt.tight_layout()

plt.show()


# ============================================================
# CELL 47 - OPTIONAL WORD CLOUD
# ============================================================

from wordcloud import WordCloud

text_for_wordcloud = " ".join(
    df["clean_text"]
)

wordcloud = WordCloud(
    width=1200,
    height=600,
    background_color="white",
    max_words=100
).generate(
    text_for_wordcloud
)

plt.figure(figsize=(15, 7))

plt.imshow(
    wordcloud,
    interpolation="bilinear"
)

plt.axis("off")

plt.title(
    "Word Cloud - Data Breaches"
)

plt.show()


# ============================================================
# CELL 48 - CLUSTER-SPECIFIC WORD CLOUDS
# ============================================================

for cluster_number in range(best_k):

    cluster_text = " ".join(
        df[
            df["Cluster"] == cluster_number
        ]["clean_text"]
    )

    if cluster_text.strip() == "":
        continue

    cluster_wordcloud = WordCloud(
        width=1000,
        height=500,
        background_color="white",
        max_words=75
    ).generate(
        cluster_text
    )

    plt.figure(figsize=(12, 6))

    plt.imshow(
        cluster_wordcloud,
        interpolation="bilinear"
    )

    plt.axis("off")

    plt.title(
        f"Word Cloud - Cluster {cluster_number}"
    )

    plt.show()


# ============================================================
# CELL 49 - SAVE CLUSTERED DATASET
# ============================================================

output_file = "Data_Breaches_KMeans_Results.csv"

df.to_csv(
    output_file,
    index=False
)

print(
    "Results saved successfully as:",
    output_file
)


# ============================================================
# CELL 50 - SAVE CLUSTER SUMMARY
# ============================================================

summary_file = "Data_Breaches_Cluster_Summary.csv"

cluster_summary.to_csv(
    summary_file,
    index=False
)

print(
    "Cluster summary saved as:",
    summary_file
)


# ============================================================
# CELL 51 - FINAL RESULTS
# ============================================================

print()
print("=" * 70)
print("FINAL ANALYSIS RESULTS")
print("=" * 70)

print(
    f"Total records: {len(df)}"
)

print(
    f"Text columns used: {len(text_columns)}"
)

print(
    f"TF-IDF features: {X.shape[1]}"
)

print(
    f"Optimal number of clusters: {best_k}"
)

print(
    f"Silhouette Score: {final_silhouette:.4f}"
)

print()
print("Cluster sizes:")

display(
    cluster_summary[
        [
            "Cluster",
            "Number of Breaches",
            "Percentage",
            "Top Keywords"
        ]
    ]
)


# ============================================================
# CELL 52 - FINAL DATASET PREVIEW
# ============================================================

display(
    df.head(20)
)


# ============================================================
# CELL 53 - CHECK OUTPUT FILES
# ============================================================

import os

print("Output files:")

for file in [
    "Data_Breaches_KMeans_Results.csv",
    "Data_Breaches_Cluster_Summary.csv"
]:

    if os.path.exists(file):
        print("✓", file)
    else:
        print("✗", file)
```
