```python
import pandas as pd

def safe_col(df, col):
    if col in df.columns:
        return df[col].fillna("Unknown").astype(str)
    else:
        return pd.Series(["Unknown"] * len(df))

df['combined_text'] = (
    safe_col(df, 'Name') + " " +
    safe_col(df, 'Genre') + " " +
    safe_col(df, 'Publisher')
)
```
