```python
# ============================================================
# Constituents Financials - K-Means + NLP Clustering
# ============================================================

# 1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# ------------------------------------------------------------
# 2. Load the dataset
# ------------------------------------------------------------

file_path = "constituents-financials.csv"

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
display(df.head())
print("\nColumns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 3. Basic data exploration
# ------------------------------------------------------------

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nSummary statistics:")
display(df.describe(include="all").T)


# ------------------------------------------------------------
# 4. Clean column names
# ------------------------------------------------------------

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print(df.columns.tolist())


# ------------------------------------------------------------
# 5. Identify numerical and text columns
# ------------------------------------------------------------

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
text_cols = df.select_dtypes(include=["object", "string"]).columns.tolist()

print("Numerical columns:")
print(numeric_cols)

print("\nText columns:")
print(text_cols)


# ------------------------------------------------------------
# 6. Prepare numerical financial data
# ------------------------------------------------------------

# Replace infinite values
df[numeric_cols] = df[numeric_cols].replace(
    [np.inf, -np.inf], np.nan
)

# Fill missing numerical values with median
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Standardize numerical variables
scaler = StandardScaler()

X_numeric = scaler.fit_transform(df[numeric_cols])

print("Numerical feature matrix:", X_numeric.shape)


# ------------------------------------------------------------
# 7. Prepare text data for NLP
# ------------------------------------------------------------

# Combine all text columns into one document per company
if len(text_cols) > 0:
    df["combined_text"] = (
        df[text_cols]
        .fillna("")
        .astype(str)
        .agg(" ".join, axis=1)
    )
else:
    df["combined_text"] = ""

display(df[["combined_text"]].head())


# ------------------------------------------------------------
# 8. TF-IDF transformation
# ------------------------------------------------------------

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=500,
    min_df=2
)

X_text = vectorizer.fit_transform(df["combined_text"])

print("NLP feature matrix:", X_text.shape)


# ------------------------------------------------------------
# 9. Combine financial + NLP features
# ------------------------------------------------------------

# Convert sparse NLP matrix to dense format
X_text_dense = X_text.toarray()

# Combine financial and NLP features
X = np.hstack([
    X_numeric,
    X_text_dense
])

print("Final feature matrix:", X.shape)


# ------------------------------------------------------------
# 10. Find a suitable number of clusters
# ------------------------------------------------------------

inertia = []
silhouette_scores = []

K_range = range(2, 11)

for k in K_range:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    inertia.append(model.inertia_)
    silhouette_scores.append(
        silhouette_score(X, labels)
    )


# Elbow plot
plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()


# Silhouette plot
plt.figure(figsize=(8, 5))
plt.plot(K_range, silhouette_scores, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score")
plt.show()


# ------------------------------------------------------------
# 11. Apply K-means
# ------------------------------------------------------------

# Change this value after examining the plots
K = 4

kmeans = KMeans(
    n_clusters=K,
    random_state=42,
    n_init=10
)

df["cluster"] = kmeans.fit_predict(X)

print("Cluster counts:")
display(df["cluster"].value_counts().sort_index())


# ------------------------------------------------------------
# 12. PCA visualization
# ------------------------------------------------------------

pca = PCA(n_components=2, random_state=42)

X_pca = pca.fit_transform(X)

plot_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "cluster": df["cluster"]
})

plt.figure(figsize=(10, 7))

for cluster in sorted(plot_df["cluster"].unique()):
    subset = plot_df[plot_df["cluster"] == cluster]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=f"Cluster {cluster}",
        alpha=0.7
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clusters")
plt.legend()
plt.show()


# ------------------------------------------------------------
# 13. Financial profile of each cluster
# ------------------------------------------------------------

cluster_summary = (
    df.groupby("cluster")[numeric_cols]
      .mean()
      .round(2)
)

display(cluster_summary)


# ------------------------------------------------------------
# 14. Cluster sizes
# ------------------------------------------------------------

cluster_sizes = (
    df["cluster"]
    .value_counts()
    .sort_index()
    .rename("Number of Companies")
)

display(cluster_sizes)


# ------------------------------------------------------------
# 15. Identify important NLP terms
# ------------------------------------------------------------

terms = vectorizer.get_feature_names_out()

# Calculate average TF-IDF value for each cluster
for cluster in sorted(df["cluster"].unique()):

    cluster_indices = df.index[df["cluster"] == cluster]

    cluster_tfidf = X_text[cluster_indices].mean(axis=0)

    scores = np.asarray(cluster_tfidf).flatten()

    top_indices = scores.argsort()[-15:][::-1]

    print(f"\nCluster {cluster} - Top NLP terms:")

    for i in top_indices:
        print(f"{terms[i]}: {scores[i]:.4f}")


# ------------------------------------------------------------
# 16. Display companies in each cluster
# ------------------------------------------------------------

# Try to find a likely company/name column
possible_name_cols = [
    "name",
    "company",
    "company_name",
    "symbol",
    "ticker"
]

name_col = None

for col in possible_name_cols:
    if col in df.columns:
        name_col = col
        break

if name_col:
    for cluster in sorted(df["cluster"].unique()):
        print(f"\nCluster {cluster}")
        display(
            df[df["cluster"] == cluster][
                [name_col] + numeric_cols + ["cluster"]
            ].head(20)
        )
else:
    print("No obvious company/name column was found.")


# ------------------------------------------------------------
# 17. Save results
# ------------------------------------------------------------

output_file = "constituents_financials_kmeans_results.csv"

df.to_csv(
    output_file,
    index=False
)

print(f"Results saved to: {output_file}")
```
