```python
# ============================================================
# TIPS.CSV - NLP + K-MEANS CLUSTERING
# Complete Jupyter Notebook
# ============================================================


# ============================================================
# CELL 1 - INSTALL LIBRARIES
# ============================================================

!pip install pandas numpy matplotlib seaborn scikit-learn nltk wordcloud


# ============================================================
# CELL 2 - IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings

warnings.filterwarnings("ignore")

# NLP
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Machine Learning
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.metrics import silhouette_score

# Word cloud
from wordcloud import WordCloud

# Download NLTK data
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

print("All libraries imported successfully.")


# ============================================================
# CELL 3 - LOAD DATASET
# ============================================================

FILE_NAME = "tips.csv"

df = pd.read_csv(FILE_NAME)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# CELL 4 - BASIC DATASET INFORMATION
# ============================================================

print("Dataset Information")
print("=" * 50)

print("\nShape:")
print(df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe(include="all"))


# ============================================================
# CELL 5 - REMOVE DUPLICATES
# ============================================================

before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)


# ============================================================
# CELL 6 - FIND A TEXT COLUMN FOR NLP
# ============================================================

# Possible names for text columns
possible_text_columns = [
    "text",
    "comment",
    "comments",
    "review",
    "reviews",
    "description",
    "message",
    "content",
    "feedback"
]

text_column = None

for column in df.columns:
    if column.lower() in possible_text_columns:
        text_column = column
        break

print("Detected text column:", text_column)

if text_column is None:
    print("\nNo text column was detected.")
    print("Available columns are:")
    print(df.columns.tolist())

    print("\nThe dataset will be processed using numerical/categorical")
    print("K-Means instead of NLP.")


# ============================================================
# PART A
# NLP + TF-IDF + K-MEANS
# ============================================================

if text_column is not None:

    print("\n")
    print("=" * 70)
    print("NLP + TF-IDF + K-MEANS")
    print("=" * 70)


    # ========================================================
    # CELL 7 - HANDLE MISSING TEXT VALUES
    # ========================================================

    df[text_column] = df[text_column].fillna("").astype(str)

    print("Missing text values handled.")


    # ========================================================
    # CELL 8 - NLP TEXT CLEANING
    # ========================================================

    stop_words = set(stopwords.words("english"))
    lemmatizer = WordNetLemmatizer()


    def clean_text(text):

        # Convert to lowercase
        text = text.lower()

        # Remove URLs
        text = re.sub(
            r"http\S+|www\S+|https\S+",
            " ",
            text
        )

        # Remove email addresses
        text = re.sub(
            r"\S+@\S+",
            " ",
            text
        )

        # Remove punctuation and numbers
        text = re.sub(
            r"[^a-zA-Z\s]",
            " ",
            text
        )

        # Split into words
        words = text.split()

        # Remove stopwords
        words = [
            word
            for word in words
            if word not in stop_words
        ]

        # Remove very short words
        words = [
            word
            for word in words
            if len(word) > 2
        ]

        # Lemmatization
        words = [
            lemmatizer.lemmatize(word)
            for word in words
        ]

        return " ".join(words)


    # Apply cleaning
    df["clean_text"] = df[text_column].apply(clean_text)

    print("Text preprocessing completed.")

    print("\nOriginal vs cleaned text:")
    display(
        df[[text_column, "clean_text"]].head(10)
    )


    # ========================================================
    # CELL 9 - REMOVE EMPTY TEXT
    # ========================================================

    df = df[
        df["clean_text"].str.strip() != ""
    ].reset_index(drop=True)

    print("Rows remaining:", len(df))


    # ========================================================
    # CELL 10 - TF-IDF
    # ========================================================

    vectorizer = TfidfVectorizer(
        max_features=3000,
        min_df=1,
        max_df=0.95,
        ngram_range=(1, 2),
        sublinear_tf=True
    )

    X = vectorizer.fit_transform(
        df["clean_text"]
    )

    print("TF-IDF completed.")
    print("TF-IDF matrix shape:", X.shape)


    # ========================================================
    # CELL 11 - FIND BEST NUMBER OF CLUSTERS
    # ========================================================

    max_k = min(10, len(df) - 1)

    K_values = range(2, max_k + 1)

    silhouette_scores = []
    inertia_values = []

    for k in K_values:

        kmeans_test = KMeans(
            n_clusters=k,
            random_state=42,
            n_init=10
        )

        labels = kmeans_test.fit_predict(X)

        inertia_values.append(
            kmeans_test.inertia_
        )

        score = silhouette_score(
            X,
            labels
        )

        silhouette_scores.append(score)

        print(
            f"K = {k} | "
            f"Silhouette Score = {score:.4f} | "
            f"Inertia = {kmeans_test.inertia_:.2f}"
        )


    # ========================================================
    # CELL 12 - SILHOUETTE SCORE GRAPH
    # ========================================================

    plt.figure(figsize=(9, 5))

    plt.plot(
        list(K_values),
        silhouette_scores,
        marker="o"
    )

    plt.xlabel("Number of Clusters (K)")
    plt.ylabel("Silhouette Score")
    plt.title("Silhouette Score for Different K Values")

    plt.xticks(list(K_values))
    plt.grid(True)

    plt.show()


    # ========================================================
    # CELL 13 - ELBOW METHOD
    # ========================================================

    plt.figure(figsize=(9, 5))

    plt.plot(
        list(K_values),
        inertia_values,
        marker="o"
    )

    plt.xlabel("Number of Clusters (K)")
    plt.ylabel("Inertia")
    plt.title("Elbow Method")

    plt.xticks(list(K_values))
    plt.grid(True)

    plt.show()


    # ========================================================
    # CELL 14 - SELECT BEST K
    # ========================================================

    best_k = list(K_values)[
        np.argmax(silhouette_scores)
    ]

    print("Best K:", best_k)

    print(
        "Best Silhouette Score:",
        round(max(silhouette_scores), 4)
    )


    # ========================================================
    # CELL 15 - FINAL K-MEANS MODEL
    # ========================================================

    kmeans = KMeans(
        n_clusters=best_k,
        random_state=42,
        n_init=10
    )

    df["cluster"] = kmeans.fit_predict(X)

    print("K-Means clustering completed.")


    # ========================================================
    # CELL 16 - CLUSTER COUNTS
    # ========================================================

    cluster_counts = (
        df["cluster"]
        .value_counts()
        .sort_index()
    )

    print("Number of records in each cluster:")
    print(cluster_counts)


    # ========================================================
    # CELL 17 - CLUSTER COUNT GRAPH
    # ========================================================

    plt.figure(figsize=(9, 5))

    sns.barplot(
        x=cluster_counts.index.astype(str),
        y=cluster_counts.values
    )

    plt.xlabel("Cluster")
    plt.ylabel("Number of Records")
    plt.title("Number of Records in Each Cluster")

    plt.show()


    # ========================================================
    # CELL 18 - IMPORTANT WORDS IN EACH CLUSTER
    # ========================================================

    terms = vectorizer.get_feature_names_out()

    order_centroids = (
        kmeans.cluster_centers_
        .argsort()[:, ::-1]
    )

    print("\nIMPORTANT WORDS PER CLUSTER")
    print("=" * 70)

    for cluster_number in range(best_k):

        top_words = [
            terms[index]
            for index in order_centroids[
                cluster_number, :20
            ]
        ]

        print(
            f"\nCluster {cluster_number}:"
        )

        print(
            ", ".join(top_words)
        )


    # ========================================================
    # CELL 19 - WORD CLOUD FOR EACH CLUSTER
    # ========================================================

    for cluster_number in range(best_k):

        cluster_text = " ".join(
            df[
                df["cluster"] == cluster_number
            ]["clean_text"]
        )

        if len(cluster_text.strip()) == 0:
            continue

        wordcloud = WordCloud(
            width=900,
            height=500,
            background_color="white",
            max_words=100
        ).generate(cluster_text)

        plt.figure(figsize=(10, 5))

        plt.imshow(
            wordcloud,
            interpolation="bilinear"
        )

        plt.axis("off")

        plt.title(
            f"Word Cloud - Cluster {cluster_number}"
        )

        plt.show()


    # ========================================================
    # CELL 20 - SHOW SAMPLE RECORDS FROM EACH CLUSTER
    # ========================================================

    for cluster_number in range(best_k):

        print("\n")
        print("=" * 70)
        print(
            f"CLUSTER {cluster_number}"
        )
        print("=" * 70)

        cluster_data = df[
            df["cluster"] == cluster_number
        ]

        display(
            cluster_data[
                [text_column, "clean_text"]
            ].head(10)
        )


    # ========================================================
    # CELL 21 - PCA VISUALIZATION
    # ========================================================

    # PCA needs a dense matrix.
    # Limit the size if the dataset is very large.

    if X.shape[0] <= 10000:

        X_dense = X.toarray()

        pca = PCA(
            n_components=2,
            random_state=42
        )

        X_pca = pca.fit_transform(
            X_dense
        )

        df["PCA1"] = X_pca[:, 0]
        df["PCA2"] = X_pca[:, 1]

        plt.figure(figsize=(10, 7))

        sns.scatterplot(
            data=df,
            x="PCA1",
            y="PCA2",
            hue="cluster",
            palette="tab10",
            s=70
        )

        plt.title(
            "K-Means Clustering Using NLP + TF-IDF"
        )

        plt.xlabel("PCA Component 1")
        plt.ylabel("PCA Component 2")

        plt.legend(
            title="Cluster"
        )

        plt.show()

    else:

        print(
            "Dataset is too large for dense PCA visualization."
        )


    # ========================================================
    # CELL 22 - SILHOUETTE SCORE OF FINAL MODEL
    # ========================================================

    final_silhouette = silhouette_score(
        X,
        df["cluster"]
    )

    print(
        "Final Silhouette Score:",
        round(final_silhouette, 4)
    )


    # ========================================================
    # CELL 23 - SAVE NLP CLUSTERING RESULTS
    # ========================================================

    output_file = "tips_nlp_kmeans_results.csv"

    df.to_csv(
        output_file,
        index=False
    )

    print(
        f"Results saved to: {output_file}"
    )


# ============================================================
# PART B
# STANDARD TIPS DATASET
# NUMERICAL + CATEGORICAL K-MEANS
# ============================================================

else:

    print("\n")
    print("=" * 70)
    print("STANDARD TIPS DATASET - K-MEANS")
    print("=" * 70)


    # ========================================================
    # CELL 24 - IDENTIFY NUMERICAL AND CATEGORICAL COLUMNS
    # ========================================================

    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_columns = df.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()

    print("Numerical columns:")
    print(numerical_columns)

    print("\nCategorical columns:")
    print(categorical_columns)


    # ========================================================
    # CELL 25 - PREPARE FEATURES
    # ========================================================

    # Remove columns that should not normally be used
    # as clustering features.

    numerical_features = [
        col for col in numerical_columns
        if col.lower() not in ["id", "index"]
    ]

    categorical_features = categorical_columns.copy()

    print(
        "\nNumerical features:",
        numerical_features
    )

    print(
        "Categorical features:",
        categorical_features
    )


    # ========================================================
    # CELL 26 - PREPROCESS FEATURES
    # ========================================================

    transformers = []

    if len(numerical_features) > 0:

        numerical_transformer = StandardScaler()

        transformers.append(
            (
                "numerical",
                numerical_transformer,
                numerical_features
            )
        )


    if len(categorical_features) > 0:

        categorical_transformer = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )

        transformers.append(
            (
                "categorical",
                categorical_transformer,
                categorical_features
            )
        )


    preprocessor = ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )


    X_processed = preprocessor.fit_transform(
        df
    )

    print(
        "\nProcessed feature matrix:",
        X_processed.shape
    )


    # ========================================================
    # CELL 27 - FIND BEST K
    # ========================================================

    max_k = min(10, len(df) - 1)

    K_values = range(
        2,
        max_k + 1
    )

    silhouette_scores = []
    inertia_values = []

    for k in K_values:

        kmeans_test = KMeans(
            n_clusters=k,
            random_state=42,
            n_init=10
        )

        labels = kmeans_test.fit_predict(
            X_processed
        )

        inertia_values.append(
            kmeans_test.inertia_
        )

        score = silhouette_score(
            X_processed,
            labels
        )

        silhouette_scores.append(
            score
        )

        print(
            f"K = {k} | "
            f"Silhouette Score = {score:.4f}"
        )


    # ========================================================
    # CELL 28 - SILHOUETTE GRAPH
    # ========================================================

    plt.figure(figsize=(9, 5))

    plt.plot(
        list(K_values),
        silhouette_scores,
        marker="o"
    )

    plt.xlabel("Number of Clusters")
    plt.ylabel("Silhouette Score")
    plt.title(
        "Silhouette Score - Tips Dataset"
    )

    plt.xticks(list(K_values))
    plt.grid(True)

    plt.show()


    # ========================================================
    # CELL 29 - ELBOW GRAPH
    # ========================================================

    plt.figure(figsize=(9, 5))

    plt.plot(
        list(K_values),
        inertia_values,
        marker="o"
    )

    plt.xlabel("Number of Clusters")
    plt.ylabel("Inertia")
    plt.title(
        "Elbow Method - Tips Dataset"
    )

    plt.xticks(list(K_values))
    plt.grid(True)

    plt.show()


    # ========================================================
    # CELL 30 - SELECT BEST K
    # ========================================================

    best_k = list(K_values)[
        np.argmax(silhouette_scores)
    ]

    print(
        "Best K:",
        best_k
    )

    print(
        "Best Silhouette Score:",
        round(
            max(silhouette_scores),
            4
        )
    )


    # ========================================================
    # CELL 31 - FINAL K-MEANS
    # ========================================================

    kmeans = KMeans(
        n_clusters=best_k,
        random_state=42,
        n_init=10
    )

    df["cluster"] = kmeans.fit_predict(
        X_processed
    )

    print(
        "K-Means clustering completed."
    )


    # ========================================================
    # CELL 32 - CLUSTER COUNTS
    # ========================================================

    cluster_counts = (
        df["cluster"]
        .value_counts()
        .sort_index()
    )

    print(
        "Records in each cluster:"
    )

    print(
        cluster_counts
    )


    # ========================================================
    # CELL 33 - CLUSTER GRAPH
    # ========================================================

    plt.figure(figsize=(9, 5))

    sns.barplot(
        x=cluster_counts.index.astype(str),
        y=cluster_counts.values
    )

    plt.xlabel("Cluster")
    plt.ylabel("Number of Records")
    plt.title(
        "Records in Each Cluster"
    )

    plt.show()


    # ========================================================
    # CELL 34 - PCA VISUALIZATION
    # ========================================================

    pca = PCA(
        n_components=2,
        random_state=42
    )

    X_pca = pca.fit_transform(
        X_processed
    )

    df["PCA1"] = X_pca[:, 0]
    df["PCA2"] = X_pca[:, 1]


    plt.figure(figsize=(10, 7))

    sns.scatterplot(
        data=df,
        x="PCA1",
        y="PCA2",
        hue="cluster",
        palette="tab10",
        s=70
    )

    plt.title(
        "K-Means Clustering - Tips Dataset"
    )

    plt.xlabel(
        "PCA Component 1"
    )

    plt.ylabel(
        "PCA Component 2"
    )

    plt.legend(
        title="Cluster"
    )

    plt.show()


    # ========================================================
    # CELL 35 - CLUSTER SUMMARY
    # ========================================================

    print(
        "Numerical averages by cluster:"
    )

    if len(numerical_features) > 0:

        cluster_summary = (
            df.groupby("cluster")[
                numerical_features
            ]
            .mean()
            .round(2)
        )

        display(
            cluster_summary
        )


    # ========================================================
    # CELL 36 - CATEGORICAL DISTRIBUTION
    # ========================================================

    for column in categorical_features:

        print(
            "\n" + "=" * 60
        )

        print(
            f"Distribution of {column} by cluster"
        )

        distribution = pd.crosstab(
            df["cluster"],
            df[column],
            normalize="index"
        ).round(3)

        display(
            distribution
        )


    # ========================================================
    # CELL 37 - FINAL SILHOUETTE SCORE
    # ========================================================

    final_silhouette = silhouette_score(
        X_processed,
        df["cluster"]
    )

    print(
        "Final Silhouette Score:",
        round(
            final_silhouette,
            4
        )
    )


    # ========================================================
    # CELL 38 - SAVE RESULTS
    # ========================================================

    output_file = (
        "tips_kmeans_results.csv"
    )

    df.to_csv(
        output_file,
        index=False
    )

    print(
        f"Results saved to: {output_file}"
    )


# ============================================================
# CELL 39 - FINAL RESULT
# ============================================================

print("\n")
print("=" * 70)
print("ANALYSIS COMPLETED")
print("=" * 70)

print("\nFinal dataset:")
display(df.head(10))

print("\nFinal columns:")
print(df.columns.tolist())

print("\nCluster distribution:")
print(
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("\nDone!")
```
