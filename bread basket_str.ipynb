````python

# Bread Basket - NLP + K-Means

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 1. Load dataset
df = pd.read_csv("bread basket_str.csv")

# 2. Clean data
df = df.drop_duplicates()
df["Item"] = df["Item"].fillna("").astype(str).str.lower()

# 3. NLP - TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(df["Item"])

# 4. K-Means
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# 5. Show clusters
print(df[["Item", "Cluster"]].head(20))

# 6. Cluster counts
print("\nCluster counts:")
print(df["Cluster"].value_counts())

# 7. PCA visualisation
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.toarray())

plt.figure(figsize=(8, 5))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Bread Basket - K-Means Clustering")
plt.show()

# 8. Save results
df.to_csv("bread_basket_clusters.csv", index=False)
```
````
