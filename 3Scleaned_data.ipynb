```python
# ============================================================
# K-MEANS + NLP CLUSTERING FOR cleaned_data.csv
# JUPYTER NOTEBOOK - FULL CODE
# ============================================================

# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

# Run this cell once if the libraries are not already installed.
# Uncomment the following line:

# !pip install pandas numpy scikit-learn matplotlib seaborn nltk wordcloud


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings

from IPython.display import display

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import TruncatedSVD

import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

warnings.filterwarnings("ignore")

print("Libraries imported successfully.")


# ============================================================
# 3. DOWNLOAD NLTK DATA
# ============================================================

nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

print("NLTK resources downloaded successfully.")


# ============================================================
# 4. LOAD DATASET
# ============================================================

FILE_NAME = "cleaned_data.csv"

df = pd.read_csv(FILE_NAME)

print("Dataset loaded successfully.")
print()
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 5. BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())


# ============================================================
# 6. REMOVE DUPLICATES
# ============================================================

df = df.drop_duplicates().reset_index(drop=True)

print("\nDataset after removing duplicates:")
print(df.shape)


# ============================================================
# 7. FIND TEXT COLUMNS
# ============================================================

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("\nText columns detected:")
for column in text_columns:
    print("-", column)


# ============================================================
# 8. SELECT TEXT COLUMN
# ============================================================

# ------------------------------------------------------------
# IMPORTANT:
#
# If the automatic selection is incorrect, change this line.
#
# Examples:
# TEXT_COLUMN = "review"
# TEXT_COLUMN = "text"
# TEXT_COLUMN = "comment"
# TEXT_COLUMN = "description"
# ------------------------------------------------------------

if len(text_columns) == 0:
    raise ValueError(
        "No text column was found in the dataset. "
        "Please check cleaned_data.csv."
    )

TEXT_COLUMN = text_columns[0]

print("\nSelected text column:", TEXT_COLUMN)


# ============================================================
# 9. CHECK TEXT DATA
# ============================================================

print("\nExample text data:")
display(df[[TEXT_COLUMN]].head(10))

print("\nNumber of missing values in text column:")
print(df[TEXT_COLUMN].isnull().sum())


# ============================================================
# 10. FILL MISSING TEXT VALUES
# ============================================================

df[TEXT_COLUMN] = df[TEXT_COLUMN].fillna("").astype(str)

print("\nMissing text values after filling:")
print(df[TEXT_COLUMN].isnull().sum())


# ============================================================
# 11. INITIALIZE NLP TOOLS
# ============================================================

stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()


# ============================================================
# 12. NLP TEXT CLEANING FUNCTION
# ============================================================

def clean_text(text):
    
    # Convert to lowercase
    text = text.lower()
    
    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )
    
    # Remove email addresses
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )
    
    # Remove HTML tags
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )
    
    # Remove numbers
    text = re.sub(
        r"\d+",
        " ",
        text
    )
    
    # Keep only letters and spaces
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )
    
    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()
    
    # Tokenize
    words = text.split()
    
    # Remove stopwords and short words
    words = [
        word
        for word in words
        if word not in stop_words
        and len(word) > 2
    ]
    
    # Lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]
    
    # Join words
    cleaned = " ".join(words)
    
    return cleaned


# ============================================================
# 13. APPLY NLP CLEANING
# ============================================================

print("\nCleaning text using NLP...")

df["clean_text"] = df[TEXT_COLUMN].apply(clean_text)

print("NLP preprocessing completed.")


# ============================================================
# 14. DISPLAY ORIGINAL VS CLEANED TEXT
# ============================================================

print("\nOriginal and cleaned text:")

display(
    df[
        [TEXT_COLUMN, "clean_text"]
    ].head(10)
)


# ============================================================
# 15. REMOVE EMPTY DOCUMENTS
# ============================================================

before = len(df)

df = df[
    df["clean_text"].str.strip() != ""
].reset_index(drop=True)

after = len(df)

print("\nDocuments before removing empty text:", before)
print("Documents after removing empty text:", after)
print("Documents removed:", before - after)


# ============================================================
# 16. TF-IDF VECTORIZATION
# ============================================================

print("\n" + "=" * 60)
print("TF-IDF VECTORIZATION")
print("=" * 60)

tfidf = TfidfVectorizer(
    max_features=5000,
    min_df=2,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X = tfidf.fit_transform(
    df["clean_text"]
)

print("\nTF-IDF matrix shape:")
print(X.shape)

print("\nNumber of TF-IDF features:")
print(len(tfidf.get_feature_names_out()))


# ============================================================
# 17. SHOW TF-IDF FEATURES
# ============================================================

feature_names = tfidf.get_feature_names_out()

print("\nFirst 50 TF-IDF features:")
print(feature_names[:50])


# ============================================================
# 18. FIND OPTIMAL NUMBER OF CLUSTERS
# ============================================================

print("\n" + "=" * 60)
print("FINDING OPTIMAL NUMBER OF CLUSTERS")
print("=" * 60)

# Test K values from 2 to 10

k_values = range(2, 11)

inertia_values = []
silhouette_values = []

for k in k_values:
    
    print(f"Testing K = {k}...")
    
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = model.fit_predict(X)
    
    # Inertia
    inertia_values.append(
        model.inertia_
    )
    
    # Silhouette score
    score = silhouette_score(
        X,
        labels
    )
    
    silhouette_values.append(score)
    
    print(
        f"   Inertia: {model.inertia_:.4f}"
    )
    
    print(
        f"   Silhouette Score: {score:.4f}"
    )


# ============================================================
# 19. DISPLAY CLUSTER EVALUATION RESULTS
# ============================================================

evaluation_df = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertia_values,
    "Silhouette Score": silhouette_values
})

print("\nCluster evaluation:")
display(evaluation_df)


# ============================================================
# 20. ELBOW METHOD PLOT
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Selecting K")
plt.xticks(list(k_values))
plt.grid(True)

plt.show()


# ============================================================
# 21. SILHOUETTE SCORE PLOT
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")
plt.xticks(list(k_values))
plt.grid(True)

plt.show()


# ============================================================
# 22. SELECT BEST K
# ============================================================

best_k = list(k_values)[
    np.argmax(silhouette_values)
]

best_score = max(silhouette_values)

print("\nBest K:", best_k)
print("Best Silhouette Score:", round(best_score, 4))


# ============================================================
# 23. TRAIN FINAL K-MEANS MODEL
# ============================================================

print("\n" + "=" * 60)
print("FINAL K-MEANS MODEL")
print("=" * 60)

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10,
    max_iter=300
)

cluster_labels = kmeans.fit_predict(X)

# Add cluster labels to dataframe
df["cluster"] = cluster_labels

print("\nK-Means clustering completed.")


# ============================================================
# 24. CLUSTER COUNTS
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("\nNumber of documents in each cluster:")

display(
    cluster_counts.to_frame(
        name="Number of Documents"
    )
)


# ============================================================
# 25. CLUSTER DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="cluster"
)

plt.xlabel("Cluster")
plt.ylabel("Number of Documents")
plt.title("Number of Documents per Cluster")

plt.show()


# ============================================================
# 26. REDUCE TF-IDF TO 2 DIMENSIONS
# ============================================================

print("\nReducing TF-IDF dimensions using SVD...")

svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = svd.fit_transform(X)

print("SVD completed.")

print(
    "Explained variance ratio:",
    svd.explained_variance_ratio_
)

print(
    "Total explained variance:",
    round(
        svd.explained_variance_ratio_.sum(),
        4
    )
)


# ============================================================
# 27. CREATE VISUALIZATION DATAFRAME
# ============================================================

visualization_df = pd.DataFrame({
    "Component_1": X_2d[:, 0],
    "Component_2": X_2d[:, 1],
    "cluster": df["cluster"]
})


# ============================================================
# 28. PLOT K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=visualization_df,
    x="Component_1",
    y="Component_2",
    hue="cluster",
    palette="tab10",
    s=70,
    alpha=0.7
)

plt.title(
    "K-Means Clusters Visualized with SVD"
)

plt.xlabel("SVD Component 1")
plt.ylabel("SVD Component 2")

plt.legend(
    title="Cluster",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()


# ============================================================
# 29. IDENTIFY TOP WORDS FOR EACH CLUSTER
# ============================================================

print("\n" + "=" * 60)
print("TOP WORDS IN EACH CLUSTER")
print("=" * 60)

terms = tfidf.get_feature_names_out()

order_centroids = (
    kmeans.cluster_centers_
    .argsort()[:, ::-1]
)

top_words_per_cluster = {}

for cluster_number in range(best_k):
    
    top_words = [
        terms[index]
        for index in order_centroids[
            cluster_number,
            :20
        ]
    ]
    
    top_words_per_cluster[
        cluster_number
    ] = top_words
    
    print(
        f"\nCluster {cluster_number}:"
    )
    
    print(
        ", ".join(top_words)
    )


# ============================================================
# 30. CREATE CLUSTER SUMMARY
# ============================================================

cluster_summary = []

for cluster_number in range(best_k):
    
    cluster_data = df[
        df["cluster"] == cluster_number
    ]
    
    cluster_summary.append({
        "Cluster": cluster_number,
        "Documents": len(cluster_data),
        "Percentage": round(
            len(cluster_data) / len(df) * 100,
            2
        ),
        "Top Words": ", ".join(
            top_words_per_cluster[
                cluster_number
            ][:10]
        )
    })

cluster_summary_df = pd.DataFrame(
    cluster_summary
)

print("\n" + "=" * 60)
print("CLUSTER SUMMARY")
print("=" * 60)

display(cluster_summary_df)


# ============================================================
# 31. SHOW EXAMPLES FROM EACH CLUSTER
# ============================================================

print("\n" + "=" * 60)
print("EXAMPLES FROM EACH CLUSTER")
print("=" * 60)

for cluster_number in range(best_k):
    
    print(
        f"\n{'=' * 60}\n"
        f"CLUSTER {cluster_number}\n"
        f"{'=' * 60}"
    )
    
    cluster_examples = df[
        df["cluster"] == cluster_number
    ][
        [TEXT_COLUMN, "clean_text"]
    ].head(10)
    
    display(cluster_examples)


# ============================================================
# 32. ADD CLUSTER LABEL / DESCRIPTION
# ============================================================

# Create a readable cluster label based on its top words.

cluster_names = {}

for cluster_number in range(best_k):
    
    words = top_words_per_cluster[
        cluster_number
    ][:3]
    
    cluster_names[
        cluster_number
    ] = " / ".join(words)

df["cluster_name"] = df["cluster"].map(
    cluster_names
)

print("\nCluster names:")
for cluster_number, name in cluster_names.items():
    print(
        f"Cluster {cluster_number}: {name}"
    )


# ============================================================
# 33. DISPLAY FINAL DATASET
# ============================================================

print("\n" + "=" * 60)
print("FINAL DATASET")
print("=" * 60)

display(df.head(20))


# ============================================================
# 34. SAVE CLUSTERED DATASET
# ============================================================

OUTPUT_FILE = "cleaned_data_with_kmeans_clusters.csv"

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    f"\nClustered dataset saved as: {OUTPUT_FILE}"
)


# ============================================================
# 35. SAVE CLUSTER SUMMARY
# ============================================================

SUMMARY_FILE = "cluster_summary.csv"

cluster_summary_df.to_csv(
    SUMMARY_FILE,
    index=False
)

print(
    f"Cluster summary saved as: {SUMMARY_FILE}"
)


# ============================================================
# 36. SAVE TOP WORDS FOR EACH CLUSTER
# ============================================================

top_words_df = pd.DataFrame(
    dict([
        (
            f"Cluster {cluster_number}",
            pd.Series(words)
        )
        for cluster_number, words
        in top_words_per_cluster.items()
    ])
)

top_words_df.to_csv(
    "cluster_top_words.csv",
    index=False
)

print(
    "Top cluster words saved as: cluster_top_words.csv"
)


# ============================================================
# 37. FINAL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(
    f"\nOriginal dataset: {FILE_NAME}"
)

print(
    f"Text column used: {TEXT_COLUMN}"
)

print(
    f"Number of documents: {len(df)}"
)

print(
    f"Number of TF-IDF features: {X.shape[1]}"
)

print(
    f"Optimal number of clusters: {best_k}"
)

print(
    f"Silhouette score: {best_score:.4f}"
)

print("\nCluster sizes:")

for cluster_number in range(best_k):
    
    count = (
        df["cluster"] == cluster_number
    ).sum()
    
    percentage = (
        count / len(df)
    ) * 100
    
    print(
        f"Cluster {cluster_number}: "
        f"{count} documents "
        f"({percentage:.2f}%)"
    )

print("\nTop words by cluster:")

for cluster_number in range(best_k):
    
    print(
        f"\nCluster {cluster_number}: "
        f"{', '.join(top_words_per_cluster[cluster_number][:10])}"
    )

print("\n" + "=" * 70)
print("PROCESS COMPLETED SUCCESSFULLY")
print("=" * 70)

```
