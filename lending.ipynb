```python
# ============================================================
# LENDING DATASET - NLP + K-MEANS CLUSTERING
# Complete Jupyter Notebook Code
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.sparse import csr_matrix, hstack

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import TruncatedSVD

warnings.filterwarnings("ignore")

print("Libraries imported successfully.")


# ============================================================
# 2. LOAD DATASET
# ============================================================

FILE_NAME = "LendingData.csv"

if not os.path.exists(FILE_NAME):
    raise FileNotFoundError(
        f"'{FILE_NAME}' was not found. "
        "Make sure the CSV file is in the same folder as your Jupyter Notebook."
    )

df = pd.read_csv(FILE_NAME)

print("=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 3. BASIC DATA EXPLORATION
# ============================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nColumn names:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")

print("\nData types:")
display(df.dtypes.to_frame("Data Type"))

print("\nMissing values:")
missing_values = df.isnull().sum().sort_values(ascending=False)
display(missing_values.to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe(include="all").T)


# ============================================================
# 4. REMOVE DUPLICATES
# ============================================================

before = len(df)

df = df.drop_duplicates()

df = df.reset_index(drop=True)

after = len(df)

print("\nDuplicates removed:", before - after)
print("Rows remaining:", after)


# ============================================================
# 5. IDENTIFY COLUMN TYPES
# ============================================================

numeric_cols = df.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_cols = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\n" + "=" * 70)
print("COLUMN TYPES")
print("=" * 70)

print("\nNumerical columns:")
print(numeric_cols)

print("\nCategorical/Text columns:")
print(categorical_cols)


# ============================================================
# 6. AUTOMATICALLY IDENTIFY TEXT COLUMNS FOR NLP
# ============================================================

text_keywords = [
    "description",
    "desc",
    "title",
    "purpose",
    "text",
    "occupation",
    "employment",
    "job",
    "reason",
    "comment",
    "note",
    "address",
    "summary",
    "detail"
]

text_cols = []

for col in categorical_cols:

    col_lower = col.lower()

    if any(keyword in col_lower for keyword in text_keywords):
        text_cols.append(col)


# If no obvious text columns are found, look for high-cardinality
# categorical columns that contain meaningful text.

if len(text_cols) == 0:

    for col in categorical_cols:

        unique_ratio = df[col].nunique(dropna=True) / max(len(df), 1)

        if unique_ratio > 0.05:

            sample = df[col].dropna().astype(str).head(100)

            if len(sample) > 0:

                average_length = sample.str.len().mean()

                if average_length > 15:
                    text_cols.append(col)


# Remove duplicated column names
text_cols = list(dict.fromkeys(text_cols))

print("\n" + "=" * 70)
print("TEXT COLUMNS SELECTED FOR NLP")
print("=" * 70)

if len(text_cols) == 0:
    print("No text columns were automatically detected.")
else:
    for col in text_cols:
        print("-", col)


# ============================================================
# 7. CREATE COMBINED TEXT COLUMN
# ============================================================

if len(text_cols) > 0:

    df["combined_text"] = (
        df[text_cols]
        .fillna("")
        .astype(str)
        .agg(" ".join, axis=1)
    )

else:

    # Create an empty column if no NLP columns exist
    df["combined_text"] = ""


# ============================================================
# 8. NLP TEXT CLEANING
# ============================================================

def clean_text(text):

    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+",
        " ",
        text
    )

    # Remove email addresses
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # Remove punctuation
    text = re.sub(
        r"[^a-zA-Z0-9\s]",
        " ",
        text
    )

    # Remove numbers
    text = re.sub(
        r"\d+",
        " ",
        text
    )

    # Remove extra whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


df["clean_text"] = df["combined_text"].apply(clean_text)

print("\nExample of cleaned text:")
display(
    df[["combined_text", "clean_text"]].head()
)


# ============================================================
# 9. TF-IDF NLP
# ============================================================

print("\n" + "=" * 70)
print("CREATING TF-IDF FEATURES")
print("=" * 70)

# If there is insufficient text, use a small fallback feature.
valid_text = df["clean_text"].str.strip()

if valid_text.str.len().sum() > 0:

    tfidf = TfidfVectorizer(
        stop_words="english",
        max_features=1000,
        min_df=1,
        max_df=0.95,
        ngram_range=(1, 2),
        sublinear_tf=True
    )

    tfidf_matrix = tfidf.fit_transform(
        df["clean_text"]
    )

    tfidf_features = tfidf.get_feature_names_out()

    print("TF-IDF matrix:", tfidf_matrix.shape)
    print("Number of NLP features:", len(tfidf_features))

else:

    # Empty sparse matrix
    tfidf_matrix = csr_matrix(
        (len(df), 0)
    )

    tfidf_features = np.array([])

    print("No usable text was found for NLP.")


# ============================================================
# 10. SHOW TOP WORDS IN ENTIRE DATASET
# ============================================================

if tfidf_matrix.shape[1] > 0:

    overall_tfidf = np.asarray(
        tfidf_matrix.mean(axis=0)
    ).flatten()

    top_word_indices = (
        overall_tfidf.argsort()[-20:][::-1]
    )

    print("\nTop 20 NLP terms in the dataset:")

    for rank, index in enumerate(
        top_word_indices,
        1
    ):
        print(
            f"{rank:2}. "
            f"{tfidf_features[index]} "
            f"({overall_tfidf[index]:.4f})"
        )


# ============================================================
# 11. IDENTIFY ID COLUMNS
# ============================================================

id_cols = []

for col in df.columns:

    col_lower = col.lower()

    if (
        col_lower == "id"
        or col_lower.endswith("_id")
        or col_lower.startswith("id_")
        or "customerid" in col_lower
        or "customer_id" in col_lower
        or "loanid" in col_lower
        or "loan_id" in col_lower
    ):

        id_cols.append(col)

print("\nID columns excluded from clustering:")
print(id_cols)


# ============================================================
# 12. SELECT NUMERICAL FEATURES
# ============================================================

numeric_features = [
    col for col in numeric_cols
    if col not in id_cols
]

# Exclude Cluster if it already exists
numeric_features = [
    col for col in numeric_features
    if col != "Cluster"
]

print("\nNumerical features used for clustering:")

for col in numeric_features:
    print("-", col)


# ============================================================
# 13. PREPARE NUMERICAL DATA
# ============================================================

if len(numeric_features) > 0:

    numerical_data = df[
        numeric_features
    ].copy()

    # Convert to numeric
    for col in numeric_features:

        numerical_data[col] = pd.to_numeric(
            numerical_data[col],
            errors="coerce"
        )

    # Replace infinite values
    numerical_data = numerical_data.replace(
        [np.inf, -np.inf],
        np.nan
    )

    # Fill missing values with median
    for col in numeric_features:

        median_value = numerical_data[col].median()

        if pd.isna(median_value):
            median_value = 0

        numerical_data[col] = (
            numerical_data[col]
            .fillna(median_value)
        )

    # Standardise numerical data
    scaler = StandardScaler()

    numerical_scaled = scaler.fit_transform(
        numerical_data
    )

    numerical_sparse = csr_matrix(
        numerical_scaled
    )

    print(
        "\nNumerical feature matrix:",
        numerical_sparse.shape
    )

else:

    numerical_sparse = csr_matrix(
        (len(df), 0)
    )

    print("\nNo numerical features available.")


# ============================================================
# 14. PREPARE CATEGORICAL DATA
# ============================================================

categorical_features = [
    col for col in categorical_cols
    if col not in text_cols
    and col != "combined_text"
    and col != "clean_text"
]

# Exclude ID columns
categorical_features = [
    col for col in categorical_features
    if col not in id_cols
]

print("\nCategorical features used:")

for col in categorical_features:
    print("-", col)


# ============================================================
# 15. ONE-HOT ENCODE CATEGORICAL DATA
# ============================================================

if len(categorical_features) > 0:

    categorical_data = df[
        categorical_features
    ].copy()

    # Convert missing values
    for col in categorical_features:
        categorical_data[col] = (
            categorical_data[col]
            .fillna("Unknown")
            .astype(str)
        )

    # Compatible with different sklearn versions
    try:

        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=True
        )

    except TypeError:

        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=True
        )

    categorical_encoded = encoder.fit_transform(
        categorical_data
    )

    print(
        "\nCategorical encoded matrix:",
        categorical_encoded.shape
    )

else:

    categorical_encoded = csr_matrix(
        (len(df), 0)
    )

    print("\nNo categorical features available.")


# ============================================================
# 16. COMBINE ALL FEATURES
# ============================================================

print("\n" + "=" * 70)
print("COMBINING FEATURES")
print("=" * 70)

feature_matrices = []

if numerical_sparse.shape[1] > 0:
    feature_matrices.append(
        numerical_sparse
    )

if categorical_encoded.shape[1] > 0:
    feature_matrices.append(
        categorical_encoded
    )

if tfidf_matrix.shape[1] > 0:
    feature_matrices.append(
        tfidf_matrix
    )

if len(feature_matrices) == 0:

    raise ValueError(
        "No usable features were found for clustering."
    )

X = hstack(
    feature_matrices
).tocsr()

print("Final feature matrix:", X.shape)


# ============================================================
# 17. FIND OPTIMAL NUMBER OF CLUSTERS
# ============================================================

print("\n" + "=" * 70)
print("FINDING OPTIMAL NUMBER OF CLUSTERS")
print("=" * 70)

# Maximum K should not exceed number of records - 1
max_k = min(
    10,
    len(df) - 1
)

if max_k < 2:

    raise ValueError(
        "The dataset needs at least 3 rows for K-Means clustering."
    )

K_values = range(
    2,
    max_k + 1
)

inertias = []
silhouette_scores = []


for k in K_values:

    print(
        f"Testing K = {k}..."
    )

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    inertias.append(
        model.inertia_
    )

    # Silhouette score can be expensive on very large datasets
    sample_size = min(
        10000,
        len(df)
    )

    score = silhouette_score(
        X,
        labels,
        sample_size=sample_size,
        random_state=42
    )

    silhouette_scores.append(
        score
    )


# ============================================================
# 18. ELBOW PLOT
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    list(K_values),
    inertias,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.title(
    "Elbow Method"
)

plt.grid(
    True,
    alpha=0.3
)

plt.show()


# ============================================================
# 19. SILHOUETTE SCORE PLOT
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    list(K_values),
    silhouette_scores,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "Silhouette Score"
)

plt.grid(
    True,
    alpha=0.3
)

plt.show()


# ============================================================
# 20. DISPLAY SCORES
# ============================================================

evaluation_table = pd.DataFrame({
    "K": list(K_values),
    "Inertia": inertias,
    "Silhouette Score": silhouette_scores
})

display(
    evaluation_table
)


# ============================================================
# 21. SELECT BEST K
# ============================================================

best_k = list(
    K_values
)[
    np.argmax(
        silhouette_scores
    )
]

print(
    "\nRecommended K based on highest Silhouette Score:",
    best_k
)


# ============================================================
# 22. FINAL K-MEANS MODEL
# ============================================================

print("\n" + "=" * 70)
print("RUNNING FINAL K-MEANS")
print("=" * 70)

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=20
)

cluster_labels = kmeans.fit_predict(
    X
)

df["Cluster"] = cluster_labels

print(
    "K-Means clustering completed."
)


# ============================================================
# 23. CLUSTER COUNTS
# ============================================================

cluster_counts = (
    df["Cluster"]
    .value_counts()
    .sort_index()
)

print("\nNumber of records in each cluster:")

display(
    cluster_counts.to_frame(
        "Number of Records"
    )
)


# ============================================================
# 24. CLUSTER SIZE VISUALISATION
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.xlabel(
    "Cluster"
)

plt.ylabel(
    "Number of Records"
)

plt.title(
    "Number of Lending Records per Cluster"
)

plt.show()


# ============================================================
# 25. REDUCE FEATURES TO 2D FOR VISUALISATION
# ============================================================

print("\nCreating 2D cluster visualisation...")

# TruncatedSVD works directly with sparse matrices
svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = svd.fit_transform(
    X
)

print(
    "Explained variance:",
    svd.explained_variance_ratio_.sum()
)


# ============================================================
# 26. PLOT K-MEANS CLUSTERS
# ============================================================

plt.figure(
    figsize=(12, 8)
)

scatter = plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=df["Cluster"],
    alpha=0.6,
    s=30
)

plt.xlabel(
    "Component 1"
)

plt.ylabel(
    "Component 2"
)

plt.title(
    "K-Means Clusters - Lending Data"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(
    True,
    alpha=0.2
)

plt.show()


# ============================================================
# 27. NUMERICAL CLUSTER ANALYSIS
# ============================================================

if len(numeric_features) > 0:

    print("\n" + "=" * 70)
    print("NUMERICAL FEATURES BY CLUSTER")
    print("=" * 70)

    numerical_cluster_summary = (
        df.groupby("Cluster")[
            numeric_features
        ]
        .mean()
        .round(2)
    )

    display(
        numerical_cluster_summary
    )


# ============================================================
# 28. MEDIAN VALUES BY CLUSTER
# ============================================================

if len(numeric_features) > 0:

    print("\nMedian numerical values by cluster:")

    numerical_median_summary = (
        df.groupby("Cluster")[
            numeric_features
        ]
        .median()
        .round(2)
    )

    display(
        numerical_median_summary
    )


# ============================================================
# 29. CATEGORICAL ANALYSIS
# ============================================================

if len(categorical_features) > 0:

    print("\n" + "=" * 70)
    print("CATEGORICAL FEATURES BY CLUSTER")
    print("=" * 70)

    for col in categorical_features:

        print("\n")
        print("-" * 70)
        print(f"Column: {col}")
        print("-" * 70)

        cross_table = pd.crosstab(
            df["Cluster"],
            df[col],
            normalize="index"
        ) * 100

        display(
            cross_table.round(2)
        )


# ============================================================
# 30. TOP NLP TERMS FOR EACH CLUSTER
# ============================================================

print("\n" + "=" * 70)
print("TOP NLP TERMS BY CLUSTER")
print("=" * 70)

if tfidf_matrix.shape[1] > 0:

    for cluster_number in range(best_k):

        cluster_indices = np.where(
            cluster_labels == cluster_number
        )[0]

        cluster_matrix = (
            tfidf_matrix[
                cluster_indices
            ]
        )

        mean_tfidf = np.asarray(
            cluster_matrix.mean(axis=0)
        ).flatten()

        top_indices = (
            mean_tfidf.argsort()[-20:][::-1]
        )

        print(
            f"\nCLUSTER {cluster_number}"
        )

        for rank, index in enumerate(
            top_indices,
            1
        ):

            print(
                f"{rank:2}. "
                f"{tfidf_features[index]:25} "
                f"{mean_tfidf[index]:.4f}"
            )

else:

    print(
        "No text features were available."
    )


# ============================================================
# 31. NLP WORD FREQUENCY BY CLUSTER
# ============================================================

if tfidf_matrix.shape[1] > 0:

    cluster_nlp_tables = {}

    for cluster_number in range(best_k):

        cluster_indices = np.where(
            cluster_labels == cluster_number
        )[0]

        cluster_matrix = (
            tfidf_matrix[
                cluster_indices
            ]
        )

        mean_scores = np.asarray(
            cluster_matrix.mean(axis=0)
        ).flatten()

        top_indices = (
            mean_scores.argsort()[-15:][::-1]
        )

        table = pd.DataFrame({
            "Word": [
                tfidf_features[i]
                for i in top_indices
            ],
            "TF-IDF Score": [
                mean_scores[i]
                for i in top_indices
            ]
        })

        cluster_nlp_tables[
            cluster_number
        ] = table

    # Display all cluster NLP tables
    for cluster_number, table in cluster_nlp_tables.items():

        print(
            f"\nCluster {cluster_number} - Top Terms"
        )

        display(
            table
        )


# ============================================================
# 32. SHOW SAMPLE RECORDS FROM EACH CLUSTER
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE RECORDS FROM EACH CLUSTER")
print("=" * 70)

# Don't display clean NLP columns unless needed
display_columns = [
    col for col in df.columns
    if col not in [
        "combined_text",
        "clean_text"
    ]
]

for cluster_number in range(best_k):

    print(
        f"\nCLUSTER {cluster_number}"
    )

    cluster_sample = (
        df[
            df["Cluster"] == cluster_number
        ][display_columns]
        .head(10)
    )

    display(
        cluster_sample
    )


# ============================================================
# 33. CALCULATE FINAL SILHOUETTE SCORE
# ============================================================

final_silhouette = silhouette_score(
    X,
    cluster_labels,
    sample_size=min(
        10000,
        len(df)
    ),
    random_state=42
)

print("\n" + "=" * 70)
print("FINAL MODEL PERFORMANCE")
print("=" * 70)

print(
    "Number of clusters:",
    best_k
)

print(
    "Final inertia:",
    kmeans.inertia_
)

print(
    "Final silhouette score:",
    round(
        final_silhouette,
        4
    )
)


# ============================================================
# 34. CREATE CLUSTER SUMMARY TABLE
# ============================================================

cluster_summary = pd.DataFrame({
    "Cluster": sorted(
        df["Cluster"].unique()
    ),
    "Number of Records": [
        (df["Cluster"] == cluster).sum()
        for cluster in sorted(
            df["Cluster"].unique()
        )
    ]
})

cluster_summary[
    "Percentage"
] = (
    cluster_summary["Number of Records"]
    / len(df)
    * 100
)

cluster_summary[
    "Percentage"
] = cluster_summary[
    "Percentage"
].round(2)

print("\n" + "=" * 70)
print("CLUSTER SUMMARY")
print("=" * 70)

display(
    cluster_summary
)


# ============================================================
# 35. SAVE CLUSTERED DATASET
# ============================================================

OUTPUT_FILE = "LendingData_KMeans_NLP_Clustered.csv"

# Drop helper NLP columns before saving if preferred
output_df = df.drop(
    columns=[
        "combined_text",
        "clean_text"
    ],
    errors="ignore"
)

output_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    "\nClustered dataset saved successfully:"
)

print(
    OUTPUT_FILE
)


# ============================================================
# 36. SAVE CLUSTER SUMMARY
# ============================================================

SUMMARY_FILE = "LendingData_Cluster_Summary.csv"

cluster_summary.to_csv(
    SUMMARY_FILE,
    index=False
)

print(
    "Cluster summary saved as:",
    SUMMARY_FILE
)


# ============================================================
# 37. SAVE NLP TERMS
# ============================================================

if tfidf_matrix.shape[1] > 0:

    nlp_results = []

    for cluster_number in range(best_k):

        cluster_indices = np.where(
            cluster_labels == cluster_number
        )[0]

        cluster_matrix = (
            tfidf_matrix[
                cluster_indices
            ]
        )

        mean_scores = np.asarray(
            cluster_matrix.mean(axis=0)
        ).flatten()

        top_indices = (
            mean_scores.argsort()[-20:][::-1]
        )

        for index in top_indices:

            nlp_results.append({
                "Cluster": cluster_number,
                "Word": tfidf_features[index],
                "TF-IDF Score": mean_scores[index]
            })

    nlp_results_df = pd.DataFrame(
        nlp_results
    )

    NLP_FILE = (
        "LendingData_Cluster_NLP_Terms.csv"
    )

    nlp_results_df.to_csv(
        NLP_FILE,
        index=False
    )

    print(
        "NLP terms saved as:",
        NLP_FILE
    )


# ============================================================
# 38. FINAL OUTPUT
# ============================================================

print("\n")
print("=" * 70)
print("ANALYSIS COMPLETE")
print("=" * 70)

print(
    f"Original dataset: {FILE_NAME}"
)

print(
    f"Final number of records: {len(df)}"
)

print(
    f"Number of clusters: {best_k}"
)

print(
    f"Silhouette score: {final_silhouette:.4f}"
)

print(
    "\nOutput files:"
)

print(
    f"1. {OUTPUT_FILE}"
)

print(
    f"2. {SUMMARY_FILE}"
)

if tfidf_matrix.shape[1] > 0:

    print(
        f"3. {NLP_FILE}"
    )

print(
    "\nK-Means + NLP analysis finished successfully."
)
```
