```python
# ============================================================
# NINTENDO 64 DATASET
# NLP + TF-IDF + K-MEANS CLUSTERING
# ============================================================


# ============================================================
# CELL 1 - INSTALL REQUIRED LIBRARIES
# ============================================================

# Run this cell only if the libraries are not already installed.

# !pip install pandas numpy matplotlib seaborn scikit-learn nltk wordcloud


# ============================================================
# CELL 2 - IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import re
import string
import warnings

warnings.filterwarnings("ignore")

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics import silhouette_score

import nltk
from nltk.corpus import stopwords
from wordcloud import WordCloud


# Download English stopwords
nltk.download("stopwords")

stop_words = set(stopwords.words("english"))

print("Libraries imported successfully.")


# ============================================================
# CELL 3 - LOAD DATASET
# ============================================================

file_path = "Nintendo - Nintendo 64.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


# ============================================================
# CELL 4 - DISPLAY FIRST 5 ROWS
# ============================================================

df.head()


# ============================================================
# CELL 5 - DISPLAY LAST 5 ROWS
# ============================================================

df.tail()


# ============================================================
# CELL 6 - DATASET INFORMATION
# ============================================================

print("Dataset Information")
print("=" * 60)

df.info()


# ============================================================
# CELL 7 - COLUMN NAMES
# ============================================================

print("Column names:")
print("=" * 60)

for i, column in enumerate(df.columns):
    print(i, ":", column)


# ============================================================
# CELL 8 - DATASET SIZE
# ============================================================

rows, columns = df.shape

print("Rows:", rows)
print("Columns:", columns)


# ============================================================
# CELL 9 - CHECK DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)


# ============================================================
# CELL 10 - REMOVE DUPLICATES
# ============================================================

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)


# ============================================================
# CELL 11 - CHECK MISSING VALUES
# ============================================================

missing = df.isnull().sum()

print("Missing values:")
print("=" * 60)

print(missing)


# ============================================================
# CELL 12 - MISSING VALUES PERCENTAGE
# ============================================================

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

missing_table = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

missing_table


# ============================================================
# CELL 13 - BASIC STATISTICS
# ============================================================

df.describe(include="all").T


# ============================================================
# CELL 14 - DATA TYPES
# ============================================================

print("Data types:")
print("=" * 60)

print(df.dtypes)


# ============================================================
# CELL 15 - IDENTIFY NUMERIC COLUMNS
# ============================================================

numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

print("Numeric columns:")
print(list(numeric_columns))


# ============================================================
# CELL 16 - IDENTIFY TEXT COLUMNS
# ============================================================

text_columns = df.select_dtypes(
    include=["object"]
).columns

print("Text columns:")
print(list(text_columns))


# ============================================================
# CELL 17 - NUMBER OF UNIQUE VALUES
# ============================================================

unique_values = df.nunique().sort_values(ascending=False)

print("Unique values per column:")
print("=" * 60)

print(unique_values)


# ============================================================
# CELL 18 - EXPLORATORY DATA ANALYSIS
# NUMERIC COLUMN DISTRIBUTIONS
# ============================================================

for column in numeric_columns:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[column].dropna(),
        bins=20
    )

    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"Distribution of {column}")

    plt.show()


# ============================================================
# CELL 19 - CORRELATION MATRIX
# ============================================================

if len(numeric_columns) > 1:

    plt.figure(figsize=(10, 7))

    correlation = df[numeric_columns].corr()

    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Matrix")

    plt.show()


# ============================================================
# CELL 20 - CREATE COMBINED TEXT
# ============================================================

# Fill missing text values
df[text_columns] = df[text_columns].fillna("")


# Combine all text columns into one column
df["combined_text"] = df[text_columns].astype(str).agg(
    " ".join,
    axis=1
)

print("Combined text created successfully.")

df[["combined_text"]].head()


# ============================================================
# CELL 21 - NLP TEXT CLEANING FUNCTION
# ============================================================

def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        "",
        text
    )

    # Remove HTML
    text = re.sub(
        r"<.*?>",
        "",
        text
    )

    # Remove numbers
    text = re.sub(
        r"\d+",
        "",
        text
    )

    # Remove punctuation
    text = text.translate(
        str.maketrans(
            "",
            "",
            string.punctuation
        )
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Remove stopwords
    words = text.split()

    words = [
        word for word in words
        if word not in stop_words
    ]

    # Join words
    text = " ".join(words)

    return text.strip()


# ============================================================
# CELL 22 - APPLY NLP CLEANING
# ============================================================

df["cleaned_text"] = df["combined_text"].apply(
    clean_text
)

df[[
    "combined_text",
    "cleaned_text"
]].head(10)


# ============================================================
# CELL 23 - CHECK EMPTY TEXT RECORDS
# ============================================================

empty_text = (
    df["cleaned_text"]
    .str.strip()
    .eq("")
    .sum()
)

print("Records with empty text:", empty_text)


# ============================================================
# CELL 24 - REMOVE EMPTY TEXT RECORDS
# ============================================================

df = df[
    df["cleaned_text"].str.strip() != ""
].copy()

df.reset_index(
    drop=True,
    inplace=True
)

print("Dataset shape:", df.shape)


# ============================================================
# CELL 25 - MOST COMMON WORDS
# ============================================================

from collections import Counter

all_words = " ".join(
    df["cleaned_text"]
).split()

word_counts = Counter(all_words)

common_words = word_counts.most_common(20)

print("Top 20 words:")
print("=" * 60)

for word, count in common_words:
    print(f"{word}: {count}")


# ============================================================
# CELL 26 - WORD FREQUENCY CHART
# ============================================================

words = [
    item[0]
    for item in common_words
]

counts = [
    item[1]
    for item in common_words
]

plt.figure(figsize=(12, 6))

plt.bar(
    words,
    counts
)

plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Top 20 Most Frequent Words")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 27 - WORD CLOUD
# ============================================================

all_text = " ".join(
    df["cleaned_text"]
)

wordcloud = WordCloud(
    width=1200,
    height=600,
    background_color="white",
    stopwords=stop_words
).generate(all_text)

plt.figure(
    figsize=(15, 7)
)

plt.imshow(
    wordcloud,
    interpolation="bilinear"
)

plt.axis("off")

plt.title(
    "Nintendo 64 Dataset Word Cloud"
)

plt.show()


# ============================================================
# CELL 28 - TF-IDF VECTORIZATION
# ============================================================

tfidf = TfidfVectorizer(
    max_features=5000,
    min_df=1,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X = tfidf.fit_transform(
    df["cleaned_text"]
)

print("TF-IDF completed.")

print(
    "Number of documents:",
    X.shape[0]
)

print(
    "Number of TF-IDF features:",
    X.shape[1]
)


# ============================================================
# CELL 29 - DISPLAY TF-IDF FEATURES
# ============================================================

feature_names = tfidf.get_feature_names_out()

print("First 50 TF-IDF features:")

print(
    feature_names[:50]
)


# ============================================================
# CELL 30 - ELBOW METHOD
# ============================================================

inertia = []

k_values = range(2, 11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)

    inertia.append(
        model.inertia_
    )


# Plot elbow curve

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    k_values,
    inertia,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.title(
    "Elbow Method for K-Means"
)

plt.xticks(
    list(k_values)
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 31 - SILHOUETTE SCORE
# ============================================================

silhouette_scores = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k} | Silhouette Score = {score:.4f}"
    )


# ============================================================
# CELL 32 - PLOT SILHOUETTE SCORES
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    k_values,
    silhouette_scores,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "Silhouette Score for Different K Values"
)

plt.xticks(
    list(k_values)
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 33 - FIND BEST K
# ============================================================

best_index = np.argmax(
    silhouette_scores
)

best_k = list(k_values)[
    best_index
]

best_score = silhouette_scores[
    best_index
]

print(
    "Best K:",
    best_k
)

print(
    "Best Silhouette Score:",
    round(best_score, 4)
)


# ============================================================
# CELL 34 - TRAIN FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=20
)

cluster_labels = kmeans.fit_predict(X)

df["Cluster"] = cluster_labels

print(
    "K-Means clustering completed."
)


# ============================================================
# CELL 35 - DISPLAY CLUSTER ASSIGNMENTS
# ============================================================

df[
    ["cleaned_text", "Cluster"]
].head(20)


# ============================================================
# CELL 36 - NUMBER OF RECORDS IN EACH CLUSTER
# ============================================================

cluster_counts = (
    df["Cluster"]
    .value_counts()
    .sort_index()
)

print(
    cluster_counts
)


# ============================================================
# CELL 37 - CLUSTER SIZE CHART
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.xlabel(
    "Cluster"
)

plt.ylabel(
    "Number of Games"
)

plt.title(
    "Number of Nintendo 64 Games in Each Cluster"
)

plt.show()


# ============================================================
# CELL 38 - CLUSTER PERCENTAGES
# ============================================================

cluster_percentage = (
    df["Cluster"]
    .value_counts(
        normalize=True
    )
    .sort_index()
    * 100
)

cluster_percentage


# ============================================================
# CELL 39 - MOST IMPORTANT WORDS PER CLUSTER
# ============================================================

terms = tfidf.get_feature_names_out()

order_centroids = (
    kmeans.cluster_centers_
    .argsort()[:, ::-1]
)

print(
    "Top words in each cluster:"
)

print("=" * 70)

for cluster_number in range(best_k):

    print(
        f"\nCLUSTER {cluster_number}"
    )

    top_words = [
        terms[index]
        for index in
        order_centroids[
            cluster_number,
            :20
        ]
    ]

    print(
        ", ".join(top_words)
    )


# ============================================================
# CELL 40 - CREATE CLUSTER SUMMARY
# ============================================================

cluster_summary = []

for cluster_number in range(best_k):

    cluster_data = df[
        df["Cluster"] == cluster_number
    ]

    top_words = [
        terms[index]
        for index in
        order_centroids[
            cluster_number,
            :10
        ]
    ]

    cluster_summary.append({

        "Cluster":
            cluster_number,

        "Number of Games":
            len(cluster_data),

        "Percentage":
            round(
                len(cluster_data)
                / len(df)
                * 100,
                2
            ),

        "Top Keywords":
            ", ".join(top_words)
    })


cluster_summary = pd.DataFrame(
    cluster_summary
)

cluster_summary


# ============================================================
# CELL 41 - WORD CLOUD FOR EACH CLUSTER
# ============================================================

for cluster_number in range(best_k):

    cluster_text = " ".join(
        df[
            df["Cluster"] == cluster_number
        ]["cleaned_text"]
    )

    if cluster_text.strip() == "":
        continue

    cluster_wordcloud = WordCloud(
        width=1000,
        height=500,
        background_color="white",
        stopwords=stop_words
    ).generate(cluster_text)

    plt.figure(
        figsize=(12, 6)
    )

    plt.imshow(
        cluster_wordcloud,
        interpolation="bilinear"
    )

    plt.axis("off")

    plt.title(
        f"Word Cloud - Cluster {cluster_number}"
    )

    plt.show()


# ============================================================
# CELL 42 - REDUCE TF-IDF DIMENSIONS
# USING Truncated SVD
# ============================================================

# TruncatedSVD works directly with the sparse TF-IDF matrix.

svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = svd.fit_transform(X)

print(
    "Explained variance:",
    svd.explained_variance_ratio_
)

print(
    "Total explained variance:",
    round(
        svd.explained_variance_ratio_.sum(),
        4
    )
)


# ============================================================
# CELL 43 - VISUALISE K-MEANS CLUSTERS
# ============================================================

plt.figure(
    figsize=(12, 8)
)

plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=df["Cluster"],
    alpha=0.7
)

plt.xlabel(
    "SVD Component 1"
)

plt.ylabel(
    "SVD Component 2"
)

plt.title(
    "Nintendo 64 Games - K-Means Clustering"
)

plt.colorbar(
    label="Cluster"
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 44 - CALCULATE FINAL SILHOUETTE SCORE
# ============================================================

final_silhouette = silhouette_score(
    X,
    df["Cluster"]
)

print(
    "Final Silhouette Score:",
    round(
        final_silhouette,
        4
    )
)


# ============================================================
# CELL 45 - DISPLAY EXAMPLES FROM EACH CLUSTER
# ============================================================

for cluster_number in range(best_k):

    print("\n")
    print("=" * 80)
    print(
        f"CLUSTER {cluster_number}"
    )
    print("=" * 80)

    cluster_data = df[
        df["Cluster"] == cluster_number
    ]

    # Display up to 10 records
    print(
        cluster_data[
            text_columns
        ].head(10).to_string(
            index=False
        )
    )


# ============================================================
# CELL 46 - FIND CLOSEST GAMES TO EACH CLUSTER CENTRE
# ============================================================

# Calculate distances from every record
# to its assigned cluster centre.

distances = kmeans.transform(X)

df["Distance_to_Centroid"] = [
    distances[i, cluster]
    for i, cluster in enumerate(
        df["Cluster"]
    )
]


# ============================================================
# CELL 47 - MOST REPRESENTATIVE GAMES
# ============================================================

for cluster_number in range(best_k):

    print("\n")
    print("=" * 80)

    print(
        f"REPRESENTATIVE GAMES - CLUSTER {cluster_number}"
    )

    print("=" * 80)

    representative_games = (
        df[
            df["Cluster"]
            == cluster_number
        ]
        .sort_values(
            "Distance_to_Centroid"
        )
        .head(10)
    )

    print(
        representative_games[
            text_columns
        ].to_string(
            index=False
        )
    )


# ============================================================
# CELL 48 - CREATE FINAL DATASET
# ============================================================

final_columns = list(
    df.columns
)

print(
    "Final columns:"
)

print(
    final_columns
)


# ============================================================
# CELL 49 - SAVE CLUSTERED DATASET
# ============================================================

output_file = (
    "Nintendo_64_NLP_KMeans_Results.csv"
)

df.to_csv(
    output_file,
    index=False
)

print(
    f"Dataset saved successfully as: {output_file}"
)


# ============================================================
# CELL 50 - SAVE CLUSTER SUMMARY
# ============================================================

summary_file = (
    "Nintendo_64_Cluster_Summary.csv"
)

cluster_summary.to_csv(
    summary_file,
    index=False
)

print(
    f"Cluster summary saved as: {summary_file}"
)


# ============================================================
# CELL 51 - FINAL RESULTS
# ============================================================

print("=" * 70)
print("FINAL K-MEANS RESULTS")
print("=" * 70)

print(
    "Original number of records:",
    rows
)

print(
    "Records used:",
    len(df)
)

print(
    "Number of clusters:",
    best_k
)

print(
    "Silhouette Score:",
    round(
        final_silhouette,
        4
    )
)

print(
    "Number of TF-IDF features:",
    X.shape[1]
)

print("=" * 70)

print("\nCluster sizes:")

print(
    df["Cluster"]
    .value_counts()
    .sort_index()
)

print("\nTop keywords per cluster:")

for cluster_number in range(best_k):

    top_words = [
        terms[index]
        for index in
        order_centroids[
            cluster_number,
            :10
        ]
    ]

    print(
        f"\nCluster {cluster_number}:"
    )

    print(
        ", ".join(top_words)
    )


# ============================================================
# CELL 52 - DISPLAY FINAL DATASET
# ============================================================

df.head(20)
```
