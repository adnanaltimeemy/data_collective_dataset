```python
# ============================================================
# NOMINAL_INDEX.CSV
# NLP + TF-IDF + K-MEANS CLUSTERING
# COMPLETE JUPYTER NOTEBOOK CODE
# ============================================================


# ============================================================
# CELL 1 — INSTALL REQUIRED LIBRARIES
# ============================================================

# Run this cell once if the packages are not already installed.

# !pip install pandas numpy matplotlib seaborn scikit-learn nltk


# ============================================================
# CELL 2 — IMPORT LIBRARIES
# ============================================================

import os
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    calinski_harabasz_score,
    davies_bouldin_score
)
from sklearn.decomposition import TruncatedSVD

warnings.filterwarnings("ignore")

print("Libraries imported successfully.")


# ============================================================
# CELL 3 — DOWNLOAD NLTK RESOURCES
# ============================================================

nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

print("NLTK resources downloaded.")


# ============================================================
# CELL 4 — LOAD DATASET
# ============================================================

file_path = "nominal_index.csv"

if not os.path.exists(file_path):
    raise FileNotFoundError(
        f"Could not find '{file_path}'. "
        "Make sure the CSV file is in the same folder as your Jupyter Notebook."
    )

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print(f"Number of rows    : {df.shape[0]}")
print(f"Number of columns : {df.shape[1]}")


# ============================================================
# CELL 5 — DISPLAY DATASET
# ============================================================

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset information:")
display(df.info())

print("\nMissing values:")
display(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())


# ============================================================
# CELL 6 — REMOVE DUPLICATE ROWS
# ============================================================

before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after = len(df)

print(f"Rows before removing duplicates : {before}")
print(f"Rows after removing duplicates  : {after}")
print(f"Duplicates removed              : {before - after}")


# ============================================================
# CELL 7 — IDENTIFY TEXT COLUMNS
# ============================================================

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Detected text columns:")
print(text_columns)

if len(text_columns) == 0:
    raise ValueError(
        "No text columns were detected in the dataset. "
        "NLP requires at least one text column."
    )


# ============================================================
# CELL 8 — CREATE COMBINED TEXT COLUMN
# ============================================================

# Convert missing values to empty strings
df[text_columns] = df[text_columns].fillna("")

# Convert all text columns to strings
for column in text_columns:
    df[column] = df[column].astype(str)

# Combine all text columns
df["combined_text"] = df[text_columns].agg(
    " ".join,
    axis=1
)

print("Combined text created successfully.")

display(
    df[text_columns + ["combined_text"]].head()
)


# ============================================================
# CELL 9 — REMOVE EMPTY TEXT RECORDS
# ============================================================

df["combined_text"] = (
    df["combined_text"]
    .fillna("")
    .str.strip()
)

empty_text = df["combined_text"].eq("").sum()

print("Rows with empty text:", empty_text)

if empty_text > 0:
    df = df[df["combined_text"] != ""].reset_index(drop=True)

print("Dataset shape after removing empty text:")
print(df.shape)


# ============================================================
# CELL 10 — NLP PREPROCESSING
# ============================================================

stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()


def preprocess_text(text):
    """
    Clean and preprocess text for NLP.
    """

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )

    # Remove email addresses
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # Remove HTML tags
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )

    # Keep alphabetic characters
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    # Remove extra whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    # Tokenisation
    words = text.split()

    # Stopword removal + lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
        and len(word) > 2
    ]

    return " ".join(words)


# Apply preprocessing
df["clean_text"] = df["combined_text"].apply(
    preprocess_text
)

print("NLP preprocessing completed.")

display(
    df[
        ["combined_text", "clean_text"]
    ].head(10)
)


# ============================================================
# CELL 11 — REMOVE RECORDS WITH EMPTY CLEANED TEXT
# ============================================================

empty_clean_text = df["clean_text"].eq("").sum()

print(
    "Records with empty cleaned text:",
    empty_clean_text
)

if empty_clean_text > 0:

    df = df[
        df["clean_text"] != ""
    ].reset_index(drop=True)

print("Final dataset shape:")
print(df.shape)


# ============================================================
# CELL 12 — TF-IDF VECTORIZATION
# ============================================================

tfidf = TfidfVectorizer(
    lowercase=True,
    max_features=5000,
    min_df=1,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X = tfidf.fit_transform(
    df["clean_text"]
)

print("TF-IDF completed.")

print("Number of documents :", X.shape[0])
print("Number of features  :", X.shape[1])


# ============================================================
# CELL 13 — DISPLAY TF-IDF FEATURES
# ============================================================

feature_names = tfidf.get_feature_names_out()

print("First 50 TF-IDF features:")

print(
    feature_names[:50]
)


# ============================================================
# CELL 14 — DETERMINE VALID K VALUES
# ============================================================

n_samples = X.shape[0]

if n_samples < 3:
    raise ValueError(
        "At least 3 text records are required for K-Means clustering."
    )

# We do not want K >= number of records.
max_k = min(10, n_samples - 1)

if max_k < 2:
    raise ValueError(
        "Not enough records to evaluate multiple clusters."
    )

k_values = range(2, max_k + 1)

print(
    "K values that will be evaluated:",
    list(k_values)
)


# ============================================================
# CELL 15 — ELBOW METHOD
# ============================================================

inertia_values = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)

    inertia_values.append(
        model.inertia_
    )


plt.figure(figsize=(9, 6))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.xticks(
    list(k_values)
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 16 — SILHOUETTE SCORE
# ============================================================

silhouette_values = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        labels
    )

    silhouette_values.append(
        score
    )


plt.figure(figsize=(9, 6))

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score")

plt.xticks(
    list(k_values)
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 17 — COMPARE CLUSTERING RESULTS
# ============================================================

evaluation_results = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertia_values,
    "Silhouette_Score": silhouette_values
})

display(evaluation_results)


# ============================================================
# CELL 18 — SELECT BEST K
# ============================================================

best_index = np.argmax(
    silhouette_values
)

best_k = list(k_values)[best_index]

best_silhouette = silhouette_values[best_index]

print(
    f"Best K according to the Silhouette Score: {best_k}"
)

print(
    f"Best Silhouette Score: {best_silhouette:.4f}"
)


# ============================================================
# CELL 19 — TRAIN FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X)

df["cluster"] = cluster_labels

print("K-Means clustering completed.")


# ============================================================
# CELL 20 — CLUSTER DISTRIBUTION
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("Number of records in each cluster:")

display(
    cluster_counts.to_frame(
        name="Number_of_Records"
    )
)


# ============================================================
# CELL 21 — CLUSTER DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(9, 6))

sns.countplot(
    data=df,
    x="cluster"
)

plt.xlabel("Cluster")
plt.ylabel("Number of Records")
plt.title("Number of Records per Cluster")

plt.show()


# ============================================================
# CELL 22 — FINAL CLUSTERING METRICS
# ============================================================

final_silhouette = silhouette_score(
    X,
    df["cluster"]
)

final_calinski = calinski_harabasz_score(
    X.toarray(),
    df["cluster"]
)

final_davies = davies_bouldin_score(
    X.toarray(),
    df["cluster"]
)

print("Final Clustering Evaluation")
print("-" * 40)

print(
    f"Silhouette Score          : {final_silhouette:.4f}"
)

print(
    f"Calinski-Harabasz Score   : {final_calinski:.4f}"
)

print(
    f"Davies-Bouldin Score      : {final_davies:.4f}"
)


# ============================================================
# CELL 23 — TOP WORDS FOR EACH CLUSTER
# ============================================================

terms = np.array(
    tfidf.get_feature_names_out()
)

print("Top terms in each cluster")
print("=" * 70)

cluster_keywords = {}

for cluster_number in range(best_k):

    centroid = (
        kmeans.cluster_centers_[cluster_number]
    )

    top_indices = centroid.argsort()[
        ::-1
    ][:20]

    top_terms = terms[
        top_indices
    ]

    cluster_keywords[
        cluster_number
    ] = top_terms.tolist()

    print(
        f"\nCluster {cluster_number}"
    )

    print(
        ", ".join(top_terms)
    )


# ============================================================
# CELL 24 — CREATE CLUSTER KEYWORDS DATAFRAME
# ============================================================

keywords_df = pd.DataFrame(
    dict(
        (
            cluster,
            pd.Series(words)
        )
        for cluster, words
        in cluster_keywords.items()
    )
)

keywords_df.index.name = "Rank"

print("Cluster keywords:")

display(keywords_df)


# ============================================================
# CELL 25 — REDUCE TF-IDF DIMENSIONS FOR VISUALISATION
# ============================================================

# TruncatedSVD is appropriate for sparse TF-IDF matrices.

n_components = min(
    2,
    X.shape[1] - 1
)

if n_components < 2:
    raise ValueError(
        "There are not enough TF-IDF features for a 2D visualisation."
    )

svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = svd.fit_transform(X)

print(
    "Explained variance ratio:",
    svd.explained_variance_ratio_
)


# ============================================================
# CELL 26 — VISUALISE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=df["cluster"],
    cmap="viridis",
    alpha=0.75,
    s=50
)

plt.xlabel("SVD Component 1")
plt.ylabel("SVD Component 2")
plt.title("K-Means Clusters")

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 27 — SHOW SAMPLE RECORDS FROM EACH CLUSTER
# ============================================================

for cluster_number in range(best_k):

    print("\n")
    print("=" * 80)
    print(f"CLUSTER {cluster_number}")
    print("=" * 80)

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    display(
        cluster_data[
            text_columns + ["cluster"]
        ].head(10)
    )


# ============================================================
# CELL 28 — CREATE A CLUSTER SUMMARY
# ============================================================

cluster_summary = []

for cluster_number in range(best_k):

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    cluster_summary.append({
        "Cluster": cluster_number,
        "Number_of_Records": len(cluster_data),
        "Percentage": (
            len(cluster_data)
            / len(df)
            * 100
        ),
        "Top_Terms": ", ".join(
            cluster_keywords[
                cluster_number
            ][:10]
        )
    })

cluster_summary_df = pd.DataFrame(
    cluster_summary
)

display(
    cluster_summary_df
)


# ============================================================
# CELL 29 — ADD SVD COORDINATES TO DATASET
# ============================================================

df["SVD_1"] = X_2d[:, 0]
df["SVD_2"] = X_2d[:, 1]

display(
    df.head()
)


# ============================================================
# CELL 30 — SAVE FINAL CLUSTERED DATASET
# ============================================================

output_file = (
    "nominal_index_kmeans_clusters.csv"
)

df.to_csv(
    output_file,
    index=False
)

print(
    f"Final clustered dataset saved to: {output_file}"
)


# ============================================================
# CELL 31 — SAVE CLUSTER SUMMARY
# ============================================================

summary_file = (
    "nominal_index_cluster_summary.csv"
)

cluster_summary_df.to_csv(
    summary_file,
    index=False
)

print(
    f"Cluster summary saved to: {summary_file}"
)


# ============================================================
# CELL 32 — SAVE CLUSTER KEYWORDS
# ============================================================

keywords_file = (
    "nominal_index_cluster_keywords.csv"
)

keywords_df.to_csv(
    keywords_file
)

print(
    f"Cluster keywords saved to: {keywords_file}"
)


# ============================================================
# CELL 33 — FINAL REPORT
# ============================================================

print("\n")
print("=" * 70)
print("FINAL NLP + K-MEANS REPORT")
print("=" * 70)

print(f"Original dataset rows       : {before}")
print(f"Final dataset rows          : {len(df)}")
print(f"Text columns used           : {text_columns}")
print(f"TF-IDF features             : {X.shape[1]}")
print(f"Optimal number of clusters  : {best_k}")
print(f"Silhouette Score            : {final_silhouette:.4f}")
print(f"Calinski-Harabasz Score     : {final_calinski:.4f}")
print(f"Davies-Bouldin Score        : {final_davies:.4f}")

print("\nCluster sizes:")

for cluster_number, count in cluster_counts.items():

    percentage = (
        count / len(df) * 100
    )

    print(
        f"Cluster {cluster_number}: "
        f"{count} records "
        f"({percentage:.2f}%)"
    )

print("\nTop terms:")

for cluster_number in range(best_k):

    print(
        f"\nCluster {cluster_number}: "
        f"{', '.join(cluster_keywords[cluster_number][:10])}"
    )

print("\nOutput files:")
print("1. nominal_index_kmeans_clusters.csv")
print("2. nominal_index_cluster_summary.csv")
print("3. nominal_index_cluster_keywords.csv")

print("\nAnalysis completed successfully.")
```
