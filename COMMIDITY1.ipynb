```python
# 1. Import libraries
import pandas as pd
import re
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 2. Load dataset
df = pd.read_csv("commodity-prices.csv")
print(df.head())
print(df.info())

# 3. Find text column automatically
text_col = df.select_dtypes(include="object").columns[0]

# 4. Clean text
def clean_text(x):
    x = str(x).lower()
    x = re.sub(r'[^a-zA-Z\s]', '', x)
    return x

df["clean_text"] = df[text_col].apply(clean_text)

# 5. NLP - TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X_text = tfidf.fit_transform(df["clean_text"])

print("TF-IDF Shape:", X_text.shape)

# 6. K-Means clustering on NLP features
k = 3
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X_text)

# 7. Display clusters
print("\nClustered Data:")
print(df[[text_col, "Cluster"]].head(20))

# 8. PCA for visualisation
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_text.toarray())

plt.figure(figsize=(8,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=df["Cluster"], cmap="viridis")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("K-Means Clustering of Commodity Prices")
plt.show()

# 9. Cluster summary
print("\nCluster Counts:")
print(df["Cluster"].value_counts().sort_index())

# 10. Save results
df.to_csv("commodity-prices-clustered.csv", index=False)

print("\nDone! Results saved as commodity-prices-clustered.csv")
```
