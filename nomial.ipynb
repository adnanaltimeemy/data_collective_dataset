```python
# ============================================================
# NOMINAL INDEX DATASET
# NLP + TF-IDF + K-MEANS CLUSTERING
# ============================================================

# ============================================================
# 1. INSTALL REQUIRED PACKAGES
# ============================================================

# Run this cell if the packages are not already installed.
# Uncomment the following line:

# !pip install pandas numpy scikit-learn nltk matplotlib seaborn


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import re
import os
import warnings

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics import silhouette_score

import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer


# Suppress unnecessary warnings
warnings.filterwarnings("ignore")


# ============================================================
# 3. DOWNLOAD NLTK RESOURCES
# ============================================================

nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")


# ============================================================
# 4. LOAD DATASET
# ============================================================

FILE_NAME = "nominal_index.csv"

if not os.path.exists(FILE_NAME):
    raise FileNotFoundError(
        f"Could not find '{FILE_NAME}'. "
        "Please place the CSV file in the same folder as this notebook."
    )

df = pd.read_csv(FILE_NAME)

print("=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 5. DATASET INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())


# ============================================================
# 6. REMOVE DUPLICATE ROWS
# ============================================================

original_rows = len(df)

df = df.drop_duplicates().reset_index(drop=True)

removed_duplicates = original_rows - len(df)

print("\nDuplicates removed:", removed_duplicates)
print("Rows remaining:", len(df))


# ============================================================
# 7. IDENTIFY TEXT COLUMNS
# ============================================================

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("\n" + "=" * 70)
print("TEXT COLUMNS")
print("=" * 70)

print("Text columns detected:")
for column in text_columns:
    print("-", column)


# If there are no text columns, stop execution
if len(text_columns) == 0:
    raise ValueError(
        "No text columns were found in the dataset. "
        "The dataset needs at least one text/string column for NLP."
    )


# ============================================================
# 8. COMBINE TEXT COLUMNS
# ============================================================

df["combined_text"] = (
    df[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
)

print("\nCombined text created.")

display(
    df[["combined_text"]].head()
)


# ============================================================
# 9. NLP PREPROCESSING
# ============================================================

stop_words = set(
    stopwords.words("english")
)

lemmatizer = WordNetLemmatizer()


def clean_text(text):

    # Convert to string
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )

    # Remove email addresses
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # Remove numbers
    text = re.sub(
        r"\d+",
        " ",
        text
    )

    # Keep alphabetic characters only
    text = re.sub(
        r"[^a-z\s]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    # Split into words
    words = text.split()

    # Remove stopwords
    words = [
        word
        for word in words
        if word not in stop_words
    ]

    # Remove very short words
    words = [
        word
        for word in words
        if len(word) > 2
    ]

    # Lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]

    return " ".join(words)


print("\nCleaning text...")

df["clean_text"] = (
    df["combined_text"]
    .apply(clean_text)
)


# ============================================================
# 10. DISPLAY BEFORE AND AFTER NLP
# ============================================================

print("\n" + "=" * 70)
print("NLP PREPROCESSING EXAMPLES")
print("=" * 70)

display(
    df[
        ["combined_text", "clean_text"]
    ].head(10)
)


# ============================================================
# 11. REMOVE EMPTY TEXT RECORDS
# ============================================================

before_empty_removal = len(df)

df = df[
    df["clean_text"].str.strip() != ""
].reset_index(drop=True)

removed_empty = (
    before_empty_removal - len(df)
)

print(
    "\nRows removed because they contained "
    f"no usable text: {removed_empty}"
)

print("Rows available for clustering:", len(df))


# ============================================================
# 12. TF-IDF VECTORIZATION
# ============================================================

print("\n" + "=" * 70)
print("TF-IDF VECTORIZATION")
print("=" * 70)

tfidf = TfidfVectorizer(
    max_features=5000,
    min_df=1,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X = tfidf.fit_transform(
    df["clean_text"]
)

print("TF-IDF matrix shape:", X.shape)

print(
    "Number of TF-IDF features:",
    len(tfidf.get_feature_names_out())
)


# ============================================================
# 13. GET TF-IDF TERMS
# ============================================================

terms = np.array(
    tfidf.get_feature_names_out()
)

print("\nFirst 30 TF-IDF terms:")

print(
    terms[:30]
)


# ============================================================
# 14. TEST DIFFERENT K VALUES
# ============================================================

print("\n" + "=" * 70)
print("SEARCHING FOR BEST NUMBER OF CLUSTERS")
print("=" * 70)


# Maximum number of clusters to test
max_k = min(
    10,
    len(df) - 1
)

if max_k < 2:
    raise ValueError(
        "The dataset must contain at least 3 usable records "
        "to perform K-Means clustering."
    )


k_values = range(
    2,
    max_k + 1
)


inertia_values = []

silhouette_values = []


for k in k_values:

    print(
        f"Testing K = {k}..."
    )

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    inertia_values.append(
        model.inertia_
    )

    # Silhouette score requires more than one cluster
    score = silhouette_score(
        X,
        labels
    )

    silhouette_values.append(
        score
    )


# ============================================================
# 15. CREATE K-MEANS EVALUATION TABLE
# ============================================================

k_results = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertia_values,
    "Silhouette Score": silhouette_values
})

print("\nK-Means evaluation results:")

display(k_results)


# ============================================================
# 16. FIND BEST K
# ============================================================

best_index = np.argmax(
    silhouette_values
)

best_k = list(k_values)[
    best_index
]

best_silhouette = silhouette_values[
    best_index
]

print("\n" + "=" * 70)
print("BEST K")
print("=" * 70)

print(
    "Recommended number of clusters:",
    best_k
)

print(
    "Best silhouette score:",
    round(best_silhouette, 4)
)


# ============================================================
# 17. ELBOW METHOD GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.title(
    "K-Means Elbow Method"
)

plt.xticks(
    list(k_values)
)

plt.grid(
    True,
    alpha=0.3
)

plt.show()


# ============================================================
# 18. SILHOUETTE SCORE GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.axvline(
    best_k,
    linestyle="--",
    label=f"Best K = {best_k}"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "Silhouette Score for Different K Values"
)

plt.xticks(
    list(k_values)
)

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.show()


# ============================================================
# 19. FINAL K-MEANS MODEL
# ============================================================

print("\n" + "=" * 70)
print("TRAINING FINAL K-MEANS MODEL")
print("=" * 70)

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(
    X
)

# Add cluster labels to dataset
df["cluster"] = cluster_labels

print(
    "K-Means clustering completed successfully."
)


# ============================================================
# 20. CLUSTER COUNTS
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("\n" + "=" * 70)
print("RECORDS PER CLUSTER")
print("=" * 70)

display(
    cluster_counts.to_frame(
        name="Number of Records"
    )
)


# ============================================================
# 21. CLUSTER DISTRIBUTION GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

sns.barplot(
    x=cluster_counts.index,
    y=cluster_counts.values
)

plt.xlabel(
    "Cluster"
)

plt.ylabel(
    "Number of Records"
)

plt.title(
    "Number of Records in Each Cluster"
)

plt.show()


# ============================================================
# 22. FIND TOP TERMS FOR EACH CLUSTER
# ============================================================

print("\n" + "=" * 70)
print("TOP TERMS IN EACH CLUSTER")
print("=" * 70)


cluster_keywords = []


for cluster_number in range(best_k):

    # Get rows belonging to cluster
    indices = np.where(
        df["cluster"].values
        == cluster_number
    )[0]

    # Average TF-IDF values
    cluster_matrix = X[
        indices
    ]

    mean_tfidf = np.asarray(
        cluster_matrix.mean(axis=0)
    ).flatten()

    # Sort terms by TF-IDF score
    top_indices = (
        mean_tfidf
        .argsort()[-20:][::-1]
    )

    top_terms = terms[
        top_indices
    ]

    print(
        f"\nCluster {cluster_number}"
    )

    print(
        ", ".join(top_terms)
    )

    cluster_keywords.append({
        "Cluster": cluster_number,
        "Number of Records": len(indices),
        "Top Terms": ", ".join(top_terms)
    })


# ============================================================
# 23. CREATE CLUSTER SUMMARY TABLE
# ============================================================

cluster_summary = pd.DataFrame(
    cluster_keywords
)

print("\n" + "=" * 70)
print("CLUSTER SUMMARY")
print("=" * 70)

display(
    cluster_summary
)


# ============================================================
# 24. REDUCE TF-IDF DIMENSIONS FOR VISUALIZATION
# ============================================================

print("\n" + "=" * 70)
print("CREATING 2-D CLUSTER VISUALIZATION")
print("=" * 70)


svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = svd.fit_transform(
    X
)


# Create visualization dataframe
visualization_df = pd.DataFrame({
    "Component_1": X_2d[:, 0],
    "Component_2": X_2d[:, 1],
    "Cluster": df["cluster"].values
})


# ============================================================
# 25. PLOT K-MEANS CLUSTERS
# ============================================================

plt.figure(
    figsize=(12, 8)
)

sns.scatterplot(
    data=visualization_df,
    x="Component_1",
    y="Component_2",
    hue="Cluster",
    palette="tab10",
    s=80,
    alpha=0.8
)

plt.title(
    "K-Means Clusters Based on NLP / TF-IDF"
)

plt.xlabel(
    "SVD Component 1"
)

plt.ylabel(
    "SVD Component 2"
)

plt.legend(
    title="Cluster",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()


# ============================================================
# 26. DISPLAY SAMPLE RECORDS FROM EACH CLUSTER
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE RECORDS FROM EACH CLUSTER")
print("=" * 70)


for cluster_number in range(best_k):

    print(
        f"\n{'=' * 70}"
    )

    print(
        f"CLUSTER {cluster_number}"
    )

    print(
        f"{'=' * 70}"
    )

    cluster_data = df[
        df["cluster"]
        == cluster_number
    ]

    # Show up to 10 records
    display(
        cluster_data[
            text_columns
            + ["cluster"]
        ].head(10)
    )


# ============================================================
# 27. FIND CLUSTER CENTROIDS AND THEIR TOP TERMS
# ============================================================

print("\n" + "=" * 70)
print("CLUSTER CENTROID TERMS")
print("=" * 70)


centroid_keywords = []


for cluster_number in range(best_k):

    centroid = kmeans.cluster_centers_[
        cluster_number
    ]

    top_indices = (
        centroid
        .argsort()[-20:][::-1]
    )

    top_terms = terms[
        top_indices
    ]

    print(
        f"\nCluster {cluster_number}:"
    )

    print(
        ", ".join(top_terms)
    )

    centroid_keywords.append({
        "Cluster": cluster_number,
        "Top Centroid Terms": ", ".join(top_terms)
    })


centroid_summary = pd.DataFrame(
    centroid_keywords
)

display(
    centroid_summary
)


# ============================================================
# 28. COMBINE CLUSTER SUMMARIES
# ============================================================

final_cluster_summary = (
    cluster_summary
    .merge(
        centroid_summary,
        on="Cluster",
        how="left"
    )
)

print("\nFinal cluster summary:")

display(
    final_cluster_summary
)


# ============================================================
# 29. ADD 2-D COORDINATES TO DATASET
# ============================================================

df["SVD_Component_1"] = X_2d[:, 0]

df["SVD_Component_2"] = X_2d[:, 1]


# ============================================================
# 30. REORDER COLUMNS
# ============================================================

# Put cluster information near the beginning/end
# without deleting any original columns.

original_columns = [
    column
    for column in df.columns
    if column not in [
        "cluster",
        "SVD_Component_1",
        "SVD_Component_2"
    ]
]

df = df[
    original_columns
    + [
        "cluster",
        "SVD_Component_1",
        "SVD_Component_2"
    ]
]


# ============================================================
# 31. SAVE CLUSTERED DATASET
# ============================================================

OUTPUT_FILE = (
    "nominal_index_clustered.csv"
)

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\n" + "=" * 70)
print("DATASET SAVED")
print("=" * 70)

print(
    f"Output file: {OUTPUT_FILE}"
)


# ============================================================
# 32. SAVE CLUSTER SUMMARY
# ============================================================

SUMMARY_FILE = (
    "nominal_index_cluster_summary.csv"
)

final_cluster_summary.to_csv(
    SUMMARY_FILE,
    index=False
)

print(
    f"Cluster summary: {SUMMARY_FILE}"
)


# ============================================================
# 33. SAVE K EVALUATION RESULTS
# ============================================================

K_RESULTS_FILE = (
    "nominal_index_k_evaluation.csv"
)

k_results.to_csv(
    K_RESULTS_FILE,
    index=False
)

print(
    f"K evaluation: {K_RESULTS_FILE}"
)


# ============================================================
# 34. FINAL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(
    f"Original dataset rows: {original_rows}"
)

print(
    f"Rows used for clustering: {len(df)}"
)

print(
    f"Number of text columns: {len(text_columns)}"
)

print(
    f"TF-IDF features: {X.shape[1]}"
)

print(
    f"Optimal K: {best_k}"
)

print(
    f"Silhouette score: {best_silhouette:.4f}"
)

print(
    "\nCluster sizes:"
)

display(
    cluster_counts.to_frame(
        name="Records"
    )
)

print(
    "\nTop terms by cluster:"
)

display(
    final_cluster_summary
)

print("\nFiles created:")
print("-", OUTPUT_FILE)
print("-", SUMMARY_FILE)
print("-", K_RESULTS_FILE)

print("\nDone!")
```
