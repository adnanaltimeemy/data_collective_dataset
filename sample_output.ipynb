```python
# NLP + K-Means Clustering

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 1. Load dataset
df = pd.read_csv("sample_output.csv")
print(df.head())
print(df.columns)

# 2. Select text column
# CHANGE 'text' to your actual text column name
text_col = "text"
df[text_col] = df[text_col].fillna("").astype(str)

# 3. NLP - TF-IDF
vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=3000
)
X = vectorizer.fit_transform(df[text_col])

# 4. K-Means
k = 4
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# 5. Show results
print("\nCluster counts:")
print(df["Cluster"].value_counts())

print("\nSample results:")
print(df[[text_col, "Cluster"]].head(10))

# 6. PCA visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.toarray())

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("NLP K-Means Clustering")
plt.show()

# 7. Save output
df.to_csv("sample_output_clustered.csv", index=False)

print("Done! Saved as sample_output_clustered.csv")
```
