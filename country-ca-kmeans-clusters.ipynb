```python
# File name
file_name = "country-ca.csv"

# Load dataset
df = pd.read_csv(file_name)

print("Dataset loaded successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
print("Dataset information:")
df.info()

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Find columns containing text
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Text columns found:")
print(text_columns)

if len(text_columns) == 0:
    raise ValueError(
        "No text column was found. "
        "This dataset does not appear to contain text data."
    )

# Calculate average text length for each text column
average_lengths = {}

for column in text_columns:
    average_lengths[column] = (
        df[column]
        .fillna("")
        .astype(str)
        .str.len()
        .mean()
    )

print("\nAverage text length:")
for column, length in average_lengths.items():
    print(f"{column}: {length:.2f}")

# Select the column containing the most text
text_column = max(
    average_lengths,
    key=average_lengths.get
)

print("\nSelected text column:", text_column)
```
