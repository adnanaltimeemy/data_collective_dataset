```python
# ============================================================
# NLP + K-MEANS CLUSTERING FOR benchmark_df.csv
# Complete Jupyter Notebook Code
# ============================================================

# ------------------------------------------------------------
# 1. INSTALL / IMPORT LIBRARIES
# ------------------------------------------------------------

# If packages are missing, uncomment the following line:
# !pip install pandas numpy matplotlib scikit-learn nltk seaborn

import os
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# 2. DOWNLOAD NLTK RESOURCES
# ------------------------------------------------------------

nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)


# ------------------------------------------------------------
# 3. LOAD DATASET
# ------------------------------------------------------------

FILE_PATH = "benchmark_df.csv"

if not os.path.exists(FILE_PATH):
    raise FileNotFoundError(
        f"Could not find '{FILE_PATH}'. "
        "Make sure the CSV file is in the same folder as this notebook."
    )

df = pd.read_csv(FILE_PATH)

print("=" * 70)
print("DATASET LOADED")
print("=" * 70)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]:,}")

print("\nColumn names:")
for i, column in enumerate(df.columns):
    print(f"{i}: {column}")


# ------------------------------------------------------------
# 4. BASIC DATASET INSPECTION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FIRST 5 ROWS")
print("=" * 70)

display(df.head())


print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

display(df.dtypes)


print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

missing_values = df.isnull().sum().sort_values(ascending=False)
display(missing_values)


print("\n" + "=" * 70)
print("DUPLICATES")
print("=" * 70)

print("Duplicate rows:", df.duplicated().sum())


# ------------------------------------------------------------
# 5. FIND TEXT COLUMNS
# ------------------------------------------------------------

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("\n" + "=" * 70)
print("TEXT COLUMNS")
print("=" * 70)

print(text_columns)


# ------------------------------------------------------------
# 6. SELECT TEXT COLUMN
# ------------------------------------------------------------
#
# IMPORTANT:
# If your dataset has a specific column containing the text
# that you want to cluster, change TEXT_COLUMN below.
#
# Examples:
#
# TEXT_COLUMN = "question"
# TEXT_COLUMN = "text"
# TEXT_COLUMN = "description"
# TEXT_COLUMN = "benchmark"
#
# If you leave it as None, the code will automatically choose
# the text column with the largest average text length.
# ------------------------------------------------------------

TEXT_COLUMN = None


if TEXT_COLUMN is None:

    if len(text_columns) == 0:
        raise ValueError(
            "No text columns were found in the dataset."
        )

    # Calculate average string length for every text column
    average_lengths = {}

    for column in text_columns:
        average_lengths[column] = (
            df[column]
            .fillna("")
            .astype(str)
            .str.len()
            .mean()
        )

    TEXT_COLUMN = max(
        average_lengths,
        key=average_lengths.get
    )


if TEXT_COLUMN not in df.columns:
    raise ValueError(
        f"TEXT_COLUMN '{TEXT_COLUMN}' does not exist in the dataset."
    )


print("\n" + "=" * 70)
print("SELECTED TEXT COLUMN")
print("=" * 70)

print("Using:", TEXT_COLUMN)


# ------------------------------------------------------------
# 7. REMOVE MISSING TEXT
# ------------------------------------------------------------

df = df.copy()

df[TEXT_COLUMN] = df[TEXT_COLUMN].fillna("").astype(str)

# Remove completely empty text rows
df = df[
    df[TEXT_COLUMN].str.strip() != ""
].copy()

df.reset_index(drop=True, inplace=True)

print("\nRows remaining:", len(df))


# ------------------------------------------------------------
# 8. REMOVE DUPLICATE TEXT
# ------------------------------------------------------------

before_duplicates = len(df)

df = df.drop_duplicates(
    subset=[TEXT_COLUMN]
).reset_index(drop=True)

after_duplicates = len(df)

print(
    f"Removed {before_duplicates - after_duplicates:,} "
    "duplicate text rows."
)


# ------------------------------------------------------------
# 9. NLP PREPROCESSING
# ------------------------------------------------------------

stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()


def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )

    # Remove email addresses
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # Remove HTML tags
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )

    # Keep only alphabetic characters
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    # Remove extra whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    # Tokenize
    words = text.split()

    # Remove stopwords
    words = [
        word
        for word in words
        if word not in stop_words
    ]

    # Remove very short words
    words = [
        word
        for word in words
        if len(word) > 2
    ]

    # Lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]

    return " ".join(words)


print("\n" + "=" * 70)
print("RUNNING NLP PREPROCESSING")
print("=" * 70)

df["clean_text"] = df[TEXT_COLUMN].apply(clean_text)

print("NLP preprocessing completed.")


# ------------------------------------------------------------
# 10. DISPLAY ORIGINAL VS CLEANED TEXT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ORIGINAL VS CLEANED TEXT")
print("=" * 70)

display(
    df[
        [TEXT_COLUMN, "clean_text"]
    ].head(10)
)


# ------------------------------------------------------------
# 11. REMOVE EMPTY CLEANED TEXT
# ------------------------------------------------------------

df = df[
    df["clean_text"].str.strip() != ""
].copy()

df.reset_index(drop=True, inplace=True)

print(
    "\nRows after removing empty cleaned text:",
    len(df)
)


# ------------------------------------------------------------
# 12. TF-IDF VECTORIZATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TF-IDF VECTORIZATION")
print("=" * 70)

tfidf = TfidfVectorizer(
    max_features=5000,
    min_df=2,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X = tfidf.fit_transform(
    df["clean_text"]
)

print("TF-IDF matrix shape:", X.shape)

print(
    "Number of TF-IDF features:",
    len(tfidf.get_feature_names_out())
)


# ------------------------------------------------------------
# 13. FIND OPTIMAL NUMBER OF CLUSTERS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SEARCHING FOR OPTIMAL NUMBER OF CLUSTERS")
print("=" * 70)

# Maximum K cannot be larger than the number of samples
MAX_K = min(10, X.shape[0] - 1)

if MAX_K < 2:
    raise ValueError(
        "Not enough text records to perform clustering."
    )

K_VALUES = range(2, MAX_K + 1)

inertias = []
silhouette_scores = []


for k in K_VALUES:

    print(f"Testing K = {k}")

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    inertia = model.inertia_

    silhouette = silhouette_score(
        X,
        labels
    )

    inertias.append(inertia)
    silhouette_scores.append(silhouette)


# ------------------------------------------------------------
# 14. ELBOW PLOT
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    list(K_VALUES),
    inertias,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.title(
    "Elbow Method for K-Means"
)

plt.xticks(
    list(K_VALUES)
)

plt.grid(
    alpha=0.3
)

plt.show()


# ------------------------------------------------------------
# 15. SILHOUETTE SCORE PLOT
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    list(K_VALUES),
    silhouette_scores,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "Silhouette Score for Different K Values"
)

plt.xticks(
    list(K_VALUES)
)

plt.grid(
    alpha=0.3
)

plt.show()


# ------------------------------------------------------------
# 16. SELECT BEST K
# ------------------------------------------------------------

best_index = np.argmax(
    silhouette_scores
)

BEST_K = list(K_VALUES)[best_index]

BEST_SILHOUETTE = silhouette_scores[best_index]

print("\n" + "=" * 70)
print("BEST K")
print("=" * 70)

print("Best number of clusters:", BEST_K)

print(
    "Best silhouette score:",
    round(BEST_SILHOUETTE, 4)
)


# ------------------------------------------------------------
# 17. TRAIN FINAL K-MEANS MODEL
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TRAINING FINAL K-MEANS MODEL")
print("=" * 70)

kmeans = KMeans(
    n_clusters=BEST_K,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X)

df["cluster"] = cluster_labels

print("K-Means training completed.")


# ------------------------------------------------------------
# 18. CLUSTER DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CLUSTER DISTRIBUTION")
print("=" * 70)

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

display(
    cluster_counts.to_frame(
        name="number_of_records"
    )
)


# ------------------------------------------------------------
# 19. CLUSTER DISTRIBUTION PLOT
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.xlabel(
    "Cluster"
)

plt.ylabel(
    "Number of Records"
)

plt.title(
    "Number of Records in Each Cluster"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()


# ------------------------------------------------------------
# 20. FIND TOP WORDS FOR EACH CLUSTER
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP WORDS FOR EACH CLUSTER")
print("=" * 70)

feature_names = (
    tfidf
    .get_feature_names_out()
)

order_centroids = (
    kmeans
    .cluster_centers_
    .argsort()[:, ::-1]
)

TOP_N_WORDS = 20

cluster_keywords = {}

for cluster_number in range(BEST_K):

    top_indices = (
        order_centroids[
            cluster_number,
            :TOP_N_WORDS
        ]
    )

    top_words = [
        feature_names[index]
        for index in top_indices
    ]

    cluster_keywords[
        cluster_number
    ] = top_words

    print(
        f"\nCluster {cluster_number}:"
    )

    print(
        ", ".join(top_words)
    )


# ------------------------------------------------------------
# 21. CREATE CLUSTER SUMMARY
# ------------------------------------------------------------

cluster_summary = []

for cluster_number in range(BEST_K):

    cluster_size = (
        df["cluster"] == cluster_number
    ).sum()

    keywords = ", ".join(
        cluster_keywords[
            cluster_number
        ]
    )

    cluster_summary.append(
        {
            "cluster": cluster_number,
            "number_of_records": cluster_size,
            "top_keywords": keywords
        }
    )


cluster_summary = pd.DataFrame(
    cluster_summary
)

print("\n" + "=" * 70)
print("CLUSTER SUMMARY")
print("=" * 70)

display(cluster_summary)


# ------------------------------------------------------------
# 22. SHOW SAMPLE RECORDS FROM EACH CLUSTER
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SAMPLE RECORDS BY CLUSTER")
print("=" * 70)

SAMPLES_PER_CLUSTER = 5

for cluster_number in range(BEST_K):

    print("\n")
    print("#" * 70)
    print(
        f"CLUSTER {cluster_number}"
    )
    print("#" * 70)

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    sample_data = cluster_data.sample(
        n=min(
            SAMPLES_PER_CLUSTER,
            len(cluster_data)
        ),
        random_state=42
    )

    display(
        sample_data[
            [TEXT_COLUMN, "clean_text", "cluster"]
        ]
    )


# ------------------------------------------------------------
# 23. PCA VISUALIZATION
# ------------------------------------------------------------
#
# For large datasets, converting the entire sparse matrix
# to dense can use a lot of memory. Therefore we first reduce
# the TF-IDF matrix using TruncatedSVD.
# ------------------------------------------------------------

from sklearn.decomposition import TruncatedSVD

print("\n" + "=" * 70)
print("CREATING 2D CLUSTER VISUALIZATION")
print("=" * 70)

if X.shape[1] >= 2:

    svd = TruncatedSVD(
        n_components=2,
        random_state=42
    )

    X_2d = svd.fit_transform(X)

else:

    X_2d = X.toarray()


plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=df["cluster"],
    alpha=0.6
)

plt.xlabel(
    "Component 1"
)

plt.ylabel(
    "Component 2"
)

plt.title(
    "K-Means Clusters Based on NLP / TF-IDF"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(
    alpha=0.2
)

plt.show()


# ------------------------------------------------------------
# 24. CALCULATE FINAL SILHOUETTE SCORE
# ------------------------------------------------------------

final_silhouette = silhouette_score(
    X,
    df["cluster"]
)

print("\n" + "=" * 70)
print("FINAL MODEL PERFORMANCE")
print("=" * 70)

print(
    f"Number of clusters : {BEST_K}"
)

print(
    f"Silhouette score   : {final_silhouette:.4f}"
)

print(
    f"Number of records  : {len(df):,}"
)

print(
    f"TF-IDF features    : {X.shape[1]:,}"
)


# ------------------------------------------------------------
# 25. ADD CLUSTER LABEL / KEYWORDS
# ------------------------------------------------------------

df["cluster_keywords"] = df[
    "cluster"
].map(
    lambda x: ", ".join(
        cluster_keywords[x][:10]
    )
)


# ------------------------------------------------------------
# 26. CREATE FINAL OUTPUT DATASET
# ------------------------------------------------------------

OUTPUT_FILE = (
    "benchmark_df_clustered.csv"
)

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\n" + "=" * 70)
print("OUTPUT SAVED")
print("=" * 70)

print(
    f"File: {OUTPUT_FILE}"
)


# ------------------------------------------------------------
# 27. SAVE CLUSTER SUMMARY
# ------------------------------------------------------------

SUMMARY_FILE = (
    "benchmark_cluster_summary.csv"
)

cluster_summary.to_csv(
    SUMMARY_FILE,
    index=False
)

print(
    f"Cluster summary: {SUMMARY_FILE}"
)


# ------------------------------------------------------------
# 28. FINAL OUTPUT PREVIEW
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL DATASET")
print("=" * 70)

print(
    "Shape:",
    df.shape
)

display(
    df.head(10)
)


# ------------------------------------------------------------
# 29. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ANALYSIS COMPLETE")
print("=" * 70)

print(
    f"""
Input file:
    {FILE_PATH}

Text column:
    {TEXT_COLUMN}

Number of records:
    {len(df):,}

TF-IDF features:
    {X.shape[1]:,}

Optimal K:
    {BEST_K}

Silhouette score:
    {final_silhouette:.4f}

Output files:
    {OUTPUT_FILE}
    {SUMMARY_FILE}

The 'cluster' column contains the K-Means cluster
assigned to every benchmark record.

The 'cluster_keywords' column contains the main
TF-IDF keywords associated with that cluster.
"""
)
```
