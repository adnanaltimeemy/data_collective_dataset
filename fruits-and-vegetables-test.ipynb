```python
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("fruits-and-vegetables-test.csv")

# Select text column (change 'name' if needed)
text = df["name"].astype(str)

# NLP cleaning
text = text.apply(lambda x: re.sub(r'[^a-zA-Z\s]', '', x.lower()))

# TF-IDF
vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(text)

# K-Means
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# Show results
print(df[["name", "Cluster"]])

# Visualisation
plt.scatter(X.toarray()[:, 0], X.toarray()[:, 1], c=df["Cluster"])
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Fruits and Vegetables - K-Means")
plt.show()

# Save results
df.to_csv("fruits-vegetables-clusters.csv", index=False)
```
