```python
# ==========================================
# K-Means Clustering + NLP
# Dataset: fairpoor.csv
# ==========================================

# Import Libraries

import pandas as pd
import numpy as np

# NLP Libraries
import nltk
import re

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Machine Learning Libraries
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# Visualisation
import matplotlib.pyplot as plt
import seaborn as sns


# Download NLP resources
nltk.download('stopwords')


# ==========================================
# Load Dataset
# ==========================================

df = pd.read_csv("fairpoor.csv")

# Display first rows
df.head()
```

```python
# Dataset shape

print(df.shape)


# Column names

df.columns


# Dataset information

df.info()


# Missing values

df.isnull().sum()
```

```python
# Remove duplicate rows

df = df.drop_duplicates()


# Fill missing values

df = df.fillna("")


df.head()
```

```python
# Display object columns

text_columns = df.select_dtypes(include=['object']).columns

print(text_columns)
```

```python
# Combine all text columns into one column

df['text_data'] = df[text_columns].astype(str).apply(
    lambda x: ' '.join(x),
    axis=1
)


df[['text_data']].head()
```

```python
# Stopwords

stop_words = set(stopwords.words('english'))

# Stemmer

stemmer = PorterStemmer()


def clean_text(text):
    
    # Lowercase
    text = text.lower()
    
    # Remove punctuation
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    
    # Tokenization
    words = text.split()
    
    # Remove stopwords and stemming
    words = [
        stemmer.stem(word)
        for word in words
        if word not in stop_words
    ]
    
    return " ".join(words)



# Apply NLP cleaning

df['clean_text'] = df['text_data'].apply(clean_text)


df[['text_data','clean_text']].head()
```

```python
df.to_csv(
    "fairpoor_kmeans_clusters.csv",
    index=False
)

print("File saved successfully!")
```
