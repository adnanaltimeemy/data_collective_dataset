```python
# 1. Import libraries
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 2. Load dataset
df = pd.read_csv("superheroes_info.csv")
print(df.head())

# 3. Combine text columns
text_cols = df.select_dtypes(include="object").columns
df["text"] = df[text_cols].fillna("").agg(" ".join, axis=1)

# 4. Clean text
def clean(x):
    x = x.lower()
    x = re.sub(r"[^a-z\s]", "", x)
    return x

df["text"] = df["text"].apply(clean)

# 5. NLP - TF-IDF
tfidf = TfidfVectorizer(stop_words="english", max_features=1000)
X = tfidf.fit_transform(df["text"])

# 6. K-Means clustering
k = 5
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# 7. Show results
print(df[["text", "Cluster"]].head(20))

# 8. Cluster sizes
print("\nCluster counts:")
print(df["Cluster"].value_counts())

# 9. Top words in each cluster
words = tfidf.get_feature_names_out()

for i in range(k):
    top = model.cluster_centers_[i].argsort()[-10:][::-1]
    print(f"\nCluster {i}:")
    print(", ".join(words[j] for j in top))

# 10. Save results
df.to_csv("superheroes_clustered.csv", index=False)

print("\nDone! File saved as superheroes_clustered.csv")
```
