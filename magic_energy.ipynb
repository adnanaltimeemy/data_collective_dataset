```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA

# 2. Load dataset
df = pd.read_csv("magic_energy.csv")
print(df.head())

# 3. Numerical data for K-Means
num = df.select_dtypes(include="number").fillna(0)

# Scale data
X = StandardScaler().fit_transform(num)

# 4. K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

print("\nK-Means Clusters:")
print(df["Cluster"].value_counts())

# 5. PCA visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.scatter(X_pca[:,0], X_pca[:,1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("K-Means Clustering")
plt.show()

# 6. NLP - use first text column
text_cols = df.select_dtypes(include="object").columns

if len(text_cols) > 0:
    text = df[text_cols[0]].fillna("")

    tfidf = TfidfVectorizer(stop_words="english")
    X_text = tfidf.fit_transform(text)

    nlp = KMeans(n_clusters=3, random_state=42, n_init=10)
    df["NLP_Cluster"] = nlp.fit_predict(X_text)

    print("\nNLP Clusters:")
    print(df["NLP_Cluster"].value_counts())

# 7. Save results
df.to_csv("magic_energy_clustered.csv", index=False)

print("\nDone! File saved as magic_energy_clustered.csv")
```
