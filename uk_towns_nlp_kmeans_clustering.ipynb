```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.cluster import KMeans

# 0. DEFINE YOUR DATA
text_data = [
    "I love machine learning and data science",
    "KMeans clustering is useful for grouping data",
    "Python is great for AI and machine learning",
    "Football and sports are popular worldwide",
    "The match was exciting and the team played well",
    "Data science involves statistics and programming",
    "AI and deep learning are parts of machine learning",
    "Sports events bring fans together"
]

# 1. Vectorize text
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(text_data)

# 2. KMeans
num_clusters = 2
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
kmeans.fit(X)

# 3. Get words
terms = vectorizer.get_feature_names_out()

# 4. Sort cluster centers
order_centroids = kmeans.cluster_centers_.argsort()[:, ::-1]

# 5. Print clusters
for i in range(num_clusters):
    print(f"\nCluster {i}:")
    top_words = [terms[ind] for ind in order_centroids[i, :10]]
    print(", ".join(top_words))
```
