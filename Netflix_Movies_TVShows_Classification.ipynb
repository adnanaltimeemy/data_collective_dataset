```python
target = "type"  # Movie / TV Show
```

```python
df = df.dropna(subset=[target])
```

```python
label_encoders = {}

for col in df.columns:
    if df[col].dtype == "object":
        le = LabelEncoder()
        df[col] = df[col].astype(str)
        df[col] = le.fit_transform(df[col])
        label_encoders[col] = le
```

```python
X = df.drop(columns=[target])
y = df[target]
```
