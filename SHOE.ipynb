````python
```python
# 1. Import libraries
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# 2. Load dataset
df = pd.read_csv("shoes_data_small.csv")

# 3. Select text column
text_col = "description"
df[text_col] = df[text_col].fillna("").astype(str)

# 4. NLP text cleaning
def clean(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    return text

df["clean_text"] = df[text_col].apply(clean)

# 5. TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(df["clean_text"])

# 6. K-Means
k = 4
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["cluster"] = model.fit_predict(X)

# 7. Show results
print(df[[text_col, "cluster"]].head(20))

# 8. PCA visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.toarray())

plt.scatter(X_pca[:,0], X_pca[:,1], c=df["cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Shoes K-Means Clustering")
plt.show()

# 9. Save results
df.to_csv("shoes_clustered.csv", index=False)
```
````
