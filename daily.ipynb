```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD

# Load dataset
df = pd.read_csv("daily.csv")
print(df.head())
print(df.columns)

# Select text column (change 'text' to your actual column name)
df = df.dropna(subset=["text"]).copy()

# NLP: convert text to TF-IDF features
vectorizer = TfidfVectorizer(stop_words="english", max_features=3000)
X = vectorizer.fit_transform(df["text"].astype(str))

# K-Means clustering
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X)

# Display top words per cluster
terms = vectorizer.get_feature_names_out()
for i, center in enumerate(kmeans.cluster_centers_):
    top_words = center.argsort()[-10:][::-1]
    print(f"Cluster {i}:",
          ", ".join(terms[j] for j in top_words))

# Visualise clusters
X_2d = TruncatedSVD(n_components=2, random_state=42).fit_transform(X)
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=df["cluster"], cmap="viridis")
plt.title("NLP + K-Means Clustering")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.show()

# Save results
df.to_csv("daily_clustered.csv", index=False)
print("Saved daily_clustered.csv")
```
