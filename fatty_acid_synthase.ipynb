```python
# ================================================================
# FATTY ACID SYNTHASE DATASET
# NLP + K-MEANS CLUSTERING
# ================================================================
#
# Input:
#     fatty_acid_synthase_val.csv
#
# Output:
#     fatty_acid_synthase_kmeans_results.csv
#
# Methods:
#     1. Data loading and inspection
#     2. Missing-value handling
#     3. Text preprocessing
#     4. TF-IDF feature extraction
#     5. Elbow method
#     6. Silhouette analysis
#     7. K-Means clustering
#     8. PCA visualization
#     9. Cluster interpretation
#    10. Save clustered dataset
#
# ================================================================


# ================================================================
# 1. INSTALL / IMPORT LIBRARIES
# ================================================================

# If required, uncomment and run:
# !pip install pandas numpy matplotlib seaborn scikit-learn nltk

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import re
import warnings

from nltk.corpus import stopwords
import nltk

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

warnings.filterwarnings("ignore")


# ================================================================
# 2. DOWNLOAD NLTK RESOURCES
# ================================================================

try:
    stop_words = set(stopwords.words("english"))
except LookupError:
    nltk.download("stopwords")
    stop_words = set(stopwords.words("english"))

print("Libraries loaded successfully.")


# ================================================================
# 3. LOAD DATASET
# ================================================================

FILE_NAME = "fatty_acid_synthase_val.csv"

try:
    df = pd.read_csv(FILE_NAME)
except FileNotFoundError:
    raise FileNotFoundError(
        f"\nCould not find '{FILE_NAME}'.\n"
        "Make sure the CSV file is uploaded to the same "
        "Jupyter Notebook directory."
    )

print("=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())


# ================================================================
# 4. DATASET INFORMATION
# ================================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nColumn names:")
for i, column in enumerate(df.columns):
    print(f"{i}: {column}")

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())


# ================================================================
# 5. REMOVE DUPLICATE ROWS
# ================================================================

original_rows = len(df)

df = df.drop_duplicates().reset_index(drop=True)

removed_duplicates = original_rows - len(df)

print("\nRemoved duplicate rows:", removed_duplicates)
print("Remaining rows:", len(df))


# ================================================================
# 6. IDENTIFY TEXT COLUMN
# ================================================================

# Find columns containing text/string data
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("\nPossible text columns:")
print(text_columns)


# ------------------------------------------------
# Try to automatically find the most appropriate
# text column.
# ------------------------------------------------

preferred_columns = [
    "text",
    "description",
    "abstract",
    "sequence",
    "protein_sequence",
    "amino_acid_sequence",
    "name",
    "title",
    "comment",
    "content",
    "annotation",
    "function",
    "gene",
    "protein"
]

text_column = None

for preferred in preferred_columns:

    for column in text_columns:

        if column.lower().strip() == preferred:
            text_column = column
            break

    if text_column is not None:
        break


# ------------------------------------------------
# If no preferred column is found, choose the
# string column with the largest average length.
# ------------------------------------------------

if text_column is None:

    if len(text_columns) == 0:
        raise ValueError(
            "No text/string column was found in the dataset."
        )

    average_lengths = {}

    for column in text_columns:

        average_lengths[column] = (
            df[column]
            .fillna("")
            .astype(str)
            .str.len()
            .mean()
        )

    text_column = max(
        average_lengths,
        key=average_lengths.get
    )


print("\nSelected text column:")
print(text_column)


# ================================================================
# 7. CHECK SELECTED TEXT COLUMN
# ================================================================

print("\n" + "=" * 70)
print("SELECTED TEXT COLUMN")
print("=" * 70)

display(
    df[[text_column]].head(10)
)


# ================================================================
# 8. TEXT CLEANING FUNCTION
# ================================================================

def clean_text(text):

    # Convert missing values to empty string
    if pd.isna(text):
        return ""

    # Convert to string
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )

    # Remove HTML tags
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )

    # Remove numbers
    text = re.sub(
        r"\d+",
        " ",
        text
    )

    # Keep letters and spaces
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    # Remove stopwords
    words = text.split()

    words = [
        word
        for word in words
        if word not in stop_words
    ]

    # Reconstruct cleaned text
    text = " ".join(words)

    return text


# ================================================================
# 9. APPLY TEXT CLEANING
# ================================================================

df["clean_text"] = df[text_column].apply(
    clean_text
)

print("\nOriginal vs cleaned text:")

display(
    df[[text_column, "clean_text"]].head(10)
)


# ================================================================
# 10. REMOVE EMPTY TEXT RECORDS
# ================================================================

before_empty_removal = len(df)

df = df[
    df["clean_text"].str.strip() != ""
].reset_index(drop=True)

removed_empty = (
    before_empty_removal - len(df)
)

print(
    "\nRecords removed because of empty text:",
    removed_empty
)

print("Records remaining:", len(df))


# ================================================================
# 11. TEXT STATISTICS
# ================================================================

df["text_length"] = df["clean_text"].apply(
    lambda x: len(x.split())
)

print("\n" + "=" * 70)
print("TEXT STATISTICS")
print("=" * 70)

print(
    "Average number of words:",
    round(df["text_length"].mean(), 2)
)

print(
    "Minimum number of words:",
    df["text_length"].min()
)

print(
    "Maximum number of words:",
    df["text_length"].max()
)

plt.figure(figsize=(10, 5))

plt.hist(
    df["text_length"],
    bins=30
)

plt.xlabel("Number of Words")
plt.ylabel("Number of Records")
plt.title("Distribution of Text Length")

plt.tight_layout()
plt.show()


# ================================================================
# 12. TF-IDF VECTORIZATION
# ================================================================

print("\n" + "=" * 70)
print("TF-IDF FEATURE EXTRACTION")
print("=" * 70)

vectorizer = TfidfVectorizer(
    max_features=5000,
    min_df=1,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X = vectorizer.fit_transform(
    df["clean_text"]
)

print("TF-IDF matrix shape:")
print(X.shape)

print(
    "Number of features:",
    len(vectorizer.get_feature_names_out())
)


# ================================================================
# 13. GET TF-IDF FEATURE NAMES
# ================================================================

feature_names = vectorizer.get_feature_names_out()

print("\nFirst 50 TF-IDF features:")

print(
    feature_names[:50]
)


# ================================================================
# 14. CREATE ELBOW METHOD
# ================================================================

print("\n" + "=" * 70)
print("ELBOW METHOD")
print("=" * 70)

# Maximum K should not be larger than number of observations
max_k = min(10, len(df) - 1)

if max_k < 2:
    raise ValueError(
        "Not enough records to perform K-Means clustering."
    )

k_values = range(2, max_k + 1)

inertia_values = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)

    inertia_values.append(
        model.inertia_
    )


plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.xticks(list(k_values))
plt.grid(True)

plt.tight_layout()
plt.show()


# ================================================================
# 15. SILHOUETTE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("SILHOUETTE ANALYSIS")
print("=" * 70)

silhouette_values = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        labels
    )

    silhouette_values.append(score)


silhouette_results = pd.DataFrame({
    "K": list(k_values),
    "Silhouette Score": silhouette_values
})

display(
    silhouette_results
)


plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.xticks(list(k_values))
plt.grid(True)

plt.tight_layout()
plt.show()


# ================================================================
# 16. SELECT BEST K
# ================================================================

best_index = np.argmax(
    silhouette_values
)

best_k = list(k_values)[best_index]

best_silhouette = silhouette_values[best_index]

print("\nBest K:", best_k)

print(
    "Best Silhouette Score:",
    round(best_silhouette, 4)
)


# ================================================================
# 17. APPLY FINAL K-MEANS
# ================================================================

print("\n" + "=" * 70)
print("FINAL K-MEANS CLUSTERING")
print("=" * 70)

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=20,
    max_iter=300
)

cluster_labels = kmeans.fit_predict(X)

df["cluster"] = cluster_labels

print("K-Means clustering completed.")

print("\nCluster labels:")
print(
    np.unique(cluster_labels)
)


# ================================================================
# 18. CLUSTER COUNTS
# ================================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("\n" + "=" * 70)
print("CLUSTER SIZE")
print("=" * 70)

display(
    cluster_counts.to_frame(
        name="Number of Records"
    )
)


# ================================================================
# 19. CLUSTER SIZE BAR CHART
# ================================================================

plt.figure(figsize=(10, 6))

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.xlabel("Cluster")
plt.ylabel("Number of Records")
plt.title("Number of Records in Each Cluster")

plt.tight_layout()
plt.show()


# ================================================================
# 20. FINAL CLUSTERING METRICS
# ================================================================

print("\n" + "=" * 70)
print("CLUSTERING METRICS")
print("=" * 70)

final_silhouette = silhouette_score(
    X,
    df["cluster"]
)

davies_bouldin = davies_bouldin_score(
    X.toarray(),
    df["cluster"]
)

calinski_harabasz = calinski_harabasz_score(
    X.toarray(),
    df["cluster"]
)

print(
    "Silhouette Score:",
    round(final_silhouette, 4)
)

print(
    "Davies-Bouldin Score:",
    round(davies_bouldin, 4)
)

print(
    "Calinski-Harabasz Score:",
    round(calinski_harabasz, 4)
)


# ================================================================
# 21. PCA DIMENSIONALITY REDUCTION
# ================================================================

print("\n" + "=" * 70)
print("PCA VISUALIZATION")
print("=" * 70)

# PCA requires dense data
X_dense = X.toarray()

pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(
    X_dense
)

df["PCA1"] = X_pca[:, 0]
df["PCA2"] = X_pca[:, 1]

print(
    "Variance explained by PC1:",
    round(pca.explained_variance_ratio_[0] * 100, 2),
    "%"
)

print(
    "Variance explained by PC2:",
    round(pca.explained_variance_ratio_[1] * 100, 2),
    "%"
)


# ================================================================
# 22. PCA CLUSTER VISUALIZATION
# ================================================================

plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=df,
    x="PCA1",
    y="PCA2",
    hue="cluster",
    palette="tab10",
    s=70,
    alpha=0.8
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title(
    "K-Means Clustering of Fatty Acid Synthase Dataset"
)

plt.legend(
    title="Cluster",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()


# ================================================================
# 23. FIND TOP TERMS FOR EACH CLUSTER
# ================================================================

print("\n" + "=" * 70)
print("TOP TERMS IN EACH CLUSTER")
print("=" * 70)

# Cluster centers
centers = kmeans.cluster_centers_

# Sort each cluster's center from highest to lowest
order_centroids = centers.argsort()[:, ::-1]

TOP_N = 20

cluster_keywords = {}

for cluster_number in range(best_k):

    top_indices = (
        order_centroids[
            cluster_number,
            :TOP_N
        ]
    )

    top_words = [
        feature_names[index]
        for index in top_indices
    ]

    cluster_keywords[cluster_number] = top_words

    print(
        f"\nCluster {cluster_number}:"
    )

    print(
        ", ".join(top_words)
    )


# ================================================================
# 24. CREATE CLUSTER KEYWORD TABLE
# ================================================================

keyword_rows = []

for cluster_number in range(best_k):

    keyword_rows.append({
        "Cluster": cluster_number,
        "Top Keywords": ", ".join(
            cluster_keywords[cluster_number]
        )
    })

keyword_df = pd.DataFrame(
    keyword_rows
)

print("\nCluster keyword summary:")

display(
    keyword_df
)


# ================================================================
# 25. DISPLAY EXAMPLES FROM EACH CLUSTER
# ================================================================

print("\n" + "=" * 70)
print("EXAMPLES FROM EACH CLUSTER")
print("=" * 70)

for cluster_number in range(best_k):

    print(
        f"\n{'=' * 70}"
    )

    print(
        f"CLUSTER {cluster_number}"
    )

    print(
        f"{'=' * 70}"
    )

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    display(
        cluster_data[
            [text_column, "cluster"]
        ].head(10)
    )


# ================================================================
# 26. CALCULATE CLUSTER CENTROIDS
# ================================================================

# Calculate average PCA coordinates per cluster
cluster_centers_pca = (
    df.groupby("cluster")[
        ["PCA1", "PCA2"]
    ]
    .mean()
)

print("\nCluster centers in PCA space:")

display(
    cluster_centers_pca
)


# ================================================================
# 27. PLOT CLUSTERS WITH CENTROIDS
# ================================================================

plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=df,
    x="PCA1",
    y="PCA2",
    hue="cluster",
    palette="tab10",
    s=60,
    alpha=0.6
)

plt.scatter(
    cluster_centers_pca["PCA1"],
    cluster_centers_pca["PCA2"],
    marker="X",
    s=250,
    c="black",
    label="Cluster Center"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title(
    "Fatty Acid Synthase K-Means Clusters with Centroids"
)

plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()


# ================================================================
# 28. CREATE FINAL RESULTS DATAFRAME
# ================================================================

# Make a copy so that the original dataframe remains available
results_df = df.copy()

print("\nFinal dataframe:")
display(
    results_df.head()
)

print(
    "\nFinal dataframe shape:",
    results_df.shape
)


# ================================================================
# 29. SAVE RESULTS
# ================================================================

OUTPUT_FILE = (
    "fatty_acid_synthase_kmeans_results.csv"
)

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    f"\nResults saved successfully to: {OUTPUT_FILE}"
)


# ================================================================
# 30. SAVE CLUSTER KEYWORDS
# ================================================================

KEYWORDS_FILE = (
    "fatty_acid_synthase_cluster_keywords.csv"
)

keyword_df.to_csv(
    KEYWORDS_FILE,
    index=False
)

print(
    f"Cluster keywords saved to: {KEYWORDS_FILE}"
)


# ================================================================
# 31. FINAL SUMMARY
# ================================================================

print("\n")
print("=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print(
    f"Input file: {FILE_NAME}"
)

print(
    f"Text column: {text_column}"
)

print(
    f"Number of records: {len(df)}"
)

print(
    f"Number of TF-IDF features: {X.shape[1]}"
)

print(
    f"Optimal K: {best_k}"
)

print(
    f"Silhouette Score: {final_silhouette:.4f}"
)

print(
    f"Davies-Bouldin Score: {davies_bouldin:.4f}"
)

print(
    f"Calinski-Harabasz Score: {calinski_harabasz:.4f}"
)

print("\nCluster sizes:")

for cluster_number, count in cluster_counts.items():

    print(
        f"  Cluster {cluster_number}: {count} records"
    )

print("\nTop keywords:")

for cluster_number in range(best_k):

    print(
        f"\nCluster {cluster_number}:"
    )

    print(
        ", ".join(
            cluster_keywords[cluster_number]
        )
    )

print("\nOutput files:")
print(
    f"1. {OUTPUT_FILE}"
)
print(
    f"2. {KEYWORDS_FILE}"
)

print("\n" + "=" * 70)
print("ANALYSIS COMPLETE")
print("=" * 70)
```
