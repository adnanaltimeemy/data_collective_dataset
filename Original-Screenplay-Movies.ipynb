````python

# 1. Import libraries
import pandas as pd
import re
import nltk
import matplotlib.pyplot as plt

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

nltk.download("stopwords")

# 2. Load dataset
df = pd.read_csv("Original-Screenplay-Movies.csv")

print(df.head())
print(df.info())

# 3. Clean column names
df.columns = df.columns.str.lower().str.replace(" ", "_")

# 4. Select text columns
# Change these names if your CSV uses different names
text_cols = [c for c in ["movie_name", "genre", "description", "plot"]
             if c in df.columns]

df["text"] = df[text_cols].fillna("").astype(str).agg(" ".join, axis=1)

# 5. NLP cleaning
stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    words = text.split()
    words = [w for w in words if w not in stop_words]
    return " ".join(words)

df["clean_text"] = df["text"].apply(clean_text)

# 6. TF-IDF
vectorizer = TfidfVectorizer(max_features=3000)
X = vectorizer.fit_transform(df["clean_text"])

print("TF-IDF shape:", X.shape)

# 7. K-Means
k = 5
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X)

# 8. Show clusters
for i in range(k):
    print("\nCluster", i)
    print(df[df["cluster"] == i][text_cols].head(10))

# 9. PCA visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.toarray())

plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=df["cluster"], alpha=0.6)
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("K-Means Clustering of Movies")
plt.colorbar(label="Cluster")
plt.show()

# 10. Save results
df.to_csv("Movie_Clusters.csv", index=False)

print("Done! Results saved as Movie_Clusters.csv")
```
````
