```python
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = str(text).lower()
    text = ''.join([c for c in text if c.isalpha() or c == ' '])
    words = text.split()
    words = [w for w in words if w not in stop_words]
    return ' '.join(words)
```
