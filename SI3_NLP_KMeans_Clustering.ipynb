```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

text_data = [
    "The economy is improving with lower inflation",
    "Stock markets are rising due to tech gains",
    "Football teams are preparing for the season",
    "The match ended in a surprising victory",
    "New AI models are changing technology",
    "Deep learning improves natural language processing"
]

k = 2

vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(text_data)

kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
kmeans.fit(X)

terms = vectorizer.get_feature_names_out()

order_centroids = kmeans.cluster_centers_.argsort()[:, ::-1]

for i in range(k):
    print(f"\nCluster {i} top words:")
    top_words = [terms[ind] for ind in order_centroids[i, :10]]
    print(top_words)
```
