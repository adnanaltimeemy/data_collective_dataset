```python
# ============================================================
# LEAGUE OF LEGENDS DATA ANALYSIS
# NLP + K-MEANS CLUSTERING
# ============================================================


# ============================================================
# CELL 1 - INSTALL LIBRARIES
# ============================================================

# Run this cell if the libraries are not already installed.

# !pip install pandas numpy scikit-learn matplotlib seaborn nltk


# ============================================================
# CELL 2 - IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings

warnings.filterwarnings("ignore")

# Machine learning
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics import silhouette_score

# NLP
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

print("Libraries imported successfully.")


# ============================================================
# CELL 3 - DOWNLOAD NLP RESOURCES
# ============================================================

nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

print("NLP resources downloaded.")


# ============================================================
# CELL 4 - LOAD DATASET
# ============================================================

file_path = "league_of_legends_data_large.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print()
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())


# ============================================================
# CELL 5 - BASIC INFORMATION
# ============================================================

print("Dataset information:")
print("=" * 60)

df.info()


# ============================================================
# CELL 6 - DISPLAY COLUMN NAMES
# ============================================================

print("Column names:")
print("=" * 60)

for i, column in enumerate(df.columns):
    print(i, ":", column)


# ============================================================
# CELL 7 - CHECK MISSING VALUES
# ============================================================

print("Missing values:")
print("=" * 60)

missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values,
    "Missing Percentage": 
        (missing_values.values / len(df)) * 100
})

display(
    missing_table
    .sort_values("Missing Values", ascending=False)
)


# ============================================================
# CELL 8 - CHECK DUPLICATES
# ============================================================

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)

if duplicates > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")

print("Dataset shape after removing duplicates:", df.shape)


# ============================================================
# CELL 9 - DATA TYPES
# ============================================================

print("Data types:")
print("=" * 60)

display(
    pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str).values
    })
)


# ============================================================
# CELL 10 - IDENTIFY NUMERICAL AND TEXT COLUMNS
# ============================================================

numeric_columns = df.select_dtypes(
    include=["int64", "int32", "float64", "float32"]
).columns.tolist()

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Numerical columns:")
for column in numeric_columns:
    print("-", column)

print()

print("Text columns:")
for column in text_columns:
    print("-", column)


# ============================================================
# CELL 11 - REMOVE POTENTIAL ID COLUMNS
# ============================================================

# ID columns usually do not contain useful information for clustering.
# We automatically detect columns whose names look like IDs.

possible_id_columns = []

for column in numeric_columns + text_columns:
    
    column_lower = column.lower()
    
    if (
        column_lower == "id"
        or column_lower.endswith("_id")
        or column_lower.endswith("id")
        or column_lower.startswith("id_")
    ):
        possible_id_columns.append(column)

print("Potential ID columns:")
print(possible_id_columns)


# ============================================================
# CELL 12 - CREATE WORKING COPY
# ============================================================

data = df.copy()

# Remove obvious ID columns
for column in possible_id_columns:
    if column in data.columns:
        data = data.drop(columns=[column])

print("Working dataset shape:", data.shape)


# ============================================================
# CELL 13 - RE-DETECT COLUMNS
# ============================================================

numeric_columns = data.select_dtypes(
    include=["int64", "int32", "float64", "float32"]
).columns.tolist()

text_columns = data.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Numerical columns:")
print(numeric_columns)

print()

print("Text columns:")
print(text_columns)


# ============================================================
# CELL 14 - NUMERICAL DATA CLEANING
# ============================================================

if len(numeric_columns) > 0:
    
    numerical_data = data[numeric_columns].copy()
    
    # Replace infinite values
    numerical_data = numerical_data.replace(
        [np.inf, -np.inf],
        np.nan
    )
    
    # Replace missing values with median
    for column in numeric_columns:
        numerical_data[column] = numerical_data[column].fillna(
            numerical_data[column].median()
        )
    
    print("Numerical data cleaned.")
    
    display(numerical_data.head())

else:
    
    numerical_data = None
    print("No numerical columns found.")


# ============================================================
# CELL 15 - SCALE NUMERICAL DATA
# ============================================================

if numerical_data is not None:
    
    scaler = StandardScaler()
    
    X_numeric = scaler.fit_transform(
        numerical_data
    )
    
    print("Numerical data scaled.")
    print("Shape:", X_numeric.shape)

else:
    
    X_numeric = None


# ============================================================
# CELL 16 - NLP TEXT PREPROCESSING
# ============================================================

stop_words = set(
    stopwords.words("english")
)

lemmatizer = WordNetLemmatizer()


def clean_text(text):
    
    # Convert to string
    text = str(text)
    
    # Convert to lowercase
    text = text.lower()
    
    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )
    
    # Remove HTML
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )
    
    # Keep alphabetic characters
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )
    
    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()
    
    # Tokenise
    words = text.split()
    
    # Remove stopwords
    words = [
        word
        for word in words
        if word not in stop_words
    ]
    
    # Remove very short words
    words = [
        word
        for word in words
        if len(word) > 2
    ]
    
    # Lemmatise
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]
    
    return " ".join(words)


# ============================================================
# CELL 17 - COMBINE TEXT COLUMNS
# ============================================================

if len(text_columns) > 0:
    
    # Convert missing text to empty strings
    text_data = data[text_columns].fillna("")
    
    # Convert everything to strings
    text_data = text_data.astype(str)
    
    # Combine all text columns into one document per row
    combined_text = text_data.agg(
        " ".join,
        axis=1
    )
    
    # Clean the text
    clean_text_data = combined_text.apply(
        clean_text
    )
    
    print("Text preprocessing completed.")
    
    display(
        pd.DataFrame({
            "Original Text": combined_text.head(10),
            "Cleaned Text": clean_text_data.head(10)
        })
    )

else:
    
    clean_text_data = None
    
    print("No text columns found.")


# ============================================================
# CELL 18 - TF-IDF
# ============================================================

if clean_text_data is not None:
    
    tfidf = TfidfVectorizer(
        max_features=1000,
        min_df=2,
        max_df=0.95,
        ngram_range=(1, 2),
        sublinear_tf=True
    )
    
    X_text = tfidf.fit_transform(
        clean_text_data
    )
    
    print("TF-IDF completed.")
    print("TF-IDF matrix shape:", X_text.shape)

else:
    
    X_text = None


# ============================================================
# CELL 19 - DISPLAY IMPORTANT WORDS
# ============================================================

if X_text is not None:
    
    feature_names = tfidf.get_feature_names_out()
    
    tfidf_scores = np.asarray(
        X_text.mean(axis=0)
    ).flatten()
    
    word_scores = pd.DataFrame({
        "Word": feature_names,
        "TF-IDF Score": tfidf_scores
    })
    
    word_scores = word_scores.sort_values(
        "TF-IDF Score",
        ascending=False
    )
    
    print("Most important words:")
    
    display(
        word_scores.head(30)
    )


# ============================================================
# CELL 20 - REDUCE NLP DIMENSIONS
# ============================================================

if X_text is not None:
    
    # Number of components cannot be larger than
    # the number of TF-IDF features.
    
    n_components = min(
        50,
        X_text.shape[1] - 1
    )
    
    if n_components >= 2:
        
        svd = TruncatedSVD(
            n_components=n_components,
            random_state=42
        )
        
        X_text_reduced = svd.fit_transform(
            X_text
        )
        
        print(
            "NLP dimensions reduced from",
            X_text.shape[1],
            "to",
            X_text_reduced.shape[1]
        )
        
    else:
        
        X_text_reduced = X_text.toarray()
        
else:
    
    X_text_reduced = None


# ============================================================
# CELL 21 - EXPLAINED VARIANCE
# ============================================================

if "svd" in globals():
    
    explained_variance = svd.explained_variance_ratio_
    
    print(
        "Total explained variance:",
        explained_variance.sum()
    )
    
    plt.figure(figsize=(10, 5))
    
    plt.plot(
        np.cumsum(explained_variance),
        marker="o"
    )
    
    plt.xlabel("Number of SVD Components")
    plt.ylabel("Cumulative Explained Variance")
    plt.title("NLP SVD Explained Variance")
    plt.grid(True)
    plt.show()


# ============================================================
# CELL 22 - COMBINE NUMERICAL + NLP FEATURES
# ============================================================

feature_sets = []

if X_numeric is not None:
    feature_sets.append(X_numeric)

if X_text_reduced is not None:
    feature_sets.append(X_text_reduced)


if len(feature_sets) == 0:
    
    raise ValueError(
        "No numerical or text features were available."
    )

elif len(feature_sets) == 1:
    
    X = feature_sets[0]

else:
    
    X = np.hstack(feature_sets)


print("Final feature matrix shape:")
print(X.shape)


# ============================================================
# CELL 23 - CHECK FINAL DATA
# ============================================================

print("Checking final feature matrix...")

print("Contains NaN:", np.isnan(X).any())
print("Contains infinity:", np.isinf(X).any())

if np.isnan(X).any():
    
    X = np.nan_to_num(X)

if np.isinf(X).any():
    
    X = np.nan_to_num(X)

print("Final matrix is ready for K-Means.")


# ============================================================
# CELL 24 - ELBOW METHOD
# ============================================================

inertias = []

k_range = range(2, 11)

print("Running K-Means for different K values...")

for k in k_range:
    
    kmeans = KMeans(
        n_clusters=k,
        init="k-means++",
        n_init=10,
        max_iter=300,
        random_state=42
    )
    
    kmeans.fit(X)
    
    inertias.append(
        kmeans.inertia_
    )


# Plot elbow curve

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_range),
    inertias,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Selecting K")
plt.xticks(list(k_range))
plt.grid(True)

plt.show()


# ============================================================
# CELL 25 - SILHOUETTE SCORE
# ============================================================

silhouette_scores = []

print("Calculating silhouette scores...")

for k in k_range:
    
    kmeans = KMeans(
        n_clusters=k,
        init="k-means++",
        n_init=10,
        max_iter=300,
        random_state=42
    )
    
    labels = kmeans.fit_predict(X)
    
    score = silhouette_score(
        X,
        labels
    )
    
    silhouette_scores.append(score)
    
    print(
        "K =", k,
        "| Silhouette Score =",
        round(score, 4)
    )


# Plot silhouette scores

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_range),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")
plt.xticks(list(k_range))
plt.grid(True)

plt.show()


# ============================================================
# CELL 26 - SELECT BEST K
# ============================================================

best_k = list(k_range)[
    np.argmax(silhouette_scores)
]

best_score = max(
    silhouette_scores
)

print("=" * 60)
print("BEST NUMBER OF CLUSTERS")
print("=" * 60)

print("Best K:", best_k)
print("Silhouette Score:", round(best_score, 4))


# ============================================================
# CELL 27 - FINAL K-MEANS MODEL
# ============================================================

final_kmeans = KMeans(
    n_clusters=best_k,
    init="k-means++",
    n_init=20,
    max_iter=500,
    random_state=42
)

cluster_labels = final_kmeans.fit_predict(
    X
)

print("K-Means clustering completed.")


# ============================================================
# CELL 28 - ADD CLUSTERS TO DATASET
# ============================================================

df_clustered = df.copy()

df_clustered["Cluster"] = cluster_labels

print("Clusters added to dataset.")

display(
    df_clustered.head()
)


# ============================================================
# CELL 29 - CLUSTER SIZES
# ============================================================

cluster_sizes = (
    df_clustered["Cluster"]
    .value_counts()
    .sort_index()
)

print("Number of records in each cluster:")
print()

display(
    cluster_sizes.to_frame(
        name="Number of Records"
    )
)


# ============================================================
# CELL 30 - CLUSTER SIZE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

cluster_sizes.plot(
    kind="bar"
)

plt.xlabel("Cluster")
plt.ylabel("Number of Records")
plt.title("Number of Records in Each Cluster")
plt.xticks(rotation=0)

plt.show()


# ============================================================
# CELL 31 - REDUCE FINAL FEATURES TO 2D
# ============================================================

visual_svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = visual_svd.fit_transform(
    X
)

print("2D visualisation created.")


# ============================================================
# CELL 32 - VISUALISE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=cluster_labels,
    cmap="tab10",
    alpha=0.6
)

plt.xlabel("Component 1")
plt.ylabel("Component 2")

plt.title(
    "League of Legends Dataset - K-Means Clusters"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.show()


# ============================================================
# CELL 33 - CLUSTER CENTRES
# ============================================================

print("Cluster centre shape:")
print(final_kmeans.cluster_centers_.shape)


# ============================================================
# CELL 34 - NUMERICAL CLUSTER ANALYSIS
# ============================================================

if len(numeric_columns) > 0:
    
    analysis_data = df_clustered[
        numeric_columns + ["Cluster"]
    ].copy()
    
    cluster_means = (
        analysis_data
        .groupby("Cluster")[numeric_columns]
        .mean()
    )
    
    print("Average numerical values by cluster:")
    
    display(
        cluster_means
    )


# ============================================================
# CELL 35 - NORMALISED CLUSTER PROFILE
# ============================================================

if len(numeric_columns) > 0:
    
    # Standardise cluster averages so we can compare
    # variables on the same scale.
    
    profile_scaler = StandardScaler()
    
    profile_values = profile_scaler.fit_transform(
        cluster_means
    )
    
    cluster_profile = pd.DataFrame(
        profile_values,
        index=cluster_means.index,
        columns=cluster_means.columns
    )
    
    print("Standardised cluster profile:")
    
    display(
        cluster_profile
    )


# ============================================================
# CELL 36 - HEATMAP OF CLUSTER PROFILES
# ============================================================

if len(numeric_columns) > 0:
    
    plt.figure(figsize=(14, 8))
    
    sns.heatmap(
        cluster_profile,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0
    )
    
    plt.title(
        "Standardised Numerical Features by Cluster"
    )
    
    plt.xlabel("Features")
    plt.ylabel("Cluster")
    
    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 37 - NLP CLUSTER ANALYSIS
# ============================================================

if clean_text_data is not None:
    
    df_clustered["Clean_Text"] = (
        clean_text_data.values
    )
    
    print("Example text records for each cluster:")
    print()
    
    for cluster in sorted(
        df_clustered["Cluster"].unique()
    ):
        
        print("=" * 70)
        print("CLUSTER", cluster)
        print("=" * 70)
        
        cluster_text = df_clustered[
            df_clustered["Cluster"] == cluster
        ]["Clean_Text"]
        
        for text in cluster_text.head(5):
            print("-", text)
        
        print()


# ============================================================
# CELL 38 - MOST IMPORTANT WORDS BY CLUSTER
# ============================================================

if X_text is not None:
    
    feature_names = tfidf.get_feature_names_out()
    
    print(
        "Important NLP words for each cluster:"
    )
    
    print()
    
    for cluster in sorted(
        df_clustered["Cluster"].unique()
    ):
        
        cluster_indices = np.where(
            cluster_labels == cluster
        )[0]
        
        cluster_tfidf = X_text[
            cluster_indices
        ]
        
        mean_tfidf = np.asarray(
            cluster_tfidf.mean(axis=0)
        ).flatten()
        
        top_indices = mean_tfidf.argsort()[
            ::-1
        ][:15]
        
        print("=" * 70)
        print("CLUSTER", cluster)
        print("=" * 70)
        
        for index in top_indices:
            print(
                feature_names[index],
                "->",
                round(mean_tfidf[index], 4)
            )
        
        print()


# ============================================================
# CELL 39 - COMPARE CLUSTER DISTRIBUTION
# ============================================================

cluster_percentage = (
    cluster_sizes / len(df_clustered)
) * 100

cluster_summary = pd.DataFrame({
    "Cluster": cluster_sizes.index,
    "Records": cluster_sizes.values,
    "Percentage": cluster_percentage.values
})

display(
    cluster_summary
)


# ============================================================
# CELL 40 - FINAL SILHOUETTE SCORE
# ============================================================

final_silhouette = silhouette_score(
    X,
    cluster_labels
)

print("=" * 60)
print("FINAL MODEL PERFORMANCE")
print("=" * 60)

print(
    "Number of clusters:",
    best_k
)

print(
    "Silhouette score:",
    round(final_silhouette, 4)
)

print(
    "Inertia:",
    round(final_kmeans.inertia_, 2)
)


# ============================================================
# CELL 41 - SAVE CLUSTERED DATASET
# ============================================================

output_file = (
    "league_of_legends_kmeans_results.csv"
)

df_clustered.to_csv(
    output_file,
    index=False
)

print(
    "Clustered dataset saved as:",
    output_file
)


# ============================================================
# CELL 42 - SAVE CLUSTER SUMMARY
# ============================================================

summary_file = (
    "league_of_legends_cluster_summary.csv"
)

cluster_summary.to_csv(
    summary_file,
    index=False
)

print(
    "Cluster summary saved as:",
    summary_file
)


# ============================================================
# CELL 43 - FINAL RESULTS
# ============================================================

print()
print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(
    "Original dataset size:",
    df.shape
)

print(
    "Final feature matrix:",
    X.shape
)

print(
    "Optimal K:",
    best_k
)

print(
    "Final silhouette score:",
    round(final_silhouette, 4)
)

print(
    "Number of clusters:",
    df_clustered["Cluster"].nunique()
)

print()
print("Cluster sizes:")

display(
    cluster_summary
)

print()
print("Analysis completed successfully.")
```
