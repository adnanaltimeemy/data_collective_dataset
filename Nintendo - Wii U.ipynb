```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 2. Load dataset
df = pd.read_csv("Nintendo - Wii U.csv")
print(df.head())
print(df.columns)

# 3. Select text columns and combine them
text_cols = df.select_dtypes(include="object").columns
df["text"] = df[text_cols].fillna("").astype(str).agg(" ".join, axis=1)

# 4. NLP - convert text into TF-IDF features
tfidf = TfidfVectorizer(stop_words="english", max_features=500)
X = tfidf.fit_transform(df["text"])

# 5. K-Means clustering
k = 4
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# 6. Display clusters
print("\nCluster counts:")
print(df["Cluster"].value_counts())

# 7. Show important words in each cluster
terms = tfidf.get_feature_names_out()

for i in range(k):
    words = terms[kmeans.cluster_centers_[i].argsort()[-10:][::-1]]
    print(f"\nCluster {i}: {', '.join(words)}")

# 8. Visualise clusters using PCA
X_pca = PCA(n_components=2).fit_transform(X.toarray())

plt.figure(figsize=(8,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Nintendo Wii U - K-Means Clustering")
plt.show()

# 9. Save results
df.to_csv("Nintendo_Wii_U_Clustered.csv", index=False)

print("\nDone! File saved as Nintendo_Wii_U_Clustered.csv")
```
