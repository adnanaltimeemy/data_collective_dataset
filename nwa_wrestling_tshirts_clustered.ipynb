```python
import os
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# 1. Create a synthetic dataset of NWA Wrestling TV Show T-Shirts
data = {
    'shirt_id': [f"NWA-TS-{i:03d}" for i in range(1, 21)],
    'shirt_name': [
        "NWA Worlds Heavyweight Championship Tee", "Ten Pounds of Gold Classic Shirt",
        "Nick Aldis National Treasure Tee", "Cody Rhodes NWA Title Victory Shirt",
        "NWA Powerrr Official Logo Tee", "NWA Powerrr Studio Wrestling Shirt",
        "Rock 'n' Roll Express Legend Tee", "Midnight Express Retro Wrestling Shirt",
        "NWA Crockett Cup Tournament Tee", "NWA 75th Anniversary Show Shirt",
        "Kamille The Brickhouse Women's Tee", "NWA World Women's Championship Shirt",
        "Trevor Murdoch Real Racehorse Tee", "Harley Race King of Wrestling Retro",
        "Dusty Rhodes American Dream Vintage", "Ric Flair Nature Boy Wooo Tee",
        "NWA USA Television Championship Tee", "NWA Shockwave Network Logo Shirt",
        "Strictly Business Faction T-Shirt", "La Rebelión Tag Team Champions Tee"
    ],
    'description': [
        "Official t-shirt featuring the historic NWA Worlds Heavyweight Championship belt graphic.",
        "Celebrate the legacy of the Ten Pounds of Gold with this vintage style wrestling shirt.",
        "Show your support for the National Treasure Nick Aldis with his signature modern tee.",
        "Commemorative shirt celebrating Cody Rhodes winning the historic NWA World Title.",
        "The official logo t-shirt for the weekly flagship show NWA Powerrr. Modern retro style.",
        "Relive the retro studio wrestling vibes with this official NWA Powerrr production shirt.",
        "Classic retro tee honoring the legendary tag team Rock 'n' Roll Express.",
        "Vintage look wrestling shirt dedicated to the iconic Midnight Express tag team.",
        "Official merchandise for the historic NWA Crockett Cup tag team tournament.",
        "Special event merchandise celebrating the massive NWA 75th Anniversary weekend show.",
        "Represent the dominant women's champion Kamille The Brickhouse with this graphic tee.",
        "Celebrate women's wrestling history with the NWA World Women's Championship shirt.",
        "Support the blue-collar champion Trevor Murdoch with his official wrestling tee.",
        "Retro vintage shirt dedicated to the legendary 8-time NWA Champion Harley Race.",
        "Vintage style tee honoring the legendary American Dream Dusty Rhodes.",
        "Classic stylized graphic tee featuring Ric Flair's iconic Nature Boy styling.",
        "Official shirt representing the NWA USA weekly show and the Television Title.",
        "Show your love for the NWA Shockwave era with this official network logo shirt.",
        "Align yourself with the dominant championship faction Strictly Business with this tee.",
        "Official tag team championship shirt for the dynamic duo La Rebelión."
    ],
    'price': [29.99, 27.99, 29.99, 32.99, 24.99, 24.99, 27.99, 27.99, 29.99, 34.99,
              28.99, 29.99, 27.99, 29.99, 29.99, 29.99, 24.99, 24.99, 29.99, 28.99]
}

df = pd.DataFrame(data)

# 2. NLP Feature Extraction using TF-IDF (Term Frequency-Inverse Document Frequency)
vectorizer = TfidfVectorizer(stop_words='english', max_features=50)
tfidf_matrix = vectorizer.fit_transform(df['description'])

# 3. K-Means Clustering on the NLP extracted features
num_clusters = 4
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
df['cluster_id'] = kmeans.fit_predict(tfidf_matrix)

# Map human-readable thematic labels to clusters
cluster_labels = {
    0: "Retro & Legends Showcase",
    1: "Weekly TV Shows & Logos",
    2: "Championships & Title Victories",
    3: "Modern Factions & Modern Talent"
}
df['cluster_theme'] = df['cluster_id'].map(lambda x: cluster_labels.get(x, f"Theme Cluster {x}"))

# 4. Save to CSV
os.makedirs('generated', exist_ok=True)
csv_path = 'generated/nwa_wrestling_tshirts_clustered.csv'
df.to_csv(csv_path, index=False)

print(f"Successfully generated dataset at: {csv_path}")
```
