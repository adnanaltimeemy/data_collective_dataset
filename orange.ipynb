```python
# 1. Import libraries
import pandas as pd
import re
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# 2. Load dataset
df = pd.read_csv("Orange.csv")
print(df.head())
print(df.columns)

# 3. Select text column
text_col = "text"   # change this if your column has another name

# 4. Clean text
def clean_text(x):
    x = str(x).lower()
    x = re.sub(r'[^a-z\s]', '', x)
    return x

df["clean_text"] = df[text_col].apply(clean_text)

# 5. NLP - TF-IDF
vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(df["clean_text"])

# 6. K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["cluster"] = model.fit_predict(X)

# 7. Show results
print(df[[text_col, "cluster"]].head(20))

# 8. Show important words in each cluster
words = vectorizer.get_feature_names_out()

for i in range(k):
    indexes = model.cluster_centers_[i].argsort()[-10:][::-1]
    print("Cluster", i, ":", [words[j] for j in indexes])

# 9. Visualise clusters
plt.scatter(X.toarray()[:, 0], X.toarray()[:, 1], c=df["cluster"])
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Means Clustering")
plt.show()

# 10. Save results
df.to_csv("Orange_clustered.csv", index=False)
```
