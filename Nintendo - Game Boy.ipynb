```python
# 1. Import libraries
import pandas as pd
import re
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 2. Load dataset
df = pd.read_csv("Nintendo - Game Boy.csv")
print(df.head())
print(df.columns)

# 3. Select text column
# Change 'Review' if your text column has a different name
text_col = "Review"

df = df.dropna(subset=[text_col])
df[text_col] = df[text_col].astype(str)

# 4. NLP cleaning
def clean(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    return text

df["clean_text"] = df[text_col].apply(clean)

# 5. TF-IDF
tfidf = TfidfVectorizer(stop_words="english", max_features=1000)
X = tfidf.fit_transform(df["clean_text"])

# 6. K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# 7. Show results
print(df[[text_col, "Cluster"]].head(10))

# 8. Show important words in each cluster
words = tfidf.get_feature_names_out()

for i in range(k):
    top_words = model.cluster_centers_[i].argsort()[-10:][::-1]
    print("Cluster", i, ":", [words[j] for j in top_words])

# 9. Visualise clusters
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.toarray())

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Nintendo Game Boy - K-Means Clustering")
plt.show()

# 10. Save results
df.to_csv("Nintendo_Game_Boy_Clusters.csv", index=False)
```
