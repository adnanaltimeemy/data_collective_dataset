```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

nltk.download("stopwords")

# Load dataset
df = pd.read_csv("letters.csv")

# Text column
text = df["text"].fillna("").astype(str).str.lower()

# NLP preprocessing
stop = set(stopwords.words("english"))
text = text.apply(lambda x: " ".join(
    w for w in x.split() if w.isalpha() and w not in stop
))

# TF-IDF
X = TfidfVectorizer().fit_transform(text)

# K-Means
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X)

# Show results
print(df[["text", "cluster"]])

# Plot
pca = PCA(n_components=2)
x = pca.fit_transform(X.toarray())
plt.scatter(x[:,0], x[:,1], c=df["cluster"])
plt.title("K-Means Clustering")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.show()

# Save
df.to_csv("letters_clustered.csv", index=False)
```
