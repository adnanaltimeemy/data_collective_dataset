```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# 2. Load dataset
df = pd.read_csv("match_events.csv")
print(df.head())
print(df.columns)

# 3. Select text column
# Change "event" if your text column has a different name
text_col = "event"
df[text_col] = df[text_col].fillna("").astype(str)

# 4. NLP: TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(df[text_col])

# 5. K-Means clustering
k = 4
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# 6. Evaluate
print("Silhouette Score:", silhouette_score(X, df["Cluster"]))

# 7. Show clusters
for i in range(k):
    print(f"\nCluster {i}")
    print(df[df["Cluster"] == i][text_col].head(10).to_list())

# 8. PCA visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.toarray())

plt.scatter(X_pca[:,0], X_pca[:,1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("K-Means Clustering of Match Events")
plt.show()

# 9. Save results
df.to_csv("match_events_clustered.csv", index=False)
print("Done!")
```
