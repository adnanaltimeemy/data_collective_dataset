```python
# Install if needed:
# !pip install pandas scikit-learn nltk matplotlib

import pandas as pd
import re
import nltk
import matplotlib.pyplot as plt

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

nltk.download('stopwords')

# 1. Load data
df = pd.read_csv("bakery_data.csv")
print(df.head())
print(df.columns)

# 2. Select text column
# Change 'Product' to your actual text column name
text_col = 'Product'

# 3. NLP cleaning
stop = set(stopwords.words('english'))

def clean(text):
    text = str(text).lower()
    text = re.sub(r'[^a-z\s]', '', text)
    return ' '.join(w for w in text.split() if w not in stop)

df['clean_text'] = df[text_col].apply(clean)

# 4. TF-IDF
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(df['clean_text'])

# 5. K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df['Cluster'] = model.fit_predict(X)

# 6. Show results
print(df[[text_col, 'Cluster']])

# 7. Cluster sizes
print("\nCluster counts:")
print(df['Cluster'].value_counts())

# 8. Visualisation
plt.scatter(
    range(len(df)),
    df['Cluster'],
    c=df['Cluster']
)
plt.xlabel("Bakery Products")
plt.ylabel("Cluster")
plt.title("Bakery Products - K-Means Clustering")
plt.show()

# 9. Save result
df.to_csv("bakery_clustered.csv", index=False)
```
