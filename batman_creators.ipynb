```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 2. Load dataset
df = pd.read_csv("batman_creators.csv")
print(df.head())
print(df.columns)

# 3. Select text column
# Change 'description' to your actual text column name
text = df["description"].fillna("").astype(str)

# 4. NLP: TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(text)

# 5. K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["cluster"] = model.fit_predict(X)

# 6. Show results
print(df[["description", "cluster"]])

# 7. Important words in each cluster
words = tfidf.get_feature_names_out()

for i in range(k):
    top = model.cluster_centers_[i].argsort()[-10:][::-1]
    print("Cluster", i, ":", [words[j] for j in top])

# 8. Visualisation
pca = PCA(n_components=2)
points = pca.fit_transform(X.toarray())

plt.scatter(points[:,0], points[:,1], c=df["cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Batman Creators - K-Means Clustering")
plt.show()

# 9. Save results
df.to_csv("batman_creators_clustered.csv", index=False)
```
