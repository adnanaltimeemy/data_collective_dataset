```python
import os
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# 1. Create a curated dataset of Ring of Honor (ROH) TV Show T-Shirts
data = {
    'shirt_id': [f"ROH-{i:03d}" for i in range(1, 16)],
    'title': [
        "ROH Classic Logo Black Tee",
        "Demolition Derby Briscoe Boys Shirt",
        "Jay Lethal Franchise Champion Tee",
        "Honor Is Real Typography Shirt",
        "The Kingdom Purple Conspiracy Tee",
        "Samoa Joe Submission Machine Vintage",
        "ROH World Championship Gold Graphic",
        "Briscoe Brothers Reach For the Sky",
        "Pure Wrestling Division Rulebook Tee",
        "Dalton Castle Peacock Boys Extravaganza",
        "Bullet Club ROH Joint Invasion Shirt",
        "ROH TV Taping Crew Graphic Tee",
        "Survival of the Fittest Tournament Shirt",
        "Matt Taven Melvin Fringe Tee",
        "Final Battle Pay-Per-View Commemorative"
    ],
    'description': [
        "The definitive classic Ring of Honor logo printed on a premium heavyweight black cotton t-shirt. Respect the code.",
        "Celebrate the legendary Briscoe Brothers with this redneck kung-fu themed graphic tee showing Demolition Derby style.",
        "Show your support for the longest-reigning ROH World Champion Jay Lethal with this elegant red and gold 'Franchise' design.",
        "A stylized typographic shirt featuring the core pillars of ROH wrestling: Honor, Integrity, and Pure Athleticism.",
        "Join Matt Taven and The Kingdom with this purple shirt highlighting the systematic conspiracy against the real champions.",
        "A throwback shirt honoring Samoa Joe's legendary undefeated championship run. Features a submission choke graphic.",
        "An intricate, high-quality gold metallic print showcasing the iconic ROH World Championship belt buckle design.",
        "The official Briscoe Boys 'Reach for the Sky' lifestyle tee. Aggressive graphics for ultimate wrestling fans.",
        "Dedicated to the technical masters. Features the official rules of the ROH Pure Wrestling Division printed on the back.",
        "Spread your wings with Dalton Castle. High-energy design featuring multi-coloured peacock feathers and bold text.",
        "Limited edition collaborative merchandise celebrating the Bullet Club's dramatic crossover invasion on ROH TV.",
        "Look like part of the Ring of Honor television production crew with this technical broadcast graphic top.",
        "Commemorative shirt listing the participants of the grueling annual Survival of the Fittest tournament.",
        "Rep the purple kingdom with this loud Matt Taven shirt mocking the 'Melvins' of the professional wrestling world.",
        "The official event merchandise for the biggest night on the Ring of Honor calendar: Final Battle."
    ]
}

df = pd.DataFrame(data)

# 2. NLP Feature Extraction using TF-IDF
# We filter out English stop words and focus on descriptive keywords
vectorizer = TfidfVectorizer(stop_words='english', max_features=50)
tfidf_matrix = vectorizer.fit_transform(df['description'])

# 3. Apply K-Means Clustering to group similar T-shirt descriptions
num_clusters = 3
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
df['cluster_id'] = kmeans.fit_predict(tfidf_matrix)

# 4. Map human-readable thematic names to the generated clusters
# (Themes are derived based on prevailing text keywords in each cluster group)
cluster_map = {
    0: "Championships & Major Events", 
    1: "Factions & Wrestler Branding", 
    2: "Classic Core Brand & Apparel"
}
df['cluster_theme'] = df['cluster_id'].map(cluster_map)

# 5. Export structured output to a CSV file
os.makedirs('generated', exist_ok=True)
csv_path = 'generated/roh_tshirts_dataset.csv'
df.to_csv(csv_path, index=False)

print(f"Dataset completely processed and saved to {csv_path}!")
print("\n--- Preview of Clustered Dataset ---")
print(df[['shirt_id', 'title', 'cluster_theme']].to_string(index=False))
```
