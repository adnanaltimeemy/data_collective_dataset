```python
# ============================================================
# JUPYTER NOTEBOOK
# NLP + TF-IDF + K-MEANS CLUSTERING
# Dataset: sampledJPairsAnnotated.csv
# ============================================================


# ============================================================
# CELL 1 — Install required packages
# ============================================================

# Run this cell if the packages are not already installed.

# !pip install pandas numpy scikit-learn matplotlib seaborn


# ============================================================
# CELL 2 — Import libraries
# ============================================================

import os
import re
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    adjusted_rand_score,
    normalized_mutual_info_score,
    confusion_matrix
)
from sklearn.decomposition import TruncatedSVD

warnings.filterwarnings("ignore")

print("Libraries imported successfully.")


# ============================================================
# CELL 3 — Configuration
# ============================================================

# CSV file
FILE_PATH = "sampledJPairsAnnotated.csv"

# Output file
OUTPUT_FILE = "sampledJPairsAnnotated_clustered.csv"

# Number of clusters to test
MIN_K = 2
MAX_K = 10

# Random seed for reproducibility
RANDOM_STATE = 42

# ------------------------------------------------------------
# OPTIONAL:
# If you know your text columns, put them here.
#
# Example:
# TEXT_COLUMNS = ["sentence1", "sentence2"]
#
# If you don't know the names, leave this as None and the
# notebook will try to identify suitable columns automatically.
# ------------------------------------------------------------

TEXT_COLUMNS = None

print("Configuration complete.")


# ============================================================
# CELL 4 — Check that the dataset exists
# ============================================================

if not os.path.exists(FILE_PATH):
    raise FileNotFoundError(
        f"Could not find '{FILE_PATH}'. "
        "Place the CSV file in the same folder as this notebook."
    )

print(f"Dataset found: {FILE_PATH}")


# ============================================================
# CELL 5 — Load dataset
# ============================================================

df = pd.read_csv(FILE_PATH)

print("Dataset loaded successfully.")
print()
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
print()

display(df.head())


# ============================================================
# CELL 6 — Inspect columns
# ============================================================

print("Column names:")
print("=" * 60)

for i, column in enumerate(df.columns):
    print(f"{i}: {column}")

print()
print("Data types:")
print("=" * 60)

print(df.dtypes)


# ============================================================
# CELL 7 — Missing-value analysis
# ============================================================

print("Missing values:")
print("=" * 60)

missing = df.isnull().sum()

missing_df = pd.DataFrame({
    "column": missing.index,
    "missing_values": missing.values,
    "missing_percentage": (
        missing.values / len(df) * 100
    ).round(2)
})

display(
    missing_df.sort_values(
        "missing_values",
        ascending=False
    )
)


# ============================================================
# CELL 8 — Display sample values from every column
# ============================================================

print("Sample values:")
print("=" * 80)

for column in df.columns:
    print(f"\nCOLUMN: {column}")
    print("-" * 80)

    values = df[column].dropna().astype(str).head(5)

    for value in values:
        print(value[:300])


# ============================================================
# CELL 9 — Automatically identify possible text columns
# ============================================================

if TEXT_COLUMNS is None:

    possible_text_columns = []

    for column in df.columns:

        # Convert non-null values to strings
        values = df[column].dropna().astype(str)

        if len(values) == 0:
            continue

        # Average text length
        average_length = values.str.len().mean()

        # Number of unique values
        unique_ratio = values.nunique() / len(values)

        # Ignore columns that look like IDs or simple numbers
        if (
            average_length > 10
            and unique_ratio > 0.01
        ):
            possible_text_columns.append(column)

    print("Automatically detected possible text columns:")
    print(possible_text_columns)

else:

    possible_text_columns = TEXT_COLUMNS

    print("Using manually specified text columns:")
    print(possible_text_columns)


# ============================================================
# CELL 10 — Select text columns
# ============================================================

# If automatic detection found nothing, display a useful error.

if len(possible_text_columns) == 0:

    raise ValueError(
        "No suitable text columns were detected. "
        "Set TEXT_COLUMNS manually in CELL 3."
    )

print("Text columns selected:")
print(possible_text_columns)


# ============================================================
# CELL 11 — Combine text columns
# ============================================================

# Fill missing values with empty strings
for column in possible_text_columns:
    df[column] = df[column].fillna("").astype(str)


# Combine all selected text columns
df["combined_text"] = (
    df[possible_text_columns]
    .agg(" ".join, axis=1)
)


print("Combined text created.")
display(
    df[possible_text_columns + ["combined_text"]].head()
)


# ============================================================
# CELL 12 — Japanese NLP cleaning
# ============================================================

def clean_text(text):

    text = str(text)

    # --------------------------------------------------------
    # Remove URLs
    # --------------------------------------------------------
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )

    # --------------------------------------------------------
    # Remove HTML tags
    # --------------------------------------------------------
    text = re.sub(
        r"<[^>]+>",
        " ",
        text
    )

    # --------------------------------------------------------
    # Remove email addresses
    # --------------------------------------------------------
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # --------------------------------------------------------
    # Normalize repeated whitespace
    # --------------------------------------------------------
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


df["clean_text"] = (
    df["combined_text"]
    .apply(clean_text)
)


print("Text cleaning completed.")

display(
    df[["combined_text", "clean_text"]].head(10)
)


# ============================================================
# CELL 13 — Remove empty documents
# ============================================================

before = len(df)

df = df[
    df["clean_text"].str.strip().str.len() > 0
].copy()

after = len(df)

print("Rows before removing empty text:", before)
print("Rows after removing empty text:", after)
print("Rows removed:", before - after)


# ============================================================
# CELL 14 — TF-IDF vectorization
# ============================================================

# Character n-grams work well as a baseline for Japanese text
# because Japanese does not require spaces between words.

vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

X = vectorizer.fit_transform(
    df["clean_text"]
)

print("TF-IDF completed.")
print()
print("Number of documents:", X.shape[0])
print("Number of TF-IDF features:", X.shape[1])
print("Matrix shape:", X.shape)


# ============================================================
# CELL 15 — Inspect TF-IDF features
# ============================================================

feature_names = vectorizer.get_feature_names_out()

print("Number of features:", len(feature_names))
print()
print("First 100 features:")

for feature in feature_names[:100]:
    print(feature)


# ============================================================
# CELL 16 — Find the best number of clusters
# ============================================================

k_values = range(
    MIN_K,
    MAX_K + 1
)

inertia_values = []
silhouette_values = []

print("Testing K values:")
print("=" * 60)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=10
    )

    labels = model.fit_predict(X)

    inertia = model.inertia_

    silhouette = silhouette_score(
        X,
        labels
    )

    inertia_values.append(inertia)
    silhouette_values.append(silhouette)

    print(
        f"K = {k:2d} | "
        f"Inertia = {inertia:.4f} | "
        f"Silhouette = {silhouette:.4f}"
    )


# ============================================================
# CELL 17 — Elbow plot
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("K-Means Elbow Method")

plt.xticks(list(k_values))
plt.grid(True)

plt.show()


# ============================================================
# CELL 18 — Silhouette plot
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.xlabel("Number of clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for K-Means")

plt.xticks(list(k_values))
plt.grid(True)

plt.show()


# ============================================================
# CELL 19 — Select best K
# ============================================================

best_index = np.argmax(
    silhouette_values
)

best_k = list(k_values)[best_index]

best_silhouette = silhouette_values[
    best_index
]

print("Best K:", best_k)
print("Best silhouette score:", best_silhouette)


# ============================================================
# CELL 20 — Final K-means model
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=RANDOM_STATE,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X)

df["cluster"] = cluster_labels

print("K-means clustering completed.")


# ============================================================
# CELL 21 — Cluster distribution
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("Records per cluster:")
print(cluster_counts)


# ============================================================
# CELL 22 — Cluster distribution plot
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    x=cluster_counts.index.astype(str),
    y=cluster_counts.values
)

plt.xlabel("Cluster")
plt.ylabel("Number of records")
plt.title("Number of Records in Each Cluster")

plt.show()


# ============================================================
# CELL 23 — Calculate final silhouette score
# ============================================================

final_silhouette = silhouette_score(
    X,
    df["cluster"]
)

print(
    f"Final Silhouette Score: "
    f"{final_silhouette:.4f}"
)


# ============================================================
# CELL 24 — Examine examples from each cluster
# ============================================================

for cluster_id in sorted(
    df["cluster"].unique()
):

    print()
    print("=" * 100)
    print(f"CLUSTER {cluster_id}")
    print("=" * 100)

    cluster_data = df[
        df["cluster"] == cluster_id
    ]

    display(
        cluster_data[
            possible_text_columns +
            ["cluster"]
        ].head(10)
    )


# ============================================================
# CELL 25 — Identify important TF-IDF features per cluster
# ============================================================

print("Important features for each cluster")
print("=" * 100)

order_centroids = (
    kmeans.cluster_centers_
    .argsort()[:, ::-1]
)

TOP_N = 20

cluster_features = {}

for cluster_id in range(best_k):

    top_indices = (
        order_centroids[
            cluster_id,
            :TOP_N
        ]
    )

    top_features = [
        feature_names[i]
        for i in top_indices
    ]

    cluster_features[
        cluster_id
    ] = top_features

    print()
    print(f"Cluster {cluster_id}:")
    print(", ".join(top_features))


# ============================================================
# CELL 26 — Convert important features into a DataFrame
# ============================================================

feature_table = pd.DataFrame(
    cluster_features
).T

feature_table.columns = [
    f"feature_{i+1}"
    for i in range(TOP_N)
]

feature_table.index.name = "cluster"

display(feature_table)


# ============================================================
# CELL 27 — Reduce dimensions using SVD
# ============================================================

svd = TruncatedSVD(
    n_components=2,
    random_state=RANDOM_STATE
)

X_2d = svd.fit_transform(X)

print(
    "Explained variance:",
    svd.explained_variance_ratio_.sum()
)


# ============================================================
# CELL 28 — Create visualization DataFrame
# ============================================================

plot_df = pd.DataFrame({
    "component_1": X_2d[:, 0],
    "component_2": X_2d[:, 1],
    "cluster": df["cluster"].values
})

display(plot_df.head())


# ============================================================
# CELL 29 — Visualize K-means clusters
# ============================================================

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=plot_df,
    x="component_1",
    y="component_2",
    hue="cluster",
    palette="tab10",
    s=60
)

plt.xlabel("SVD Component 1")
plt.ylabel("SVD Component 2")
plt.title("K-Means Clusters — TF-IDF + SVD")

plt.legend(
    title="Cluster",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 30 — Calculate cluster sizes
# ============================================================

cluster_summary = (
    df.groupby("cluster")
    .size()
    .reset_index(name="number_of_records")
)

cluster_summary["percentage"] = (
    cluster_summary["number_of_records"]
    / len(df)
    * 100
).round(2)

display(cluster_summary)


# ============================================================
# CELL 31 — Find representative examples
# ============================================================

# Calculate distance of each document to its cluster center.
#
# This gives us examples that are relatively close to the
# center of each cluster.

distances = kmeans.transform(X)

representative_rows = []

for cluster_id in range(best_k):

    cluster_indices = np.where(
        df["cluster"].values == cluster_id
    )[0]

    cluster_distances = distances[
        cluster_indices,
        cluster_id
    ]

    closest_positions = (
        cluster_indices[
            np.argsort(cluster_distances)[:5]
        ]
    )

    for position in closest_positions:

        representative_rows.append({
            "cluster": cluster_id,
            "row_index": df.index[position],
            "distance_to_centroid":
                distances[position, cluster_id],
            "text":
                df.iloc[position]["clean_text"]
        })


representative_df = pd.DataFrame(
    representative_rows
)

display(representative_df)


# ============================================================
# CELL 32 — Check for annotated/label columns
# ============================================================

# Look for columns that might contain annotation labels.

possible_label_columns = []

label_keywords = [
    "label",
    "class",
    "category",
    "annotation",
    "annotated",
    "target",
    "gold",
    "type"
]

for column in df.columns:

    column_lower = column.lower()

    if any(
        keyword in column_lower
        for keyword in label_keywords
    ):
        possible_label_columns.append(
            column
        )

print(
    "Possible annotation/label columns:"
)

print(
    possible_label_columns
)


# ============================================================
# CELL 33 — Optional: compare K-means clusters to annotation
# ============================================================

if len(possible_label_columns) > 0:

    label_column = possible_label_columns[0]

    print(
        f"Using label column: {label_column}"
    )

    valid = df[label_column].notna()

    true_labels = df.loc[
        valid,
        label_column
    ].astype(str)

    predicted_labels = df.loc[
        valid,
        "cluster"
    ]

    ari = adjusted_rand_score(
        true_labels,
        predicted_labels
    )

    nmi = normalized_mutual_info_score(
        true_labels,
        predicted_labels
    )

    print()
    print(
        f"Adjusted Rand Index (ARI): "
        f"{ari:.4f}"
    )

    print(
        f"Normalized Mutual Information (NMI): "
        f"{nmi:.4f}"
    )

else:

    print(
        "No obvious annotation/label column "
        "was detected."
    )


# ============================================================
# CELL 34 — Cluster vs annotation cross-tabulation
# ============================================================

if len(possible_label_columns) > 0:

    label_column = possible_label_columns[0]

    cross_tab = pd.crosstab(
        df["cluster"],
        df[label_column]
    )

    print(
        f"Cluster vs {label_column}:"
    )

    display(cross_tab)


# ============================================================
# CELL 35 — Cluster vs annotation heatmap
# ============================================================

if len(possible_label_columns) > 0:

    label_column = possible_label_columns[0]

    cross_tab = pd.crosstab(
        df["cluster"],
        df[label_column]
    )

    plt.figure(
        figsize=(
            max(8, cross_tab.shape[1] * 0.8),
            max(5, cross_tab.shape[0] * 0.7)
        )
    )

    sns.heatmap(
        cross_tab,
        annot=True,
        fmt="d"
    )

    plt.xlabel(label_column)
    plt.ylabel("K-Means Cluster")
    plt.title(
        "K-Means Clusters vs Annotation Labels"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# CELL 36 — Save clustered dataset
# ============================================================

df.to_csv(
    OUTPUT_FILE,
    index=False,
    encoding="utf-8-sig"
)

print(
    f"Clustered dataset saved as:\n"
    f"{OUTPUT_FILE}"
)


# ============================================================
# CELL 37 — Save cluster feature summary
# ============================================================

FEATURE_OUTPUT = (
    "kmeans_cluster_features.csv"
)

feature_table.to_csv(
    FEATURE_OUTPUT,
    encoding="utf-8-sig"
)

print(
    f"Cluster feature summary saved as:\n"
    f"{FEATURE_OUTPUT}"
)


# ============================================================
# CELL 38 — Save cluster summary
# ============================================================

SUMMARY_OUTPUT = (
    "kmeans_cluster_summary.csv"
)

cluster_summary.to_csv(
    SUMMARY_OUTPUT,
    index=False,
    encoding="utf-8-sig"
)

print(
    f"Cluster summary saved as:\n"
    f"{SUMMARY_OUTPUT}"
)


# ============================================================
# CELL 39 — Final results
# ============================================================

print()
print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(
    f"Original dataset rows: {before}"
)

print(
    f"Rows used for clustering: {len(df)}"
)

print(
    f"Number of TF-IDF features: {X.shape[1]}"
)

print(
    f"Optimal K: {best_k}"
)

print(
    f"Silhouette Score: {final_silhouette:.4f}"
)

print()
print("Cluster sizes:")

display(cluster_summary)

print()
print("Output files:")
print(f"1. {OUTPUT_FILE}")
print(f"2. {FEATURE_OUTPUT}")
print(f"3. {SUMMARY_OUTPUT}")


# ============================================================
# CELL 40 — Display final dataset
# ============================================================

display(df.head(20))
```
