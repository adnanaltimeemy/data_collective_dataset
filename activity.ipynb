```python
# Install if needed:
# !pip install pandas scikit-learn nltk matplotlib

import pandas as pd
import re
import nltk
import matplotlib.pyplot as plt

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

nltk.download("stopwords")

# 1. Load dataset
df = pd.read_csv("activity.csv")

# 2. Select text column
text = df["activity"].fillna("").astype(str)

# 3. NLP cleaning
stop = set(stopwords.words("english"))
text = text.apply(lambda x: " ".join(
    w for w in re.sub(r"[^a-zA-Z\s]", "", x.lower()).split()
    if w not in stop
))

# 4. Convert text to numbers
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(text)

# 5. K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# 6. Show results
print(df[["activity", "Cluster"]].head(20))

# 7. Show important words in each cluster
words = tfidf.get_feature_names_out()

for i in range(k):
    top = model.cluster_centers_[i].argsort()[-10:][::-1]
    print(f"\nCluster {i}:",
          ", ".join(words[j] for j in top))

# 8. Save results
df.to_csv("activity_clustered.csv", index=False)
print("\nSaved as activity_clustered.csv")
```
