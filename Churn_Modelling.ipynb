```python
# ============================================================
# CUSTOMER CHURN ANALYSIS USING K-MEANS + NLP
# Dataset: Churn_Modelling.csv
# ============================================================

# 1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# NLP libraries
import re
from collections import Counter

# ------------------------------------------------------------
# 2. Load the dataset
# ------------------------------------------------------------

df = pd.read_csv("Churn_Modelling.csv")

print("Dataset shape:", df.shape)
display(df.head())

# ------------------------------------------------------------
# 3. Basic dataset information
# ------------------------------------------------------------

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

# ------------------------------------------------------------
# 4. Remove unnecessary columns
# ------------------------------------------------------------

# RowNumber, CustomerId and Surname are normally identifiers,
# rather than useful numerical clustering variables.

columns_to_drop = ["RowNumber", "CustomerId", "Surname"]

df_model = df.drop(
    columns=[col for col in columns_to_drop if col in df.columns]
).copy()

display(df_model.head())

# ------------------------------------------------------------
# 5. Convert categorical variables to numerical variables
# ------------------------------------------------------------

# One-hot encode Geography and Gender
categorical_columns = [
    col for col in ["Geography", "Gender"]
    if col in df_model.columns
]

df_encoded = pd.get_dummies(
    df_model,
    columns=categorical_columns,
    drop_first=True
)

print("Encoded dataset:")
display(df_encoded.head())

# ------------------------------------------------------------
# 6. Separate target variable
# ------------------------------------------------------------

# 'Exited' is the churn indicator:
# 0 = customer stayed
# 1 = customer exited

if "Exited" in df_encoded.columns:
    X = df_encoded.drop("Exited", axis=1)
    y = df_encoded["Exited"]
else:
    X = df_encoded.copy()
    y = None

# Convert boolean columns to integers
X = X.astype(float)

print("Features used for clustering:")
print(X.columns.tolist())

# ------------------------------------------------------------
# 7. Standardise the data
# ------------------------------------------------------------

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaled data shape:", X_scaled.shape)

# ------------------------------------------------------------
# 8. Find the best number of K-Means clusters
#    using the Elbow Method
# ------------------------------------------------------------

inertia = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Choosing K")
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 9. Calculate silhouette scores
# ------------------------------------------------------------

silhouette_scores = []

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

plt.figure(figsize=(8, 5))
plt.plot(K_range, silhouette_scores, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")
plt.grid(True)
plt.show()

# Display scores
for k, score in zip(K_range, silhouette_scores):
    print(f"K = {k}: Silhouette Score = {score:.4f}")

# ------------------------------------------------------------
# 10. Apply K-Means
# ------------------------------------------------------------

# Change this value if your elbow/silhouette analysis
# suggests a different number of clusters.
optimal_k = 4

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

# Add cluster labels to the dataset
df_model["Cluster"] = clusters

print("\nCluster counts:")
print(df_model["Cluster"].value_counts().sort_index())

# ------------------------------------------------------------
# 11. Visualise clusters using PCA
# ------------------------------------------------------------

pca = PCA(n_components=2, random_state=42)

X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=clusters,
    alpha=0.6
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Customer Clusters")
plt.colorbar(scatter, label="Cluster")
plt.show()

print(
    "Explained variance:",
    pca.explained_variance_ratio_
)

# ------------------------------------------------------------
# 12. Analyse each cluster
# ------------------------------------------------------------

cluster_summary = df_model.groupby("Cluster").mean(numeric_only=True)

display(cluster_summary)

# ------------------------------------------------------------
# 13. Number of customers in each cluster
# ------------------------------------------------------------

cluster_sizes = (
    df_model["Cluster"]
    .value_counts()
    .sort_index()
)

print("Customers per cluster:")
display(cluster_sizes)

plt.figure(figsize=(8, 5))
cluster_sizes.plot(kind="bar")

plt.xlabel("Cluster")
plt.ylabel("Number of Customers")
plt.title("Number of Customers in Each Cluster")
plt.xticks(rotation=0)
plt.show()

# ------------------------------------------------------------
# 14. Churn rate by cluster
# ------------------------------------------------------------

if y is not None:

    # Add original churn value
    df_model["Exited"] = y.values

    churn_by_cluster = (
        df_model.groupby("Cluster")["Exited"]
        .mean()
        .sort_index()
    )

    print("Churn rate by cluster:")
    display(churn_by_cluster)

    plt.figure(figsize=(8, 5))

    churn_by_cluster.plot(kind="bar")

    plt.xlabel("Cluster")
    plt.ylabel("Churn Rate")
    plt.title("Customer Churn Rate by K-Means Cluster")
    plt.xticks(rotation=0)
    plt.show()

# ============================================================
# NLP SECTION
# ============================================================

# The standard Churn_Modelling dataset usually has very little
# natural-language text. The 'Surname' column is a name, not
# meaningful customer-review text.
#
# Therefore, NLP is demonstrated separately using the available
# text field. If your CSV has a Reviews/Comments/Feedback column,
# replace 'Surname' below with that column.

# ------------------------------------------------------------
# 15. Check text columns
# ------------------------------------------------------------

text_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

print("Text/categorical columns:")
print(text_columns)

# ------------------------------------------------------------
# 16. Basic NLP preprocessing
# ------------------------------------------------------------

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[^a-z\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

# Example using Surname if no feedback/review column exists
if "Feedback" in df.columns:
    text_column = "Feedback"
elif "Review" in df.columns:
    text_column = "Review"
elif "Comment" in df.columns:
    text_column = "Comment"
else:
    text_column = "Surname"

df["Clean_Text"] = df[text_column].apply(clean_text)

display(
    df[[text_column, "Clean_Text"]].head(10)
)

# ------------------------------------------------------------
# 17. Tokenisation
# ------------------------------------------------------------

df["Tokens"] = df["Clean_Text"].apply(
    lambda x: x.split()
)

display(
    df[[text_column, "Tokens"]].head(10)
)

# ------------------------------------------------------------
# 18. Word frequency analysis
# ------------------------------------------------------------

all_words = []

for tokens in df["Tokens"]:
    all_words.extend(tokens)

word_frequency = Counter(all_words)

print("Most common words:")
print(word_frequency.most_common(20))

# ------------------------------------------------------------
# 19. Plot word frequency
# ------------------------------------------------------------

common_words = word_frequency.most_common(15)

words = [item[0] for item in common_words]
frequencies = [item[1] for item in common_words]

plt.figure(figsize=(10, 5))

plt.bar(words, frequencies)

plt.xlabel("Word")
plt.ylabel("Frequency")
plt.title("Most Frequent Words")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 20. NLP with TF-IDF
# ------------------------------------------------------------

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=100
)

X_text = tfidf.fit_transform(
    df["Clean_Text"]
)

print("TF-IDF matrix shape:", X_text.shape)

# ------------------------------------------------------------
# 21. K-Means clustering using NLP features
# ------------------------------------------------------------

if X_text.shape[1] > 0:

    nlp_kmeans = KMeans(
        n_clusters=optimal_k,
        random_state=42,
        n_init=10
    )

    nlp_clusters = nlp_kmeans.fit_predict(X_text)

    df["NLP_Cluster"] = nlp_clusters

    print("\nNLP Cluster counts:")
    print(
        df["NLP_Cluster"]
        .value_counts()
        .sort_index()
    )

# ------------------------------------------------------------
# 22. Display important TF-IDF words
# ------------------------------------------------------------

feature_names = tfidf.get_feature_names_out()

for cluster_number in range(optimal_k):

    centroid = nlp_kmeans.cluster_centers_[cluster_number]

    top_indices = centroid.argsort()[-10:][::-1]

    top_words = [
        feature_names[i]
        for i in top_indices
    ]

    print(
        f"\nCluster {cluster_number} important words:"
    )
    print(top_words)

# ------------------------------------------------------------
# 23. Save the results
# ------------------------------------------------------------

df_model.to_csv(
    "Churn_KMeans_Results.csv",
    index=False
)

df.to_csv(
    "Churn_NLP_Results.csv",
    index=False
)

print("Analysis completed successfully!")
print("Files created:")
print("- Churn_KMeans_Results.csv")
print("- Churn_NLP_Results.csv")
```
