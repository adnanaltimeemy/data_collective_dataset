```python
import pandas as pd
import numpy as np
import random

# Reproducibility
np.random.seed(42)
random.seed(42)

# Number of employees
n = 1470

# ---------------------------------------------------------
# Employee basic information
# ---------------------------------------------------------

employee_numbers = np.arange(1, n + 1)

ages = np.random.randint(18, 61, n)

genders = np.random.choice(
    ["Male", "Female"],
    size=n,
    p=[0.60, 0.40]
)

marital_status = np.random.choice(
    ["Single", "Married", "Divorced"],
    size=n,
    p=[0.32, 0.47, 0.21]
)

departments = np.random.choice(
    ["Sales", "Research & Development", "Human Resources"],
    size=n,
    p=[0.30, 0.65, 0.05]
)

jobs = np.random.choice(
    [
        "Sales Executive",
        "Research Scientist",
        "Laboratory Technician",
        "Manufacturing Director",
        "Healthcare Representative",
        "Manager",
        "Sales Representative",
        "Research Director",
        "Human Resources"
    ],
    size=n
)

# ---------------------------------------------------------
# Employment information
# ---------------------------------------------------------

years_at_company = np.random.randint(0, 41, n)

years_in_current_role = np.minimum(
    years_at_company,
    np.random.randint(0, 19, n)
)

years_since_last_promotion = np.minimum(
    years_at_company,
    np.random.randint(0, 16, n)
)

years_with_current_manager = np.minimum(
    years_at_company,
    np.random.randint(0, 18, n)
)

total_working_years = np.maximum(
    ages - np.random.randint(18, 26, n),
    years_at_company
)

num_companies_worked = np.random.randint(0, 10, n)

# ---------------------------------------------------------
# Salary and compensation
# ---------------------------------------------------------

monthly_income = np.random.randint(1000, 20000, n)

monthly_rate = np.random.randint(2000, 27000, n)

daily_rate = np.random.randint(100, 1500, n)

hourly_rate = np.random.randint(30, 110, n)

percent_salary_hike = np.random.randint(11, 26, n)

stock_option_level = np.random.randint(0, 4, n)

# ---------------------------------------------------------
# Work-related variables
# ---------------------------------------------------------

business_travel = np.random.choice(
    ["Non-Travel", "Travel_Rarely", "Travel_Frequently"],
    size=n,
    p=[0.10, 0.71, 0.19]
)

overtime = np.random.choice(
    ["Yes", "No"],
    size=n,
    p=[0.28, 0.72]
)

distance_from_home = np.random.randint(1, 30, n)

job_level = np.random.randint(1, 6, n)

job_involvement = np.random.randint(1, 5, n)

job_satisfaction = np.random.randint(1, 5, n)

environment_satisfaction = np.random.randint(1, 5, n)

relationship_satisfaction = np.random.randint(1, 5, n)

work_life_balance = np.random.randint(1, 5, n)

performance_rating = np.random.choice(
    [3, 4],
    size=n,
    p=[0.85, 0.15]
)

training_times_last_year = np.random.randint(0, 7, n)

num_projects = np.random.randint(1, 11, n)

average_weekly_hours = np.random.randint(30, 61, n)

# ---------------------------------------------------------
# Education
# ---------------------------------------------------------

education = np.random.choice(
    [
        "Below College",
        "College",
        "Bachelor",
        "Master",
        "Doctor"
    ],
    size=n,
    p=[0.10, 0.20, 0.40, 0.25, 0.05]
)

education_field = np.random.choice(
    [
        "Life Sciences",
        "Medical",
        "Marketing",
        "Technical Degree",
        "Human Resources",
        "Other"
    ],
    size=n
)

# ---------------------------------------------------------
# Employee attrition
# ---------------------------------------------------------

# Create a probability of attrition influenced by
# overtime, satisfaction, income, distance, etc.

attrition_score = (
    0.8 * (overtime == "Yes").astype(int)
    + 0.5 * (job_satisfaction <= 2).astype(int)
    + 0.5 * (environment_satisfaction <= 2).astype(int)
    + 0.4 * (work_life_balance <= 2).astype(int)
    + 0.3 * (distance_from_home > 20).astype(int)
    + 0.4 * (years_at_company <= 2).astype(int)
    - 0.3 * (job_involvement >= 3).astype(int)
    - 0.2 * (monthly_income > 8000).astype(int)
)

attrition_probability = 1 / (1 + np.exp(-(attrition_score - 1.3)))

attrition = np.where(
    np.random.random(n) < attrition_probability,
    "Yes",
    "No"
)

# ---------------------------------------------------------
# NLP / Text column
# ---------------------------------------------------------

positive_comments = [
    "Good work environment and supportive manager",
    "Excellent opportunities for career growth",
    "Great team and positive workplace",
    "I enjoy working with my colleagues",
    "The company provides good training",
    "Flexible working environment",
    "Management is supportive and helpful"
]

negative_comments = [
    "High workload and frequent overtime",
    "Limited opportunities for promotion",
    "Poor work life balance",
    "Salary is not competitive",
    "Management needs improvement",
    "Long working hours are stressful",
    "I feel there is limited career growth"
]

neutral_comments = [
    "The job is satisfactory",
    "Work environment is average",
    "The workload is manageable",
    "Training is adequate",
    "The role meets expectations"
]

comments = []

for i in range(n):

    if attrition[i] == "Yes":
        comments.append(random.choice(negative_comments))
    elif random.random() < 0.6:
        comments.append(random.choice(positive_comments))
    else:
        comments.append(random.choice(neutral_comments))


# ---------------------------------------------------------
# Create DataFrame
# ---------------------------------------------------------

df = pd.DataFrame({
    "EmployeeNumber": employee_numbers,
    "Age": ages,
    "Gender": genders,
    "MaritalStatus": marital_status,
    "Department": departments,
    "JobRole": jobs,
    "BusinessTravel": business_travel,
    "OverTime": overtime,
    "Education": education,
    "EducationField": education_field,

    "MonthlyIncome": monthly_income,
    "MonthlyRate": monthly_rate,
    "DailyRate": daily_rate,
    "HourlyRate": hourly_rate,
    "PercentSalaryHike": percent_salary_hike,
    "StockOptionLevel": stock_option_level,

    "DistanceFromHome": distance_from_home,
    "JobLevel": job_level,
    "JobInvolvement": job_involvement,
    "JobSatisfaction": job_satisfaction,
    "EnvironmentSatisfaction": environment_satisfaction,
    "RelationshipSatisfaction": relationship_satisfaction,
    "WorkLifeBalance": work_life_balance,
    "PerformanceRating": performance_rating,

    "YearsAtCompany": years_at_company,
    "YearsInCurrentRole": years_in_current_role,
    "YearsSinceLastPromotion": years_since_last_promotion,
    "YearsWithCurrentManager": years_with_current_manager,
    "TotalWorkingYears": total_working_years,
    "NumCompaniesWorked": num_companies_worked,

    "TrainingTimesLastYear": training_times_last_year,
    "NumProjects": num_projects,
    "AverageWeeklyHours": average_weekly_hours,

    "Attrition": attrition,

    # NLP column
    "EmployeeComment": comments
})

# ---------------------------------------------------------
# Save dataset
# ---------------------------------------------------------

df.to_csv(
    "HR-Employee-Attrition-All.csv",
    index=False
)

print("Dataset created successfully!")
print("File: HR-Employee-Attrition-All.csv")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```
