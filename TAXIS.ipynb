```python
# ============================================================
# TAXI DATASET ANALYSIS
# K-MEANS CLUSTERING + NLP
# ============================================================


# ============================================================
# CELL 1 - INSTALL / IMPORT LIBRARIES
# ============================================================

# If required, uncomment and run:
# !pip install pandas numpy matplotlib seaborn scikit-learn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.feature_extraction.text import TfidfVectorizer

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")


# ============================================================
# CELL 2 - LOAD DATASET
# ============================================================

file_path = "taxis.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())


# ============================================================
# CELL 3 - DATASET INFORMATION
# ============================================================

print("Dataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()


# ============================================================
# CELL 4 - FIRST 10 ROWS
# ============================================================

display(df.head(10))


# ============================================================
# CELL 5 - LAST 10 ROWS
# ============================================================

display(df.tail(10))


# ============================================================
# CELL 6 - CHECK MISSING VALUES
# ============================================================

missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values,
    "Missing Percentage": 
        (missing_values.values / len(df)) * 100
})

display(
    missing_table.sort_values(
        by="Missing Values",
        ascending=False
    )
)


# ============================================================
# CELL 7 - CHECK DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")

print("New dataset shape:", df.shape)


# ============================================================
# CELL 8 - DATA TYPES
# ============================================================

print("Data types:")

display(
    pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str).values
    })
)


# ============================================================
# CELL 9 - STATISTICAL SUMMARY
# ============================================================

print("Numerical statistical summary:")

display(df.describe())


# ============================================================
# CELL 10 - UNIQUE VALUES
# ============================================================

for column in df.columns:
    print("\n===================================")
    print("Column:", column)
    print("Unique values:", df[column].nunique())

    if df[column].nunique() <= 20:
        print(df[column].unique())


# ============================================================
# CELL 11 - IDENTIFY NUMERICAL COLUMNS
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical columns:")
print(numeric_columns)


# ============================================================
# CELL 12 - IDENTIFY TEXT COLUMNS
# ============================================================

text_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

print("Text / categorical columns:")
print(text_columns)


# ============================================================
# CELL 13 - NUMERICAL DATA DISTRIBUTIONS
# ============================================================

for column in numeric_columns:

    plt.figure(figsize=(8, 5))

    sns.histplot(
        df[column].dropna(),
        kde=True
    )

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.show()


# ============================================================
# CELL 14 - CORRELATION MATRIX
# ============================================================

if len(numeric_columns) >= 2:

    plt.figure(
        figsize=(12, 8)
    )

    correlation = df[numeric_columns].corr()

    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Matrix")
    plt.show()


# ============================================================
# CELL 15 - AUTOMATICALLY SELECT FEATURES FOR K-MEANS
# ============================================================

# Common taxi-related column names.
# The code will use only columns that actually exist.

preferred_features = [
    "distance",
    "fare",
    "tip",
    "tolls",
    "total",
    "duration",
    "passengers",
    "pickup",
    "dropoff"
]

features = []

for column in preferred_features:

    if column in df.columns:

        if pd.api.types.is_numeric_dtype(df[column]):
            features.append(column)


# If fewer than 2 preferred numerical columns exist,
# use numerical columns automatically.

if len(features) < 2:

    features = numeric_columns.copy()

print("Features selected for K-Means:")
print(features)


# ============================================================
# CELL 16 - PREPARE K-MEANS DATA
# ============================================================

X = df[features].copy()

# Convert everything to numeric
for column in features:
    X[column] = pd.to_numeric(
        X[column],
        errors="coerce"
    )

# Replace infinity values
X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

# Fill missing numerical values with median
for column in X.columns:

    median_value = X[column].median()

    X[column] = X[column].fillna(
        median_value
    )

print("K-Means input data:")
display(X.head())


# ============================================================
# CELL 17 - CHECK NEGATIVE VALUES
# ============================================================

print("Minimum values:")

display(
    X.min().to_frame("Minimum")
)


# ============================================================
# CELL 18 - STANDARDIZE FEATURES
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=features,
    index=df.index
)

print("Standardized data:")
display(X_scaled.head())


# ============================================================
# CELL 19 - ELBOW METHOD
# ============================================================

inertia = []

K_range = range(2, 11)

for k in K_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)

    inertia.append(
        model.inertia_
    )


plt.figure(figsize=(10, 6))

plt.plot(
    K_range,
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.xticks(K_range)

plt.grid(True)

plt.show()


# ============================================================
# CELL 20 - SILHOUETTE SCORE
# ============================================================

silhouette_scores = []

for k in K_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(
        X_scaled
    )

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k}, "
        f"Silhouette Score = {score:.4f}"
    )


# Plot silhouette scores

plt.figure(figsize=(10, 6))

plt.plot(
    K_range,
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.title(
    "Silhouette Score for Different K Values"
)

plt.xticks(K_range)

plt.grid(True)

plt.show()


# ============================================================
# CELL 21 - SELECT BEST K
# ============================================================

best_k = K_range[
    np.argmax(silhouette_scores)
]

print("Best K according to silhouette score:", best_k)

# You can manually change this value if the Elbow Method
# suggests a different number.

K = best_k

print("Final number of clusters:", K)


# ============================================================
# CELL 22 - FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=K,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(
    X_scaled
)

df["KMeans_Cluster"] = clusters

print("K-Means clustering completed.")


# ============================================================
# CELL 23 - CLUSTER COUNTS
# ============================================================

cluster_counts = (
    df["KMeans_Cluster"]
    .value_counts()
    .sort_index()
)

print("Number of taxi trips in each cluster:")

display(
    cluster_counts.to_frame(
        "Number of Trips"
    )
)


# ============================================================
# CELL 24 - CLUSTER PERCENTAGES
# ============================================================

cluster_percentage = (
    df["KMeans_Cluster"]
    .value_counts(
        normalize=True
    )
    .sort_index()
    * 100
)

display(
    cluster_percentage.to_frame(
        "Percentage"
    )
)


# ============================================================
# CELL 25 - CLUSTER VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="KMeans_Cluster"
)

plt.title("Number of Taxi Trips per Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Trips")

plt.show()


# ============================================================
# CELL 26 - CLUSTER PROFILE
# ============================================================

cluster_profile = (
    df.groupby("KMeans_Cluster")[features]
    .mean()
)

print("Average characteristics of each cluster:")

display(cluster_profile)


# ============================================================
# CELL 27 - MEDIAN CLUSTER PROFILE
# ============================================================

cluster_median = (
    df.groupby("KMeans_Cluster")[features]
    .median()
)

print("Median characteristics of each cluster:")

display(cluster_median)


# ============================================================
# CELL 28 - PCA FOR K-MEANS
# ============================================================

pca = PCA(
    n_components=2
)

X_pca = pca.fit_transform(
    X_scaled
)

print(
    "Explained variance:",
    pca.explained_variance_ratio_
)

print(
    "Total explained variance:",
    pca.explained_variance_ratio_.sum()
)


# ============================================================
# CELL 29 - PCA CLUSTER PLOT
# ============================================================

plt.figure(
    figsize=(10, 7)
)

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=df["KMeans_Cluster"],
    alpha=0.6
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title(
    "K-Means Taxi Clusters Using PCA"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.show()


# ============================================================
# CELL 30 - ADD PCA VALUES TO DATASET
# ============================================================

df["PCA_1"] = X_pca[:, 0]
df["PCA_2"] = X_pca[:, 1]

display(
    df[
        [
            "KMeans_Cluster",
            "PCA_1",
            "PCA_2"
        ]
    ].head()
)


# ============================================================
# ============================================================
# NLP SECTION
# ============================================================
# ============================================================


# ============================================================
# CELL 31 - FIND TEXT COLUMNS
# ============================================================

text_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

print("Text columns found:")

for column in text_columns:
    print("-", column)


# ============================================================
# CELL 32 - CHECK TEXT COLUMN CONTENT
# ============================================================

for column in text_columns:

    print("\n================================")
    print("COLUMN:", column)
    print("================================")

    display(
        df[column]
        .dropna()
        .astype(str)
        .head(10)
    )


# ============================================================
# CELL 33 - CREATE NLP TEXT DATA
# ============================================================

if len(text_columns) > 0:

    df["NLP_Text"] = (
        df[text_columns]
        .fillna("")
        .astype(str)
        .agg(" ".join, axis=1)
    )

else:

    # If there are no text columns,
    # create text from useful taxi information.

    df["NLP_Text"] = ""

    for column in features:

        df["NLP_Text"] += (
            " "
            + df[column]
            .astype(str)
        )


print("NLP text created.")

display(
    df[["NLP_Text"]].head(10)
)


# ============================================================
# CELL 34 - CLEAN TEXT
# ============================================================

import re

def clean_text(text):

    text = str(text).lower()

    # Remove numbers
    text = re.sub(
        r"\d+",
        " ",
        text
    )

    # Remove punctuation
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


df["Clean_Text"] = (
    df["NLP_Text"]
    .apply(clean_text)
)

display(
    df[
        [
            "NLP_Text",
            "Clean_Text"
        ]
    ].head(10)
)


# ============================================================
# CELL 35 - TF-IDF
# ============================================================

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=1000,
    min_df=2
)

tfidf_matrix = vectorizer.fit_transform(
    df["Clean_Text"]
)

print(
    "TF-IDF matrix shape:",
    tfidf_matrix.shape
)


# ============================================================
# CELL 36 - GET TF-IDF WORDS
# ============================================================

words = vectorizer.get_feature_names_out()

print("Number of words:", len(words))

print("\nFirst 50 words:")

print(
    words[:50]
)


# ============================================================
# CELL 37 - NLP K-MEANS
# ============================================================

nlp_kmeans = KMeans(
    n_clusters=K,
    random_state=42,
    n_init=10
)

nlp_clusters = nlp_kmeans.fit_predict(
    tfidf_matrix
)

df["NLP_Cluster"] = nlp_clusters

print("NLP K-Means completed.")


# ============================================================
# CELL 38 - NLP CLUSTER COUNTS
# ============================================================

nlp_cluster_counts = (
    df["NLP_Cluster"]
    .value_counts()
    .sort_index()
)

display(
    nlp_cluster_counts.to_frame(
        "Number of Records"
    )
)


# ============================================================
# CELL 39 - IMPORTANT WORDS FOR EACH NLP CLUSTER
# ============================================================

terms = vectorizer.get_feature_names_out()

for cluster_number in range(K):

    print("\n")
    print("=" * 60)
    print(
        f"NLP CLUSTER {cluster_number}"
    )
    print("=" * 60)

    cluster_center = (
        nlp_kmeans
        .cluster_centers_[cluster_number]
    )

    top_indices = (
        cluster_center
        .argsort()[-15:][::-1]
    )

    top_words = [
        terms[index]
        for index in top_indices
    ]

    print(
        "Important words:"
    )

    print(
        ", ".join(top_words)
    )


# ============================================================
# CELL 40 - SILHOUETTE SCORE FOR NLP
# ============================================================

if K > 1:

    nlp_silhouette = silhouette_score(
        tfidf_matrix,
        nlp_clusters
    )

    print(
        "NLP Silhouette Score:",
        round(nlp_silhouette, 4)
    )


# ============================================================
# CELL 41 - PCA FOR NLP DATA
# ============================================================

# Convert sparse TF-IDF matrix to a 2D representation

nlp_pca = PCA(
    n_components=2,
    random_state=42
)

tfidf_dense = tfidf_matrix.toarray()

nlp_pca_result = nlp_pca.fit_transform(
    tfidf_dense
)

print(
    "NLP explained variance:",
    nlp_pca.explained_variance_ratio_
)


# ============================================================
# CELL 42 - NLP CLUSTER VISUALIZATION
# ============================================================

plt.figure(
    figsize=(10, 7)
)

scatter = plt.scatter(
    nlp_pca_result[:, 0],
    nlp_pca_result[:, 1],
    c=df["NLP_Cluster"],
    alpha=0.6
)

plt.xlabel("NLP Principal Component 1")
plt.ylabel("NLP Principal Component 2")

plt.title(
    "NLP K-Means Clusters"
)

plt.colorbar(
    scatter,
    label="NLP Cluster"
)

plt.show()


# ============================================================
# CELL 43 - ADD NLP PCA TO DATASET
# ============================================================

df["NLP_PCA_1"] = (
    nlp_pca_result[:, 0]
)

df["NLP_PCA_2"] = (
    nlp_pca_result[:, 1]
)


# ============================================================
# CELL 44 - COMPARE NUMERICAL AND NLP CLUSTERS
# ============================================================

cluster_comparison = pd.crosstab(
    df["KMeans_Cluster"],
    df["NLP_Cluster"]
)

print(
    "Comparison between numerical K-Means "
    "and NLP clusters:"
)

display(cluster_comparison)


# ============================================================
# CELL 45 - HEATMAP OF CLUSTER COMPARISON
# ============================================================

plt.figure(
    figsize=(10, 6)
)

sns.heatmap(
    cluster_comparison,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Numerical K-Means vs NLP Clusters"
)

plt.xlabel("NLP Cluster")
plt.ylabel("K-Means Cluster")

plt.show()


# ============================================================
# CELL 46 - EXAMPLE RECORDS FROM EACH CLUSTER
# ============================================================

for cluster in sorted(
    df["KMeans_Cluster"].unique()
):

    print("\n")
    print("=" * 60)
    print(
        f"K-MEANS CLUSTER {cluster}"
    )
    print("=" * 60)

    cluster_data = df[
        df["KMeans_Cluster"] == cluster
    ]

    display(
        cluster_data.head(5)
    )


# ============================================================
# CELL 47 - NLP EXAMPLES FROM EACH CLUSTER
# ============================================================

for cluster in sorted(
    df["NLP_Cluster"].unique()
):

    print("\n")
    print("=" * 60)
    print(
        f"NLP CLUSTER {cluster}"
    )
    print("=" * 60)

    cluster_data = df[
        df["NLP_Cluster"] == cluster
    ]

    display(
        cluster_data[
            [
                "NLP_Text",
                "Clean_Text",
                "NLP_Cluster"
            ]
        ].head(5)
    )


# ============================================================
# CELL 48 - OUTLIER CHECK
# ============================================================

print("Checking numerical data for possible outliers...")

for column in features:

    Q1 = X[column].quantile(0.25)
    Q3 = X[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = X[
        (X[column] < lower_bound) |
        (X[column] > upper_bound)
    ]

    print(
        f"{column}: "
        f"{len(outliers)} possible outliers"
    )


# ============================================================
# CELL 49 - SAVE K-MEANS RESULTS
# ============================================================

kmeans_output = "taxis_kmeans_results.csv"

df.to_csv(
    kmeans_output,
    index=False
)

print(
    "Saved:",
    kmeans_output
)


# ============================================================
# CELL 50 - SAVE NLP RESULTS
# ============================================================

nlp_output = "taxis_nlp_results.csv"

df.to_csv(
    nlp_output,
    index=False
)

print(
    "Saved:",
    nlp_output
)


# ============================================================
# CELL 51 - FINAL DATASET
# ============================================================

print(
    "Final dataset shape:",
    df.shape
)

print("\nFinal columns:")

for column in df.columns:
    print("-", column)

display(
    df.head(20)
)


# ============================================================
# CELL 52 - FINAL CLUSTER SUMMARY
# ============================================================

final_summary = (
    df.groupby("KMeans_Cluster")
    .agg(
        Number_of_Trips=(
            "KMeans_Cluster",
            "size"
        )
    )
)

# Add numerical averages
for column in features:

    final_summary[
        f"Average_{column}"
    ] = (
        df.groupby("KMeans_Cluster")[column]
        .mean()
    )

display(final_summary)


# ============================================================
# CELL 53 - FINAL NLP SUMMARY
# ============================================================

nlp_summary = (
    df.groupby("NLP_Cluster")
    .size()
    .reset_index(
        name="Number_of_Records"
    )
)

print("NLP cluster summary:")

display(nlp_summary)


# ============================================================
# CELL 54 - FINAL RESULTS
# ============================================================

print("=" * 70)
print("TAXI DATASET ANALYSIS COMPLETE")
print("=" * 70)

print(
    "\nNumber of records:",
    len(df)
)

print(
    "Number of numerical K-Means clusters:",
    K
)

print(
    "K-Means silhouette score:",
    round(
        max(silhouette_scores),
        4
    )
)

if K > 1:
    print(
        "NLP silhouette score:",
        round(
            nlp_silhouette,
            4
        )
    )

print("\nOutput files:")
print("1. taxis_kmeans_results.csv")
print("2. taxis_nlp_results.csv")

print("\nAnalysis finished successfully.")
```
