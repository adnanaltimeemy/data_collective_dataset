```python
# 1. Import libraries
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 2. Load dataset
df = pd.read_csv("crime.csv")
print(df.head())
print(df.columns)

# 3. Select text column
text_col = "Description"   # Change to your actual text column

df = df.dropna(subset=[text_col])

# 4. NLP cleaning
nltk.download("stopwords")
stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    text = " ".join(w for w in text.split() if w not in stop_words)
    return text

df["clean_text"] = df[text_col].apply(clean_text)

# 5. Convert text to TF-IDF
tfidf = TfidfVectorizer(max_features=1000)
X = tfidf.fit_transform(df["clean_text"])

# 6. K-Means clustering
k = 5
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# 7. Show clusters
for i in range(k):
    print("\nCluster", i)
    print(df[df["Cluster"] == i][text_col].head(5).to_list())

# 8. Cluster size
print("\nCluster Counts:")
print(df["Cluster"].value_counts().sort_index())

# 9. Save result
df.to_csv("crime_clustered.csv", index=False)

# 10. Simple visualisation
plt.scatter(X[:, 0].toarray(), X[:, 1].toarray(), c=df["Cluster"])
plt.title("Crime K-Means Clusters")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
