```python
# 1. Import libraries
import pandas as pd
import re
import nltk
import matplotlib.pyplot as plt
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

nltk.download("stopwords")

# 2. Load dataset
df = pd.read_csv("Sample - Coffee Chain.csv")
print(df.head())
print(df.columns)

# 3. NLP text cleaning
stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    return " ".join(w for w in text.split() if w not in stop_words)

df["Clean_Text"] = df["Review"].fillna("").apply(clean_text)

# 4. Convert text to TF-IDF
tfidf = TfidfVectorizer(max_features=500)
X = tfidf.fit_transform(df["Clean_Text"])

# 5. K-Means clustering
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# 6. Show results
print("\nCluster Counts:")
print(df["Cluster"].value_counts())

print("\nSample Results:")
print(df[["Review", "Cluster"]].head(10))

# 7. Important words in each cluster
terms = tfidf.get_feature_names_out()

for i in range(4):
    words = kmeans.cluster_centers_[i].argsort()[-10:][::-1]
    print(f"\nCluster {i}:")
    print([terms[j] for j in words])

# 8. Save results
df.to_csv("Coffee_Chain_KMeans_Result.csv", index=False)

print("\nDone! File saved as Coffee_Chain_KMeans_Result.csv")
```
