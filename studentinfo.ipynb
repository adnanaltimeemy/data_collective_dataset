```python
# NLP + K-Means on studentInfo.csv

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# Load data
df = pd.read_csv("studentInfo.csv")
print(df.head())
print(df.columns)

# Select text column (change if needed)
text_col = "Comments"
df[text_col] = df[text_col].fillna("").astype(str)

# NLP: TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(df[text_col])

# K-Means
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# Show clusters
print(df[["Cluster", text_col]].head(20))

# Important words in each cluster
words = tfidf.get_feature_names_out()

for i in range(k):
    top = model.cluster_centers_[i].argsort()[-10:][::-1]
    print(f"Cluster {i}:", [words[j] for j in top])

# Visualisation
pca = PCA(n_components=2)
X2 = pca.fit_transform(X.toarray())

plt.scatter(X2[:,0], X2[:,1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Student Clusters")
plt.show()

# Save result
df.to_csv("studentInfo_clustered.csv", index=False)
```
