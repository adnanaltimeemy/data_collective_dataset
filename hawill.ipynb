```python
# ============================================================
# HAWAIIAN TEXT DATASET - NLP + K-MEANS CLUSTERING
# ============================================================

# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

# Run this cell if the libraries are not already installed.
# Uncomment the following line if needed:

# !pip install pandas numpy matplotlib seaborn scikit-learn


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score


# ============================================================
# 3. LOAD THE DATASET
# ============================================================

file_name = "hawaiian.csv"

df = pd.read_csv(file_name)

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 4. BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDataset shape:")
print(df.shape)


# ============================================================
# 5. FIND THE TEXT COLUMN
# ============================================================

# Common possible names for a text column

possible_text_columns = [
    "text",
    "sentence",
    "description",
    "hawaiian",
    "word",
    "phrase",
    "content",
    "translation",
    "name"
]

text_column = None

# Look for a known text column
for column in possible_text_columns:
    if column in df.columns:
        text_column = column
        break

# If none of the above are found,
# automatically select the first object/string column
if text_column is None:

    string_columns = df.select_dtypes(
        include=["object"]
    ).columns

    if len(string_columns) > 0:
        text_column = string_columns[0]

    else:
        raise ValueError(
            "No text column was found in hawaiian.csv"
        )

print("\nText column selected:")
print(text_column)


# ============================================================
# 6. REMOVE MISSING TEXT VALUES
# ============================================================

df = df.dropna(
    subset=[text_column]
).copy()

# Convert text to string
df[text_column] = df[text_column].astype(str)

print("\nRows after removing missing text:")
print(len(df))


# ============================================================
# 7. TEXT CLEANING FUNCTION
# ============================================================

def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Keep:
    # a-z
    # Hawaiian vowels with macrons
    # ʻokina
    # apostrophes
    # spaces
    text = re.sub(
        r"[^a-zāēīōūʻ' ]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Remove spaces at beginning/end
    text = text.strip()

    return text


# Apply cleaning
df["clean_text"] = df[text_column].apply(clean_text)


# ============================================================
# 8. DISPLAY ORIGINAL AND CLEANED TEXT
# ============================================================

print("\n" + "=" * 60)
print("TEXT CLEANING RESULTS")
print("=" * 60)

display(
    df[
        [text_column, "clean_text"]
    ].head(10)
)


# ============================================================
# 9. REMOVE EMPTY TEXT RECORDS
# ============================================================

df = df[
    df["clean_text"].str.len() > 0
].copy()

df.reset_index(
    drop=True,
    inplace=True
)

print("\nRows remaining:")
print(len(df))


# ============================================================
# 10. CREATE TF-IDF FEATURES
# ============================================================

print("\n" + "=" * 60)
print("TF-IDF NLP")
print("=" * 60)

vectorizer = TfidfVectorizer(

    # Use individual words and two-word combinations
    ngram_range=(1, 2),

    # Ignore terms appearing in almost all documents
    max_df=0.95,

    # Keep terms appearing at least once
    min_df=1,

    # Convert to lowercase
    lowercase=True,

    # Remove accents is disabled because Hawaiian
    # macrons are meaningful
    strip_accents=None
)

X = vectorizer.fit_transform(
    df["clean_text"]
)

print("TF-IDF matrix created.")

print("Number of documents:", X.shape[0])
print("Number of TF-IDF features:", X.shape[1])


# ============================================================
# 11. DISPLAY TF-IDF FEATURES
# ============================================================

feature_names = vectorizer.get_feature_names_out()

print("\nFirst 30 TF-IDF features:")
print(feature_names[:30])


# ============================================================
# 12. FIND THE BEST NUMBER OF CLUSTERS
# ============================================================

print("\n" + "=" * 60)
print("FINDING BEST NUMBER OF CLUSTERS")
print("=" * 60)

silhouette_scores = []

# Maximum number of clusters to test
max_k = min(
    10,
    len(df) - 1
)

if max_k < 2:
    raise ValueError(
        "The dataset needs at least 3 text records."
    )


for k in range(2, max_k + 1):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k} | "
        f"Silhouette Score = {score:.4f}"
    )


# ============================================================
# 13. PLOT SILHOUETTE SCORES
# ============================================================

k_values = list(
    range(2, max_k + 1)
)

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    k_values,
    silhouette_scores,
    marker="o",
    linewidth=2
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "Silhouette Score for Selecting K"
)

plt.xticks(
    k_values
)

plt.grid(
    True,
    alpha=0.3
)

plt.show()


# ============================================================
# 14. SELECT BEST K
# ============================================================

best_index = np.argmax(
    silhouette_scores
)

best_k = k_values[
    best_index
]

best_score = silhouette_scores[
    best_index
]

print("\n" + "=" * 60)
print("BEST NUMBER OF CLUSTERS")
print("=" * 60)

print("Best K:", best_k)
print("Best Silhouette Score:", round(best_score, 4))


# ============================================================
# 15. TRAIN FINAL K-MEANS MODEL
# ============================================================

print("\n" + "=" * 60)
print("TRAINING FINAL K-MEANS MODEL")
print("=" * 60)

kmeans = KMeans(

    n_clusters=best_k,

    random_state=42,

    n_init=10
)

df["cluster"] = kmeans.fit_predict(
    X
)

print("K-Means training completed.")


# ============================================================
# 16. CLUSTER COUNTS
# ============================================================

print("\n" + "=" * 60)
print("NUMBER OF RECORDS IN EACH CLUSTER")
print("=" * 60)

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print(cluster_counts)


# ============================================================
# 17. PLOT CLUSTER COUNTS
# ============================================================

plt.figure(
    figsize=(10, 6)
)

sns.countplot(
    data=df,
    x="cluster"
)

plt.title(
    "Number of Hawaiian Text Records per Cluster"
)

plt.xlabel(
    "Cluster"
)

plt.ylabel(
    "Number of Records"
)

plt.show()


# ============================================================
# 18. GET IMPORTANT WORDS FOR EACH CLUSTER
# ============================================================

print("\n" + "=" * 60)
print("IMPORTANT WORDS IN EACH CLUSTER")
print("=" * 60)

terms = vectorizer.get_feature_names_out()

number_of_words = min(
    15,
    len(terms)
)

cluster_keywords = {}

for cluster_number in range(best_k):

    # Get cluster centre
    cluster_center = (
        kmeans.cluster_centers_[cluster_number]
    )

    # Sort features by importance
    top_indices = (
        cluster_center
        .argsort()[-number_of_words:][::-1]
    )

    top_words = [
        terms[index]
        for index in top_indices
    ]

    cluster_keywords[
        cluster_number
    ] = top_words

    print(
        f"\nCluster {cluster_number}:"
    )

    print(
        ", ".join(top_words)
    )


# ============================================================
# 19. DISPLAY CLUSTER RESULTS
# ============================================================

print("\n" + "=" * 60)
print("CLUSTERED DATA")
print("=" * 60)

display(
    df[
        [
            text_column,
            "clean_text",
            "cluster"
        ]
    ].sort_values(
        "cluster"
    )
)


# ============================================================
# 20. SHOW SAMPLE RECORDS FROM EACH CLUSTER
# ============================================================

print("\n" + "=" * 60)
print("SAMPLE RECORDS FROM EACH CLUSTER")
print("=" * 60)

for cluster_number in range(best_k):

    print(
        f"\n{'=' * 20} "
        f"CLUSTER {cluster_number} "
        f"{'=' * 20}"
    )

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    display(
        cluster_data[
            [text_column]
        ].head(10)
    )


# ============================================================
# 21. PCA FOR VISUALISATION
# ============================================================

print("\n" + "=" * 60)
print("PCA VISUALISATION")
print("=" * 60)

# PCA needs a dense matrix
X_dense = X.toarray()

pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(
    X_dense
)

df["PCA1"] = X_pca[:, 0]

df["PCA2"] = X_pca[:, 1]

print(
    "Explained variance:",
    pca.explained_variance_ratio_
)

print(
    "Total explained variance:",
    round(
        sum(pca.explained_variance_ratio_),
        4
    )
)


# ============================================================
# 22. PLOT K-MEANS CLUSTERS
# ============================================================

plt.figure(
    figsize=(12, 8)
)

sns.scatterplot(

    data=df,

    x="PCA1",

    y="PCA2",

    hue="cluster",

    palette="tab10",

    s=100,

    alpha=0.8
)

plt.title(
    "K-Means Clustering of Hawaiian Text"
)

plt.xlabel(
    "Principal Component 1"
)

plt.ylabel(
    "Principal Component 2"
)

plt.legend(
    title="Cluster"
)

plt.grid(
    True,
    alpha=0.3
)

plt.show()


# ============================================================
# 23. CALCULATE FINAL SILHOUETTE SCORE
# ============================================================

final_silhouette = silhouette_score(
    X,
    df["cluster"]
)

print("\n" + "=" * 60)
print("MODEL EVALUATION")
print("=" * 60)

print(
    "Final Silhouette Score:",
    round(final_silhouette, 4)
)


# ============================================================
# 24. CALCULATE WITHIN-CLUSTER SUM OF SQUARES
# ============================================================

print(
    "K-Means Inertia:",
    round(kmeans.inertia_, 4)
)


# ============================================================
# 25. CREATE CLUSTER SUMMARY
# ============================================================

cluster_summary = []

for cluster_number in range(best_k):

    cluster_size = len(
        df[
            df["cluster"] == cluster_number
        ]
    )

    keywords = ", ".join(
        cluster_keywords[
            cluster_number
        ][:10]
    )

    cluster_summary.append({

        "Cluster": cluster_number,

        "Number_of_records": cluster_size,

        "Important_words": keywords

    })


summary_df = pd.DataFrame(
    cluster_summary
)

print("\n" + "=" * 60)
print("CLUSTER SUMMARY")
print("=" * 60)

display(summary_df)


# ============================================================
# 26. SAVE CLUSTERED DATASET
# ============================================================

output_file = (
    "hawaiian_kmeans_results.csv"
)

df.to_csv(
    output_file,
    index=False
)

print(
    f"\nClustered dataset saved as: "
    f"{output_file}"
)


# ============================================================
# 27. SAVE CLUSTER SUMMARY
# ============================================================

summary_file = (
    "hawaiian_cluster_summary.csv"
)

summary_df.to_csv(
    summary_file,
    index=False
)

print(
    f"Cluster summary saved as: "
    f"{summary_file}"
)


# ============================================================
# 28. FINAL RESULTS
# ============================================================

print("\n" + "=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print(
    "Original dataset:",
    file_name
)

print(
    "Number of records:",
    len(df)
)

print(
    "Text column:",
    text_column
)

print(
    "TF-IDF features:",
    X.shape[1]
)

print(
    "Best number of clusters:",
    best_k
)

print(
    "Silhouette score:",
    round(final_silhouette, 4)
)

print(
    "Output file:",
    output_file
)

print(
    "Summary file:",
    summary_file
)

print("=" * 60)
print("DONE")
print("=" * 60)
```
