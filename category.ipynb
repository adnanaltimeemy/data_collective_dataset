```python
# ============================================================
# CATEGORY DEMO - NLP + K-MEANS CLUSTERING
# ============================================================


# ============================================================
# CELL 1: Install required libraries
# ============================================================

# Run this cell once in Jupyter Notebook

!pip install pandas numpy matplotlib scikit-learn nltk


# ============================================================
# CELL 2: Import libraries
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics import silhouette_score

import nltk
from nltk.corpus import stopwords

# Download English stopwords
nltk.download("stopwords")

print("Libraries imported successfully.")


# ============================================================
# CELL 3: Load the CSV dataset
# ============================================================

file_path = "category_demo.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 10 rows:")
display(df.head(10))


# ============================================================
# CELL 4: Dataset information
# ============================================================

print("Dataset information:")
df.info()

print("\n\nMissing values:")
print(df.isnull().sum())

print("\n\nNumber of duplicate rows:")
print(df.duplicated().sum())


# ============================================================
# CELL 5: Remove duplicate rows
# ============================================================

df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)


# ============================================================
# CELL 6: Identify the text column
# ============================================================

print("Available columns:")

for i, column in enumerate(df.columns):
    print(i, ":", column)


# ------------------------------------------------------------
# IMPORTANT:
#
# If your dataset has a column called "category", use:
#
# text_column = "category"
#
# If it has "description", use:
#
# text_column = "description"
#
# Change the line below if necessary.
# ------------------------------------------------------------

possible_columns = [
    "category",
    "text",
    "description",
    "name",
    "title",
    "product",
    "content"
]

text_column = None

for column in possible_columns:
    if column in df.columns:
        text_column = column
        break

# If none of the expected names exist,
# automatically select the first text column.

if text_column is None:

    object_columns = df.select_dtypes(
        include=["object"]
    ).columns

    if len(object_columns) == 0:
        raise ValueError(
            "No text column was found. "
            "Please specify the text column manually."
        )

    text_column = object_columns[0]

print("\nSelected text column:", text_column)


# ============================================================
# CELL 7: Create a working copy of the text
# ============================================================

df["text"] = df[text_column].fillna("").astype(str)

print("Text data:")
display(df[[text_column, "text"]].head(10))


# ============================================================
# CELL 8: NLP TEXT CLEANING
# ============================================================

# Convert text to lowercase
df["clean_text"] = df["text"].str.lower()

# Remove punctuation and special characters
df["clean_text"] = df["clean_text"].str.replace(
    r"[^a-zA-Z0-9\s]",
    " ",
    regex=True
)

# Remove numbers
df["clean_text"] = df["clean_text"].str.replace(
    r"\d+",
    " ",
    regex=True
)

# Remove extra spaces
df["clean_text"] = df["clean_text"].str.replace(
    r"\s+",
    " ",
    regex=True
)

# Remove leading/trailing spaces
df["clean_text"] = df["clean_text"].str.strip()

print("Cleaned text:")
display(
    df[
        [text_column, "clean_text"]
    ].head(10)
)


# ============================================================
# CELL 9: Remove English stopwords
# ============================================================

stop_words = set(
    stopwords.words("english")
)

def remove_stopwords(text):
    
    words = text.split()
    
    words = [
        word
        for word in words
        if word not in stop_words
    ]
    
    return " ".join(words)


df["clean_text"] = df["clean_text"].apply(
    remove_stopwords
)

print("After stopword removal:")
display(
    df[
        [text_column, "clean_text"]
    ].head(10)
)


# ============================================================
# CELL 10: Remove empty text rows
# ============================================================

before = len(df)

df = df[
    df["clean_text"].str.strip() != ""
].copy()

df = df.reset_index(drop=True)

after = len(df)

print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)


# ============================================================
# CELL 11: TF-IDF Vectorization
# ============================================================

tfidf_vectorizer = TfidfVectorizer(
    max_features=5000,
    min_df=1,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X = tfidf_vectorizer.fit_transform(
    df["clean_text"]
)

print("TF-IDF completed.")

print("Number of documents:", X.shape[0])
print("Number of TF-IDF features:", X.shape[1])


# ============================================================
# CELL 12: Display TF-IDF vocabulary
# ============================================================

feature_names = (
    tfidf_vectorizer
    .get_feature_names_out()
)

print("Number of features:", len(feature_names))

print("\nFirst 50 features:")

print(
    feature_names[:50]
)


# ============================================================
# CELL 13: Convert TF-IDF to DataFrame
# ============================================================

# This is optional and can use significant memory
# for large datasets.

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=feature_names
)

print("TF-IDF DataFrame shape:")
print(tfidf_df.shape)

display(tfidf_df.head())


# ============================================================
# CELL 14: ELBOW METHOD
# ============================================================

# Determine maximum number of clusters

if len(df) < 3:
    raise ValueError(
        "The dataset needs at least 3 rows for clustering."
    )

max_k = min(10, len(df) - 1)

k_values = range(2, max_k + 1)

inertia_values = []

for k in k_values:
    
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    model.fit(X)
    
    inertia_values.append(
        model.inertia_
    )


# Plot elbow curve

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.xticks(list(k_values))

plt.grid(True)

plt.show()


# ============================================================
# CELL 15: SILHOUETTE SCORE
# ============================================================

silhouette_values = []

for k in k_values:
    
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = model.fit_predict(X)
    
    score = silhouette_score(
        X,
        labels
    )
    
    silhouette_values.append(score)


# Display scores

silhouette_results = pd.DataFrame({
    "K": list(k_values),
    "Silhouette Score": silhouette_values
})

display(silhouette_results)


# Plot silhouette scores

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.xticks(list(k_values))

plt.grid(True)

plt.show()


# ============================================================
# CELL 16: Select the best K
# ============================================================

best_k = list(k_values)[
    np.argmax(silhouette_values)
]

best_score = max(
    silhouette_values
)

print("Recommended K:", best_k)
print("Best silhouette score:", best_score)


# ============================================================
# CELL 17: OPTIONAL - Manually choose K
# ============================================================

# If you want to choose K yourself, uncomment this line.

# best_k = 4

print("Final K used for clustering:", best_k)


# ============================================================
# CELL 18: Train FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df["cluster"] = kmeans.fit_predict(X)

print("K-Means clustering completed.")


# ============================================================
# CELL 19: Display clustered dataset
# ============================================================

print("Clustered data:")

display(
    df[
        [
            text_column,
            "clean_text",
            "cluster"
        ]
    ].sort_values("cluster")
)


# ============================================================
# CELL 20: Number of records in each cluster
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("Number of records in each cluster:")

display(
    cluster_counts.to_frame(
        name="Number of Records"
    )
)


# ============================================================
# CELL 21: Percentage of records in each cluster
# ============================================================

cluster_percentage = (
    df["cluster"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

cluster_summary = pd.DataFrame({
    "Number of Records": cluster_counts,
    "Percentage": cluster_percentage.round(2)
})

print("Cluster summary:")

display(cluster_summary)


# ============================================================
# CELL 22: FIND TOP WORDS FOR EACH CLUSTER
# ============================================================

terms = (
    tfidf_vectorizer
    .get_feature_names_out()
)

# K-Means cluster centers
centroids = kmeans.cluster_centers_

# Sort words by importance
ordered_centroids = (
    centroids.argsort(axis=1)[:, ::-1]
)

print("Top words for each cluster:\n")

cluster_keywords = {}

for cluster_number in range(best_k):
    
    top_word_indexes = (
        ordered_centroids[
            cluster_number,
            :15
        ]
    )
    
    top_words = [
        terms[index]
        for index in top_word_indexes
    ]
    
    cluster_keywords[
        cluster_number
    ] = top_words
    
    print(
        f"Cluster {cluster_number}:"
    )
    
    print(
        ", ".join(top_words)
    )
    
    print()


# ============================================================
# CELL 23: CREATE CLUSTER KEYWORDS TABLE
# ============================================================

keyword_rows = []

for cluster_number in range(best_k):
    
    keyword_rows.append({
        "Cluster": cluster_number,
        "Top Keywords": ", ".join(
            cluster_keywords[
                cluster_number
            ]
        )
    })

keywords_df = pd.DataFrame(
    keyword_rows
)

display(keywords_df)


# ============================================================
# CELL 24: SHOW EXAMPLES FROM EACH CLUSTER
# ============================================================

for cluster_number in range(best_k):
    
    print("\n")
    print("=" * 70)
    print(f"CLUSTER {cluster_number}")
    print("=" * 70)
    
    cluster_data = df[
        df["cluster"] == cluster_number
    ]
    
    print(
        "Number of records:",
        len(cluster_data)
    )
    
    print(
        "Important words:"
    )
    
    print(
        ", ".join(
            cluster_keywords[
                cluster_number
            ][:10]
        )
    )
    
    print("\nExamples:")
    
    display(
        cluster_data[
            [text_column]
        ].head(10)
    )


# ============================================================
# CELL 25: REDUCE DIMENSIONS FOR VISUALIZATION
# ============================================================

svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = svd.fit_transform(X)

print(
    "Explained variance:",
    svd.explained_variance_ratio_.sum()
)


# ============================================================
# CELL 26: VISUALIZE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=df["cluster"],
    alpha=0.7,
    s=50
)

plt.xlabel("SVD Component 1")
plt.ylabel("SVD Component 2")

plt.title(
    "K-Means Clustering - NLP TF-IDF"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 27: VISUALIZE CLUSTER CENTERS
# ============================================================

# Transform cluster centers into 2D space

centers_2d = svd.transform(
    kmeans.cluster_centers_
)

plt.figure(figsize=(12, 8))

plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=df["cluster"],
    alpha=0.5,
    s=40
)

plt.scatter(
    centers_2d[:, 0],
    centers_2d[:, 1],
    marker="X",
    s=250,
    edgecolors="black",
    linewidths=2
)

# Add cluster labels

for cluster_number in range(best_k):
    
    x = centers_2d[
        cluster_number,
        0
    ]
    
    y = centers_2d[
        cluster_number,
        1
    ]
    
    plt.annotate(
        f"Cluster {cluster_number}",
        (x, y),
        xytext=(8, 8),
        textcoords="offset points",
        fontsize=12
    )

plt.xlabel("SVD Component 1")
plt.ylabel("SVD Component 2")

plt.title(
    "K-Means Clusters and Cluster Centers"
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 28: CREATE FINAL RESULTS DATAFRAME
# ============================================================

results = df[
    [
        text_column,
        "clean_text",
        "cluster"
    ]
].copy()

print("Final results:")
display(results.head(20))


# ============================================================
# CELL 29: Sort results by cluster
# ============================================================

results_sorted = results.sort_values(
    by="cluster"
)

display(
    results_sorted.head(50)
)


# ============================================================
# CELL 30: SAVE CLUSTERED DATA
# ============================================================

output_file = (
    "category_demo_clustered.csv"
)

results_sorted.to_csv(
    output_file,
    index=False
)

print(
    f"File saved successfully: {output_file}"
)


# ============================================================
# CELL 31: SAVE CLUSTER SUMMARY
# ============================================================

summary_file = (
    "category_demo_cluster_summary.csv"
)

cluster_summary.to_csv(
    summary_file
)

print(
    f"Cluster summary saved: {summary_file}"
)


# ============================================================
# CELL 32: SAVE CLUSTER KEYWORDS
# ============================================================

keywords_file = (
    "category_demo_cluster_keywords.csv"
)

keywords_df.to_csv(
    keywords_file,
    index=False
)

print(
    f"Cluster keywords saved: {keywords_file}"
)


# ============================================================
# CELL 33: FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("FINAL CLUSTERING SUMMARY")
print("=" * 70)

print(
    "Dataset:",
    file_path
)

print(
    "Text column:",
    text_column
)

print(
    "Number of records:",
    len(df)
)

print(
    "Number of TF-IDF features:",
    X.shape[1]
)

print(
    "Number of clusters:",
    best_k
)

print(
    "Silhouette score:",
    round(best_score, 4)
)

print("\nCluster sizes:")

for cluster_number in range(best_k):
    
    count = (
        df["cluster"]
        .value_counts()
        .get(cluster_number, 0)
    )
    
    print(
        f"Cluster {cluster_number}: {count} records"
    )

print("\nCluster keywords:")

for cluster_number in range(best_k):
    
    print(
        f"\nCluster {cluster_number}:"
    )
    
    print(
        ", ".join(
            cluster_keywords[
                cluster_number
            ][:10]
        )
    )

print("\nOutput files:")
print("1.", output_file)
print("2.", summary_file)
print("3.", keywords_file)

print("\nClustering completed successfully!")
```
