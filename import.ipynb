```python
# ============================================================
# K-MEANS NLP CLUSTERING - COMPLETE JUPYTER NOTEBOOK
# Dataset: import.csv
# ============================================================


# ============================================================
# CELL 1 - INSTALL REQUIRED PACKAGES
# ============================================================

# Run this cell if the packages are not already installed.

!pip install pandas numpy matplotlib scikit-learn nltk


# ============================================================
# CELL 2 - IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
import string
import warnings

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

warnings.filterwarnings("ignore")


# Download NLTK resources
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

print("Libraries imported successfully.")


# ============================================================
# CELL 3 - LOAD DATASET
# ============================================================

file_name = "import.csv"

df = pd.read_csv(file_name)

print("Dataset loaded successfully.")
print()
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print()
print("Column names:")
print(df.columns.tolist())


# ============================================================
# CELL 4 - DISPLAY DATASET
# ============================================================

print("First 10 rows:")

display(df.head(10))

print()
print("Dataset information:")

df.info()


# ============================================================
# CELL 5 - CHECK MISSING VALUES
# ============================================================

print("Missing values in each column:")
display(df.isnull().sum().sort_values(ascending=False))


# ============================================================
# CELL 6 - FIND TEXT COLUMN
# ============================================================

# Find columns containing text
text_columns = df.select_dtypes(include=["object"]).columns.tolist()

print("Text/object columns found:")
print(text_columns)

if len(text_columns) == 0:
    raise ValueError(
        "No text column was found in the CSV file."
    )

# Automatically use the first text column
text_column = text_columns[0]

print()
print("Text column selected:", text_column)

# If you know the exact column name, replace the above line with:
#
# text_column = "your_column_name"


# ============================================================
# CELL 7 - REMOVE EMPTY TEXT
# ============================================================

df = df.dropna(subset=[text_column]).copy()

df[text_column] = df[text_column].astype(str)

# Remove empty strings
df = df[df[text_column].str.strip() != ""]

# Reset index
df.reset_index(drop=True, inplace=True)

print("Rows remaining:", len(df))


# ============================================================
# CELL 8 - CREATE NLP CLEANING FUNCTION
# ============================================================

stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()


def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )

    # Remove email addresses
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # Remove HTML
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )

    # Remove numbers
    text = re.sub(
        r"\d+",
        " ",
        text
    )

    # Remove punctuation
    text = text.translate(
        str.maketrans(
            "",
            "",
            string.punctuation
        )
    )

    # Keep alphabetic words only
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    # Split into words
    words = text.split()

    # Remove stopwords and very short words
    words = [
        word
        for word in words
        if word not in stop_words
        and len(word) > 2
    ]

    # Lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]

    # Join words
    text = " ".join(words)

    return text


# ============================================================
# CELL 9 - APPLY NLP CLEANING
# ============================================================

df["clean_text"] = df[text_column].apply(clean_text)

print("Original text:")
display(df[[text_column]].head(10))

print()
print("Cleaned text:")
display(df[["clean_text"]].head(10))


# ============================================================
# CELL 10 - REMOVE EMPTY CLEANED TEXT
# ============================================================

df = df[
    df["clean_text"].str.strip() != ""
].copy()

df.reset_index(drop=True, inplace=True)

print("Rows after NLP cleaning:", len(df))


# ============================================================
# CELL 11 - TF-IDF VECTORIZATION
# ============================================================

tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

X = tfidf.fit_transform(
    df["clean_text"]
)

print("TF-IDF completed.")

print()
print("TF-IDF matrix shape:")
print(X.shape)


# ============================================================
# CELL 12 - DISPLAY TF-IDF FEATURES
# ============================================================

feature_names = tfidf.get_feature_names_out()

print("Number of TF-IDF features:", len(feature_names))

print()
print("First 50 features:")
print(feature_names[:50])


# ============================================================
# CELL 13 - FIND OPTIMAL NUMBER OF CLUSTERS
# ============================================================

# Test different values of K

max_k = min(10, len(df) - 1)

if max_k < 2:
    raise ValueError(
        "The dataset needs at least 3 text records for clustering."
    )

k_values = range(2, max_k + 1)

inertia_values = []
silhouette_values = []


for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    inertia_values.append(
        model.inertia_
    )

    silhouette_values.append(
        silhouette_score(X, labels)
    )


# ============================================================
# CELL 14 - ELBOW METHOD
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Selecting K")

plt.xticks(list(k_values))

plt.grid(True)

plt.show()


# ============================================================
# CELL 15 - SILHOUETTE SCORE
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.xticks(list(k_values))

plt.grid(True)

plt.show()


# ============================================================
# CELL 16 - DISPLAY K RESULTS
# ============================================================

results = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertia_values,
    "Silhouette Score": silhouette_values
})

display(results)


# ============================================================
# CELL 17 - AUTOMATICALLY SELECT BEST K
# ============================================================

best_k = results.loc[
    results["Silhouette Score"].idxmax(),
    "K"
]

best_k = int(best_k)

print("Best K according to silhouette score:", best_k)


# ============================================================
# CELL 18 - TRAIN FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X)

df["cluster"] = cluster_labels

print("K-Means clustering completed.")


# ============================================================
# CELL 19 - CLUSTER COUNTS
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("Number of records in each cluster:")

display(
    cluster_counts.to_frame(
        name="Number of Records"
    )
)


# ============================================================
# CELL 20 - CLUSTER PERCENTAGES
# ============================================================

cluster_percentage = (
    df["cluster"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

cluster_summary = pd.DataFrame({
    "Cluster": cluster_percentage.index,
    "Records": cluster_counts.values,
    "Percentage": cluster_percentage.values
})

cluster_summary["Percentage"] = (
    cluster_summary["Percentage"].round(2)
)

display(cluster_summary)


# ============================================================
# CELL 21 - FIND TOP WORDS FOR EACH CLUSTER
# ============================================================

terms = tfidf.get_feature_names_out()

cluster_keywords = {}


for cluster_number in range(best_k):

    # Get centroid for this cluster
    centroid = kmeans.cluster_centers_[
        cluster_number
    ]

    # Sort words by importance
    top_indices = centroid.argsort()[
        ::-1
    ][:20]

    top_words = [
        terms[index]
        for index in top_indices
    ]

    cluster_keywords[
        cluster_number
    ] = top_words


# ============================================================
# CELL 22 - DISPLAY TOP WORDS
# ============================================================

for cluster_number, words in cluster_keywords.items():

    print()
    print("=" * 70)
    print(f"CLUSTER {cluster_number}")
    print("=" * 70)

    print(", ".join(words))


# ============================================================
# CELL 23 - CREATE CLUSTER SUMMARY
# ============================================================

cluster_summary_text = []

for cluster_number in range(best_k):

    words = cluster_keywords[
        cluster_number
    ]

    count = (
        df["cluster"] == cluster_number
    ).sum()

    percentage = (
        count / len(df)
    ) * 100

    cluster_summary_text.append({
        "Cluster": cluster_number,
        "Records": count,
        "Percentage": round(percentage, 2),
        "Top Keywords": ", ".join(words[:10])
    })


cluster_summary_df = pd.DataFrame(
    cluster_summary_text
)

display(cluster_summary_df)


# ============================================================
# CELL 24 - SHOW EXAMPLES FROM EACH CLUSTER
# ============================================================

for cluster_number in range(best_k):

    print()
    print("=" * 80)
    print(f"EXAMPLES FROM CLUSTER {cluster_number}")
    print("=" * 80)

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    display(
        cluster_data[
            [text_column, "clean_text", "cluster"]
        ].head(10)
    )


# ============================================================
# CELL 25 - PCA REDUCTION
# ============================================================

# Convert TF-IDF data into 2 dimensions
# for visualisation.

pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(
    X.toarray()
)

print("PCA completed.")

print(
    "Explained variance:",
    pca.explained_variance_ratio_
)


# ============================================================
# CELL 26 - CREATE PCA DATAFRAME
# ============================================================

pca_df = pd.DataFrame({
    "PCA1": X_pca[:, 0],
    "PCA2": X_pca[:, 1],
    "Cluster": df["cluster"].values
})

display(pca_df.head())


# ============================================================
# CELL 27 - VISUALISE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(12, 8))

for cluster_number in range(best_k):

    points = pca_df[
        pca_df["Cluster"] == cluster_number
    ]

    plt.scatter(
        points["PCA1"],
        points["PCA2"],
        label=f"Cluster {cluster_number}",
        alpha=0.6
    )


plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")

plt.title(
    "K-Means Clustering of NLP Text Data"
)

plt.legend()

plt.grid(True)

plt.show()


# ============================================================
# CELL 28 - ADD CLUSTER NAMES
# ============================================================

# Automatically create cluster names from
# the top 3 keywords.

cluster_names = {}

for cluster_number in range(best_k):

    keywords = cluster_keywords[
        cluster_number
    ][:3]

    name = (
        "Cluster "
        + str(cluster_number)
        + " - "
        + ", ".join(keywords)
    )

    cluster_names[
        cluster_number
    ] = name


df["cluster_name"] = df[
    "cluster"
].map(cluster_names)


# ============================================================
# CELL 29 - DISPLAY FINAL DATA
# ============================================================

print("Final clustered dataset:")

display(df.head(20))


# ============================================================
# CELL 30 - DISPLAY CLUSTER NAMES
# ============================================================

print("Cluster names:")

for cluster_number, name in cluster_names.items():

    print(
        f"{cluster_number}: {name}"
    )


# ============================================================
# CELL 31 - SAVE CLUSTERED DATASET
# ============================================================

output_file = "import_clustered.csv"

df.to_csv(
    output_file,
    index=False
)

print()
print(
    f"Clustered dataset saved as: {output_file}"
)


# ============================================================
# CELL 32 - SAVE CLUSTER SUMMARY
# ============================================================

summary_file = "cluster_summary.csv"

cluster_summary_df.to_csv(
    summary_file,
    index=False
)

print(
    f"Cluster summary saved as: {summary_file}"
)


# ============================================================
# CELL 33 - SAVE TOP KEYWORDS
# ============================================================

keywords_rows = []

for cluster_number in range(best_k):

    for rank, word in enumerate(
        cluster_keywords[cluster_number],
        start=1
    ):

        keywords_rows.append({
            "Cluster": cluster_number,
            "Rank": rank,
            "Keyword": word
        })


keywords_df = pd.DataFrame(
    keywords_rows
)

keywords_file = "cluster_keywords.csv"

keywords_df.to_csv(
    keywords_file,
    index=False
)

print(
    f"Cluster keywords saved as: {keywords_file}"
)


# ============================================================
# CELL 34 - FINAL RESULTS
# ============================================================

print()
print("=" * 80)
print("FINAL RESULTS")
print("=" * 80)

print(
    f"Original rows: {pd.read_csv(file_name).shape[0]}"
)

print(
    f"Rows used for clustering: {len(df)}"
)

print(
    f"Text column: {text_column}"
)

print(
    f"Number of clusters: {best_k}"
)

print()

print("Cluster sizes:")

display(cluster_summary_df)

print()

print("Top keywords:")

display(keywords_df)


# ============================================================
# CELL 35 - OPTIONAL: SEARCH A SPECIFIC CLUSTER
# ============================================================

# Change the number below to inspect a cluster.

selected_cluster = 0

selected_data = df[
    df["cluster"] == selected_cluster
]

print(
    f"Records in cluster {selected_cluster}:"
)

display(
    selected_data[
        [text_column, "clean_text", "cluster_name"]
    ]
)


# ============================================================
# CELL 36 - OPTIONAL: EXPORT EACH CLUSTER
# ============================================================

for cluster_number in range(best_k):

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    filename = (
        f"cluster_{cluster_number}.csv"
    )

    cluster_data.to_csv(
        filename,
        index=False
    )

    print(
        f"Saved: {filename}"
    )

print()
print("All clustering files have been created.")
```
