````python

# 1. Import libraries
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# 2. Load dataset
df = pd.read_csv("shoes_data_small.csv")
print(df.head())
print(df.columns)

# 3. Select text column
# Change 'description' to your actual text column name
text = df["description"].fillna("").astype(str)

# 4. NLP text cleaning
def clean(x):
    x = x.lower()
    x = re.sub(r"[^a-z\s]", " ", x)
    return x

text = text.apply(clean)

# 5. Convert text to TF-IDF
vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(text)

# 6. K-Means clustering
k = 3
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["cluster"] = model.fit_predict(X)

# 7. Results
print("\nClustered Data:")
print(df[["description", "cluster"]].head(20))

# 8. Evaluate model
score = silhouette_score(X, df["cluster"])
print("\nSilhouette Score:", round(score, 3))

# 9. Show important words in each cluster
words = vectorizer.get_feature_names_out()

for i in range(k):
    top = model.cluster_centers_[i].argsort()[-10:][::-1]
    print(f"\nCluster {i}:",
          ", ".join(words[j] for j in top))
```

**Note:** If your dataset does not have a column called `description`, replace:

```python
df["description"]
```

with the actual text column name. You can find it with:

```python
print(df.columns)
```
````
