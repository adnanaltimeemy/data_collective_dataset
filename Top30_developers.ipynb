```python
# NLP + K-Means Clustering for Top30 Developers

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 1. Load dataset
df = pd.read_csv("Top30_developers.csv")
print(df.head())
print(df.columns)

# 2. Select text column
text_col = "Skills"          # Change if needed
text = df[text_col].fillna("").astype(str)

# 3. NLP: TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(text)

# 4. K-Means
k = 4
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# 5. Show clusters
print("\nClustered Developers:")
print(df[[text_col, "Cluster"]])

# 6. Important words in each cluster
words = tfidf.get_feature_names_out()

for i in range(k):
    center = model.cluster_centers_[i]
    top = center.argsort()[-5:][::-1]
    print(f"\nCluster {i}:",
          ", ".join(words[j] for j in top))

# 7. Visualize clusters
pca = PCA(n_components=2)
X_2D = pca.fit_transform(X.toarray())

plt.figure(figsize=(8,5))
plt.scatter(X_2D[:,0], X_2D[:,1], c=df["Cluster"], s=100)
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Top 30 Developers - K-Means Clusters")
plt.show()

# 8. Save result
df.to_csv("Top30_developers_clustered.csv", index=False)
print("\nSaved: Top30_developers_clustered.csv")
```
