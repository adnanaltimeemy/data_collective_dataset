```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Load dataset
df = pd.read_csv("currencies.csv")
print(df.head())
print(df.columns)

# Select text column
text_col = "currency"   # Change if your column has another name

# Clean text
df[text_col] = df[text_col].fillna("").astype(str).str.lower()

# NLP: TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(df[text_col])

# K-Means
k = 4
model = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

# Show results
print("\nClustered Data:")
print(df[[text_col, "Cluster"]])

# Cluster counts
print("\nCluster Counts:")
print(df["Cluster"].value_counts())

# Visualize
plt.scatter(X.toarray()[:, 0], X.toarray()[:, 1], c=df["Cluster"])
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Means Clustering")
plt.show()

# Save result
df.to_csv("currencies_clustered.csv", index=False)
print("\nSaved as currencies_clustered.csv")
```
