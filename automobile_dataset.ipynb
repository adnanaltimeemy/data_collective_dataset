```python
# ============================================================
# AUTOMOBILE DATASET - NLP + K-MEANS CLUSTERING
# JUPYTER NOTEBOOK
# ============================================================

# ------------------------------------------------------------
# CELL 1: Install required libraries
# ------------------------------------------------------------
# Run this cell if the libraries are not already installed.

!pip install pandas numpy matplotlib seaborn scikit-learn nltk


# ------------------------------------------------------------
# CELL 2: Import libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings

warnings.filterwarnings("ignore")

# Machine Learning
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

# NLP
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download NLP resources
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")


# ------------------------------------------------------------
# CELL 3: Load the dataset
# ------------------------------------------------------------

file_path = "automobile_dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()


# ------------------------------------------------------------
# CELL 4: Basic information
# ------------------------------------------------------------

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()


# ------------------------------------------------------------
# CELL 5: Statistical summary
# ------------------------------------------------------------

print("Statistical Summary:")
display(df.describe(include="all").T)


# ------------------------------------------------------------
# CELL 6: Check missing values
# ------------------------------------------------------------

missing_values = df.isnull().sum()

print("Missing values:")
print(missing_values[missing_values > 0])


# Visualize missing values
plt.figure(figsize=(12, 6))

missing_values[missing_values > 0].sort_values(ascending=False).plot(
    kind="bar"
)

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# CELL 7: Remove duplicate rows
# ------------------------------------------------------------

print("Duplicate rows before removal:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate rows after removal:", df.duplicated().sum())


# ------------------------------------------------------------
# CELL 8: Identify numerical and categorical columns
# ------------------------------------------------------------

numerical_columns = df.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical/Text columns:")
print(categorical_columns)


# ------------------------------------------------------------
# CELL 9: Fill missing values
# ------------------------------------------------------------

# Numerical columns -> median
for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())

# Categorical/text columns -> most frequent value
for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()

        if len(mode_value) > 0:
            df[col] = df[col].fillna(mode_value[0])
        else:
            df[col] = df[col].fillna("Unknown")

print("Missing values after cleaning:")
print(df.isnull().sum().sum())


# ------------------------------------------------------------
# CELL 10: Display cleaned dataset
# ------------------------------------------------------------

display(df.head())

print("Cleaned dataset shape:", df.shape)


# ============================================================
# NLP SECTION
# ============================================================

# ------------------------------------------------------------
# CELL 11: NLP preprocessing function
# ------------------------------------------------------------

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()


def clean_text(text):
    """
    Clean text using basic NLP preprocessing:
    1. Convert to lowercase
    2. Remove special characters
    3. Tokenize
    4. Remove stopwords
    5. Lemmatize words
    """

    text = str(text).lower()

    # Remove numbers and special characters
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenization
    words = text.split()

    # Stopword removal + lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)


# ------------------------------------------------------------
# CELL 12: Create combined text column
# ------------------------------------------------------------

# Convert all categorical columns into strings
if len(categorical_columns) > 0:

    df["combined_text"] = df[categorical_columns].astype(str).agg(
        " ".join, axis=1
    )

else:

    # If there are no text columns
    df["combined_text"] = "automobile"


print("Example of combined text:")
display(df["combined_text"].head())


# ------------------------------------------------------------
# CELL 13: Apply NLP cleaning
# ------------------------------------------------------------

df["clean_text"] = df["combined_text"].apply(clean_text)

print("Cleaned text:")
display(df[["combined_text", "clean_text"]].head())


# ------------------------------------------------------------
# CELL 14: TF-IDF Vectorization
# ------------------------------------------------------------

tfidf = TfidfVectorizer(
    max_features=500,
    min_df=1,
    max_df=0.95,
    ngram_range=(1, 2)
)

text_features = tfidf.fit_transform(df["clean_text"])

print("TF-IDF matrix shape:", text_features.shape)


# ------------------------------------------------------------
# CELL 15: Display TF-IDF vocabulary
# ------------------------------------------------------------

feature_names = tfidf.get_feature_names_out()

print("Number of TF-IDF features:", len(feature_names))

print("\nFirst 50 features:")
print(feature_names[:50])


# ============================================================
# NUMERICAL DATA PROCESSING
# ============================================================

# ------------------------------------------------------------
# CELL 16: Scale numerical features
# ------------------------------------------------------------

if len(numerical_columns) > 0:

    scaler = StandardScaler()

    numerical_features = scaler.fit_transform(
        df[numerical_columns]
    )

    print(
        "Numerical feature matrix shape:",
        numerical_features.shape
    )

else:

    numerical_features = np.empty((len(df), 0))

    print("No numerical columns found.")


# ------------------------------------------------------------
# CELL 17: Combine NLP + Numerical features
# ------------------------------------------------------------

from scipy.sparse import hstack, csr_matrix

# Convert numerical features to sparse matrix
numerical_sparse = csr_matrix(numerical_features)

# Combine TF-IDF and numerical features
combined_features = hstack(
    [text_features, numerical_sparse]
)

print(
    "Final feature matrix shape:",
    combined_features.shape
)


# ============================================================
# FIND OPTIMAL NUMBER OF CLUSTERS
# ============================================================

# ------------------------------------------------------------
# CELL 18: Elbow Method
# ------------------------------------------------------------

inertia = []

# Test K values from 2 to 10
k_values = range(2, 11)

for k in k_values:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(combined_features)

    inertia.append(kmeans.inertia_)


# Plot elbow curve
plt.figure(figsize=(10, 6))

plt.plot(
    k_values,
    inertia,
    marker="o"
)

plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.grid(True)
plt.show()


# ------------------------------------------------------------
# CELL 19: Silhouette Score
# ------------------------------------------------------------

silhouette_scores = []

for k in k_values:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(combined_features)

    score = silhouette_score(
        combined_features,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k}, Silhouette Score = {score:.4f}"
    )


# ------------------------------------------------------------
# CELL 20: Plot Silhouette Scores
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    k_values,
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score for Different K Values")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.grid(True)
plt.show()


# ------------------------------------------------------------
# CELL 21: Automatically select best K
# ------------------------------------------------------------

best_k = k_values[
    np.argmax(silhouette_scores)
]

print("Best number of clusters:", best_k)

print(
    "Best silhouette score:",
    max(silhouette_scores)
)


# ============================================================
# K-MEANS CLUSTERING
# ============================================================

# ------------------------------------------------------------
# CELL 22: Train final K-Means model
# ------------------------------------------------------------

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(
    combined_features
)

print("K-Means clustering completed!")


# ------------------------------------------------------------
# CELL 23: Display cluster counts
# ------------------------------------------------------------

cluster_counts = df["Cluster"].value_counts().sort_index()

print("Number of automobiles in each cluster:")
print(cluster_counts)


# ------------------------------------------------------------
# CELL 24: Cluster distribution graph
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.barplot(
    x=cluster_counts.index,
    y=cluster_counts.values
)

plt.title("Number of Automobiles in Each Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Automobiles")

plt.show()


# ============================================================
# PCA VISUALIZATION
# ============================================================

# ------------------------------------------------------------
# CELL 25: Reduce features to 2 dimensions
# ------------------------------------------------------------

# PCA works efficiently after converting the sparse matrix
# to a dense matrix for reasonably sized datasets.

if combined_features.shape[1] > 2:

    # Limit dimensions before PCA if dataset is very large
    if combined_features.shape[1] > 1000:

        from sklearn.decomposition import TruncatedSVD

        svd = TruncatedSVD(
            n_components=50,
            random_state=42
        )

        reduced_features = svd.fit_transform(
            combined_features
        )

        pca = PCA(
            n_components=2,
            random_state=42
        )

        pca_features = pca.fit_transform(
            reduced_features
        )

    else:

        dense_features = combined_features.toarray()

        pca = PCA(
            n_components=2,
            random_state=42
        )

        pca_features = pca.fit_transform(
            dense_features
        )

else:

    pca_features = combined_features.toarray()


# ------------------------------------------------------------
# CELL 26: PCA cluster visualization
# ------------------------------------------------------------

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    pca_features[:, 0],
    pca_features[:, 1],
    c=df["Cluster"],
    s=50,
    alpha=0.7
)

plt.title("K-Means Automobile Clusters")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)
plt.show()


# ============================================================
# CLUSTER ANALYSIS
# ============================================================

# ------------------------------------------------------------
# CELL 27: Cluster summary for numerical features
# ------------------------------------------------------------

if len(numerical_columns) > 0:

    cluster_summary = df.groupby(
        "Cluster"
    )[numerical_columns].mean()

    print("Average numerical values by cluster:")

    display(
        cluster_summary
    )


# ------------------------------------------------------------
# CELL 28: Median values by cluster
# ------------------------------------------------------------

if len(numerical_columns) > 0:

    cluster_median = df.groupby(
        "Cluster"
    )[numerical_columns].median()

    print("Median numerical values by cluster:")

    display(
        cluster_median
    )


# ------------------------------------------------------------
# CELL 29: Categorical cluster analysis
# ------------------------------------------------------------

for col in categorical_columns:

    print("\n======================================")
    print("Column:", col)
    print("======================================")

    result = pd.crosstab(
        df["Cluster"],
        df[col],
        normalize="index"
    ) * 100

    display(
        result.round(2)
    )


# ============================================================
# TOP WORDS IN EACH CLUSTER
# ============================================================

# ------------------------------------------------------------
# CELL 30: Find important NLP terms for clusters
# ------------------------------------------------------------

tfidf_array = text_features.toarray()

for cluster_number in sorted(df["Cluster"].unique()):

    # Get rows belonging to cluster
    cluster_indices = (
        df["Cluster"] == cluster_number
    ).values

    # Calculate average TF-IDF score
    mean_tfidf = tfidf_array[
        cluster_indices
    ].mean(axis=0)

    # Sort features by importance
    top_indices = mean_tfidf.argsort()[-15:][::-1]

    top_words = [
        feature_names[i]
        for i in top_indices
    ]

    print(
        f"\nCluster {cluster_number} - Top NLP Terms:"
    )

    print(", ".join(top_words))


# ============================================================
# VISUALIZE TOP TERMS
# ============================================================

# ------------------------------------------------------------
# CELL 31: Top terms for each cluster
# ------------------------------------------------------------

for cluster_number in sorted(df["Cluster"].unique()):

    cluster_indices = (
        df["Cluster"] == cluster_number
    ).values

    mean_tfidf = tfidf_array[
        cluster_indices
    ].mean(axis=0)

    top_indices = mean_tfidf.argsort()[-10:][::-1]

    words = [
        feature_names[i]
        for i in top_indices
    ]

    scores = [
        mean_tfidf[i]
        for i in top_indices
    ]

    plt.figure(figsize=(10, 5))

    plt.barh(
        words[::-1],
        scores[::-1]
    )

    plt.title(
        f"Top NLP Terms - Cluster {cluster_number}"
    )

    plt.xlabel("Average TF-IDF Score")
    plt.ylabel("Terms")

    plt.tight_layout()
    plt.show()


# ============================================================
# SAVE RESULTS
# ============================================================

# ------------------------------------------------------------
# CELL 32: Save clustered dataset
# ------------------------------------------------------------

output_file = "automobile_dataset_clustered.csv"

df.to_csv(
    output_file,
    index=False
)

print(
    f"Clustered dataset saved as: {output_file}"
)


# ------------------------------------------------------------
# CELL 33: Display final dataset
# ------------------------------------------------------------

display(df.head(20))


# ------------------------------------------------------------
# CELL 34: Final cluster report
# ------------------------------------------------------------

print("\n==========================================")
print("FINAL K-MEANS CLUSTERING REPORT")
print("==========================================")

print("Original/cleaned dataset shape:", df.shape)

print(
    "Number of clusters:",
    df["Cluster"].nunique()
)

print(
    "Best K:",
    best_k
)

print(
    "Silhouette Score:",
    round(max(silhouette_scores), 4)
)

print("\nCluster sizes:")

for cluster, count in cluster_counts.items():

    percentage = (
        count / len(df)
    ) * 100

    print(
        f"Cluster {cluster}: "
        f"{count} automobiles "
        f"({percentage:.2f}%)"
    )

print("\nClustering completed successfully!")
```
