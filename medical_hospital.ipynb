```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 2. Load dataset
df = pd.read_csv("medical_hospital.csv")

print(df.head())
print(df.columns)

# 3. Select text column
# Change "Description" to your actual text column name
text = df["Description"].fillna("").astype(str)

# 4. NLP - TF-IDF
tfidf = TfidfVectorizer(stop_words="english")
X = tfidf.fit_transform(text)

# 5. K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# 6. Show results
print(df[["Description", "Cluster"]].head(20))

# 7. Visualise clusters
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.toarray())

plt.scatter(X_pca[:,0], X_pca[:,1], c=df["Cluster"])
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("K-Means Clustering")
plt.show()

# 8. Save result
df.to_csv("medical_hospital_clustered.csv", index=False)
```
