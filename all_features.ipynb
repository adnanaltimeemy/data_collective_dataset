```python
# ============================================================
# K-MEANS + NLP CLUSTERING
# INPUT FILE: all_features.csv
# OUTPUT FILE: all_features_clustered.csv
# ============================================================


# ============================================================
# CELL 1 - INSTALL LIBRARIES
# ============================================================

# Run this cell if the libraries are not already installed.

# !pip install pandas numpy scipy scikit-learn matplotlib seaborn


# ============================================================
# CELL 2 - IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.sparse import csr_matrix, hstack

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics import silhouette_score

import warnings

warnings.filterwarnings("ignore")

print("Libraries imported successfully.")


# ============================================================
# CELL 3 - LOAD all_features.csv
# ============================================================

file_name = "all_features.csv"

df = pd.read_csv(file_name)

print("File loaded successfully!")
print("File name:", file_name)

print("\nDataset shape:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# CELL 4 - DATASET INFORMATION
# ============================================================

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
for column in df.columns:
    print("-", column)

print("\nData types:")
display(df.dtypes)


# ============================================================
# CELL 5 - CHECK MISSING VALUES
# ============================================================

print("Missing values in each column:")

missing_values = df.isnull().sum()

display(
    missing_values[
        missing_values > 0
    ].sort_values(ascending=False)
)


# ============================================================
# CELL 6 - IDENTIFY NUMERIC AND TEXT COLUMNS
# ============================================================

numeric_cols = df.select_dtypes(
    include=np.number
).columns.tolist()

text_cols = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Numeric columns:")
print(numeric_cols)

print("\nText columns:")
print(text_cols)


# ============================================================
# CELL 7 - CREATE COMBINED TEXT COLUMN
# ============================================================

if len(text_cols) > 0:

    df["combined_text"] = (
        df[text_cols]
        .fillna("")
        .astype(str)
        .agg(" ".join, axis=1)
    )

else:

    df["combined_text"] = ""


print("Combined text created.")

print("\nExample text:")
display(
    df[["combined_text"]].head()
)


# ============================================================
# CELL 8 - CLEAN TEXT
# ============================================================

df["combined_text"] = (
    df["combined_text"]
    .str.lower()
    .str.replace(r"[^a-zA-Z0-9\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("Text cleaning completed.")

display(
    df[["combined_text"]].head()
)


# ============================================================
# CELL 9 - NLP USING TF-IDF
# ============================================================

if df["combined_text"].str.strip().ne("").any():

    tfidf = TfidfVectorizer(
        stop_words="english",
        max_features=5000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95
    )

    X_text = tfidf.fit_transform(
        df["combined_text"]
    )

    print("TF-IDF completed.")

    print(
        "Number of documents:",
        X_text.shape[0]
    )

    print(
        "Number of TF-IDF features:",
        X_text.shape[1]
    )

else:

    X_text = None

    print(
        "No usable text columns were found."
    )


# ============================================================
# CELL 10 - PROCESS NUMERIC FEATURES
# ============================================================

if len(numeric_cols) > 0:

    numeric_data = df[numeric_cols].copy()

    # Replace infinity values
    numeric_data = numeric_data.replace(
        [np.inf, -np.inf],
        np.nan
    )

    # Fill missing numeric values
    numeric_data = numeric_data.fillna(
        numeric_data.median()
    )

    # Standardize numerical features
    scaler = StandardScaler()

    X_numeric = scaler.fit_transform(
        numeric_data
    )

    print("Numeric preprocessing completed.")

    print(
        "Numeric feature shape:",
        X_numeric.shape
    )

else:

    X_numeric = None

    print(
        "No numeric columns found."
    )


# ============================================================
# CELL 11 - COMBINE NLP + NUMERIC FEATURES
# ============================================================

if X_text is not None and X_numeric is not None:

    X_numeric_sparse = csr_matrix(
        X_numeric
    )

    X = hstack(
        [
            X_text,
            X_numeric_sparse
        ]
    )

elif X_text is not None:

    X = X_text

elif X_numeric is not None:

    X = csr_matrix(
        X_numeric
    )

else:

    raise ValueError(
        "No usable features were found."
    )


print("Final feature matrix created.")

print(
    "Final feature matrix shape:",
    X.shape
)


# ============================================================
# CELL 12 - ELBOW METHOD
# ============================================================

k_values = range(2, 11)

inertia_values = []

print("Running K-Means for different K values...")

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)

    inertia_values.append(
        model.inertia_
    )

    print(
        "K =", k,
        "| Inertia =", round(
            model.inertia_,
            2
        )
    )


# ============================================================
# CELL 13 - PLOT ELBOW CURVE
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.title(
    "Elbow Method for K-Means"
)

plt.xticks(
    list(k_values)
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 14 - SILHOUETTE SCORE
# ============================================================

silhouette_values = []

print("Calculating silhouette scores...")

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    # Use a sample for very large datasets
    if X.shape[0] > 10000:

        random_state = np.random.RandomState(42)

        sample_indices = random_state.choice(
            X.shape[0],
            10000,
            replace=False
        )

        score = silhouette_score(
            X[sample_indices],
            labels[sample_indices]
        )

    else:

        score = silhouette_score(
            X,
            labels
        )

    silhouette_values.append(
        score
    )

    print(
        "K =", k,
        "| Silhouette Score =",
        round(score, 4)
    )


# ============================================================
# CELL 15 - PLOT SILHOUETTE SCORES
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "Silhouette Score for K-Means"
)

plt.xticks(
    list(k_values)
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 16 - SELECT BEST K
# ============================================================

best_k = list(k_values)[
    np.argmax(
        silhouette_values
    )
]

print(
    "Best K according to Silhouette Score:",
    best_k
)


# ============================================================
# CELL 17 - RUN FINAL K-MEANS
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=20
)

cluster_labels = kmeans.fit_predict(
    X
)

df["cluster"] = cluster_labels

print("K-Means clustering completed.")


# ============================================================
# CELL 18 - SHOW CLUSTER COUNTS
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("Number of records in each cluster:")

display(
    cluster_counts.to_frame(
        name="Number of Records"
    )
)


# ============================================================
# CELL 19 - CLUSTER DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    x="cluster",
    data=df
)

plt.xlabel(
    "Cluster"
)

plt.ylabel(
    "Number of Records"
)

plt.title(
    "Number of Records in Each Cluster"
)

plt.show()


# ============================================================
# CELL 20 - REDUCE DIMENSIONS FOR VISUALIZATION
# ============================================================

print(
    "Reducing features to 2 dimensions..."
)

svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = svd.fit_transform(
    X
)

print(
    "2D feature shape:",
    X_2d.shape
)


# ============================================================
# CELL 21 - VISUALIZE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=df["cluster"],
    cmap="viridis",
    alpha=0.7
)

plt.xlabel(
    "SVD Component 1"
)

plt.ylabel(
    "SVD Component 2"
)

plt.title(
    "K-Means Clustering Visualization"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 22 - SHOW SAMPLE RECORDS FROM EACH CLUSTER
# ============================================================

print(
    "Sample records from each cluster:"
)

for cluster_id in sorted(
    df["cluster"].unique()
):

    print("\n")
    print("=" * 80)
    print(
        "CLUSTER",
        cluster_id
    )
    print("=" * 80)

    cluster_data = df[
        df["cluster"] == cluster_id
    ]

    display(
        cluster_data.head(5)
    )


# ============================================================
# CELL 23 - TOP WORDS FOR EACH CLUSTER
# ============================================================

if X_text is not None:

    feature_names = np.array(
        tfidf.get_feature_names_out()
    )

    print(
        "Top NLP words for each cluster:"
    )

    for cluster_id in sorted(
        df["cluster"].unique()
    ):

        cluster_indices = (
            df["cluster"].values
            == cluster_id
        )

        cluster_text = X_text[
            cluster_indices
        ]

        mean_tfidf = np.asarray(
            cluster_text.mean(axis=0)
        ).flatten()

        top_indices = (
            mean_tfidf
            .argsort()[-20:][::-1]
        )

        top_words = (
            feature_names[top_indices]
        )

        print("\nCluster", cluster_id)

        print(
            ", ".join(top_words)
        )


# ============================================================
# CELL 24 - CREATE CLUSTER SUMMARY
# ============================================================

cluster_summary = (
    df.groupby("cluster")
      .size()
      .reset_index(
          name="number_of_records"
      )
)

print(
    "Cluster Summary:"
)

display(
    cluster_summary
)


# ============================================================
# CELL 25 - SAVE CLUSTERED DATASET
# ============================================================

output_file = (
    "all_features_clustered.csv"
)

df.to_csv(
    output_file,
    index=False
)

print(
    "Dataset saved successfully!"
)

print(
    "Output file:",
    output_file
)


# ============================================================
# CELL 26 - SAVE CLUSTER RESULTS
# ============================================================

cluster_results = pd.DataFrame()

cluster_results["cluster"] = (
    df["cluster"]
)

cluster_results["text"] = (
    df["combined_text"]
)

cluster_results.to_csv(
    "cluster_results.csv",
    index=False
)

print(
    "Cluster results saved as:",
    "cluster_results.csv"
)


# ============================================================
# CELL 27 - FINAL RESULTS
# ============================================================

print("\n")
print("=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print(
    "Input file:",
    "all_features.csv"
)

print(
    "Number of records:",
    len(df)
)

print(
    "Number of original features:",
    len(df.columns) - 2
)

print(
    "Number of clusters:",
    best_k
)

print(
    "Output file:",
    "all_features_clustered.csv"
)

print(
    "Additional output:",
    "cluster_results.csv"
)

print("\nCluster sizes:")

display(
    df["cluster"]
    .value_counts()
    .sort_index()
    .to_frame(
        name="Records"
    )
)

print("\nDone!")
```
